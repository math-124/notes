# 5.2: Bases and Dimension

Last time, we've discussed the concept of _subspaces_ which makes precise the notion of 'flat objects through the origin'. An important source of examples of subspaces are supplied by the _column space_ of a matrix $A$, which is the subspace spanned by its column vectors.

We've seen through several examples that subspaces can come in different shapes: it can simply the origin (so one point), and sometimes it can be a line or a plane. One way to distinguish these shapes is to quantify them by their _dimension_. We intuitively say that lines are one-dimensional and planes are two-dimensional; our goal today is to make this intuition precise.

## Spanning sets and linear independence

Suppose we are given a subspace $V$ of $\mathbb{R}^m$ and a collection of vectors $S= \left\{ \vec v_1,\ldots, \vec v_n\right\}$ inside $V$. We say that $S$ is a **spanning set** of $V$, or that **$S$ spans $V$**, if the span of $S$ contains all vectors in $V$.

Let us unpack this definition a bit more. The span of $S$ is simply all linear combinations of $\vec v_1,\ldots, \vec v_n$. Hence $S$ is a spanning set if every vector in $V$ can be written as a linear combination of $\vec v_1,\ldots, \vec v_n$.

Let us look at some examples to have a more solid grasp of this example.

### Example 0: Column space

By definition, the column vectors of a matrix $A$ is a spanning set of the column space of $A$.

### Example 1: A line in $\mathbb{R}^2$

Consider the subspace

$$
V=\left\{\begin{bmatrix}x\\y\end{bmatrix}\in\mathbb{R}^2:y=2x\right\}.
$$

The set $S=\left\{\begin{bmatrix}1\\2\end{bmatrix}\right\}$ spans $V$. Indeed, every vector in $V$ has the form

$$
\begin{bmatrix}x\\y\end{bmatrix} = \begin{bmatrix}x\\2x\end{bmatrix} =x\begin{bmatrix}1\\2\end{bmatrix}.
$$

Geometrically, $V$ is a line through the origin. Another possible spanning set is
$$ S' = \left\{\begin{bmatrix}-2\\-4\end{bmatrix}\right\}.$$

### Example 2: A redundant spanning set for $\mathbb{R}^2$

Let

$$
S=\left\{\begin{bmatrix}1\\0\end{bmatrix},\begin{bmatrix}0\\1\end{bmatrix},\begin{bmatrix}1\\1\end{bmatrix}\right\}.
$$

This set spans all of $\mathbb{R}^2$, since every vector satisfies

$$
\begin{bmatrix}a\\b\end{bmatrix}=a\begin{bmatrix}1\\0\end{bmatrix}+b\begin{bmatrix}0\\1\end{bmatrix}.
$$

Note that the third vector is not needed at all; this shows that a spanning set may contain redundant vectors. We will elaborate on this redundancy soon.

### Example 3: A plane in $\mathbb{R}^3$

Consider the following:
$$
W=\left\{\begin{bmatrix}x\\y\\z\end{bmatrix}\in\mathbb{R}^3:z=0\right\}.
$$
Verify that $W$ is a subspace by checking that it is closed under addition and scalar multiplication.

The set
$$
S=\left\{\begin{bmatrix}1\\0\\0\end{bmatrix},\begin{bmatrix}0\\1\\0 \end{bmatrix}\right\}
$$
spans $W$. To see this, note that
$$
\begin{bmatrix}x\\y\\0\end{bmatrix}=x\begin{bmatrix}1\\0\\0\end{bmatrix} + y\begin{bmatrix}0\\1\\0\end{bmatrix}.
$$
Hence $S$ is a spanning set of $W$. Geometrically, $W$ is the $xy$-plane that goes through the origin.

The examples above give us a very important idea: the size of a spanning set might be a good candidate for the 'dimension' of a subspace. Indeed in the first example, we see that a line -- which we think of as 1-dimensional -- has a spanning set of one element. In the third example, the $xy$-plane -- which we think of as 2-dimensional -- has a spanning set of two elements.

The main issue with this idea is the possible redundancy in the spanning set. In the first example, we see that the spanning set
$$
S=\left\{\begin{bmatrix}1\\0\end{bmatrix},\begin{bmatrix}0\\1\end{bmatrix},\begin{bmatrix}1\\1\end{bmatrix}\right\}.
$$
of size 3 spans the whole of $\mathbb{R}^2$, which is 2-dimensional. But we could've chosen a smaller spanning set
$$ S' = \left\{\begin{bmatrix}1\\0\end{bmatrix},\begin{bmatrix}0\\1\end{bmatrix}\right\}$$
of $\mathbb{R}^2$ instead.

Let's now make the concept of redundancy precise: We say that a collection of vectors $S$ is **linearly dependent** if we can choose a smaller set $S'$ contained in $S$ that still spans the same subspace as $S$. Otherwise, we say that $S$ is **linearly independent**.

It is clear that $S$ in Example 1 is linearly independent, as any smaller $S'$ contained in $S$ must be empty. For Example 3, suppose we choose a smaller
$$ S' = \left\{\begin{bmatrix}1\\0\\0\end{bmatrix}\right\} $$
contained in $S$. This is no longer a spanning set of the subspace $W$ since vectors like
$$ \begin{bmatrix}0\\1\\0\end{bmatrix},\ \begin{bmatrix}1\\2\\0\end{bmatrix},\ \ldots $$
are no longer spanned by $S'$. The same goes for the other smaller choice
$$ S' = \left\{\begin{bmatrix}0\\1\\0\end{bmatrix}\right\}.$$
Hence we see that $S$ is linearly independent.

How about Example 2? We are given a linearly dependent $S$, and one can very similarly check that the smaller set
$$ S' = \left\{\begin{bmatrix}1\\0\end{bmatrix},\begin{bmatrix}0\\1\end{bmatrix}\right\}$$
of size 2 -- which is the anticipated dimension of $\mathbb{R}^2$ -- is linearly independent.

## Bases and dimension

A **basis** of a subspace is a linearly independent spanning set. The spanning sets $S$ given in Example 1 and 3 are both linearly independent and hence are bases of the respective subspaces. The $S$ given in Example 2 is linearly dependent, but by reducing to a smaller $S'$ we obtained a linearly independent spanning set and hence a basis.

We now see from all three examples that the size of a basis precisely matches what we intuitively think of as dimension. Hence we can now rigorously define the **dimension** of a subspace simply as the size of its basis.

One has to be careful with this definition since one subspace can have many different bases. For instance, both
$$ S' = \left\{\begin{bmatrix}1\\0\end{bmatrix},\begin{bmatrix}0\\1\end{bmatrix}\right\}$$
and
$$ S'' = \left\{\begin{bmatrix}-2\\0\end{bmatrix},\begin{bmatrix}1\\1\end{bmatrix}\right\}$$
are bases of $\mathbb{R}^2$. We have also seen two different examples of bases in Example 1. 

Nevertheless, it is known that **every basis of a subspace has the same size**, so we can rest assured that the concept of dimension is not ill-defined.

Another fact (that we won't prove in this course) is that one can start with a spanning set $S$ of a subspace $V$ and remove the redundant vectors one-by-one to obtain a basis $S'$ of $V$, just like we did for Example 3.

## Orthonormal basis

In this course, we will focus on a special class of bases called **orthonormal bases**. These are bases where each of the vectors have norm 1 and are orthogonal to each other. Rephrasing this in terms of the dot product, an orthonormal basis is a basis
$$ S = \left\{ v_1, v_2, \dots, v_n \right\}$$
where
$$ v_i \cdot v_i = 1 $$
for all $i$ and
$$ v_i \cdot v_j = 0$$
for all distinct $i$ and $j$.

We have already seen an example of an orthonormal basis: the _standard basis vectors_ $\vec e_1, \vec e_2 ,\ldots, \vec e_n$ form an orthonormal basis of $\mathbb{R}^n$ -- which we now call the **standard basis** -- since we have $\vec e_i\cdot \vec e_i = 1 $ for every $i$, while $\vec e_i \cdot \vec e_j = 0$ for all distinct $i,j$. More concretely, check that
$$ \vec e_1 = \begin{bmatrix} 1 \\ 0 \end{bmatrix},\ \vec e_2 = \begin{bmatrix} 0 \\ 1 \end{bmatrix}$$
forms an orthonormal basis of $\mathbb{R}^2$.

The standard basis isn't the only orthonormal basis though. There are other examples, such as
$$ \vec v_1 = \frac{1}{\sqrt{2}}\begin{bmatrix} 1 \\ 1 \end{bmatrix},\ \vec v_2 = \frac{1}{\sqrt{2}} \begin{bmatrix} -1 \\ 1 \end{bmatrix}; \tag{$\dag$}$$
or
$$ \vec w_1 = \frac{1}{\sqrt{5}}\begin{bmatrix} 2 \\ 1 \end{bmatrix},\ \vec w_2 = \frac{1}{\sqrt{5}} \begin{bmatrix} -1 \\ 2 \end{bmatrix}; \tag{$\ddag$}$$

It's worth noting that these bases aren't random; they can be obtained by applying a rotation map to the standard basis $\vec e_1, \vec e_2$. Note that rotations preserve lengths and angles, so orthogonality and unit length of the basis vectors are preserved under rotations.

The reason why we focus on orthonormal bases is because they provide a very nice _coordinate system_ for the subspace. Suppose a subspace $V$ is given an orthonormal basis $S = \left\{ \vec v_1, \vec v_2, \ldots, \vec v_n \right\}$. Given a vector $\vec w$ in $V$, we wish to write $\vec w$ as a linear combination of $\vec v_1, \vec v_2, \ldots, \vec v_n$, or in other words _decompose_ $\vec w$ in terms of the basis $S$. Note that this is possible since $S$ spans $V$, so every vector in $V$ can be written as a linear combination of the basis vectors. 

In other words, we want to write
$$ \vec w = a_1 \vec v_1 + a_2 \vec v_2 + \cdots + a_n \vec v_n \tag{$*$}$$
for some scalars $a_1,a_2,\ldots,a_n$ whose values we need to find. There is an easy, clever trick for finding those values for orthonormal bases: taking the dot product of each side of ($*$) with $\vec v_i$, we obtain
$$ \vec w\cdot \vec v_i = a_1(\vec v_1 \cdot \vec v_i) + a_2(\vec v_2 \cdot \vec v_i) + \cdots + a_n(\vec v_n\cdot \vec v_i) = a_i,$$
since all of those $\vec v_j \cdot \vec v_i$ cancel out except for the case when $j=i$. Hence we simply find 
$$ a_i = \vec w \cdot \vec v_i.$$

This is most evidently demonstrated by the standard basis. Suppose we are given the vector
$$\vec w = \begin{bmatrix} 3 \\ 2\end{bmatrix} $$
and want to write it as a sum of the standard basis vectors. Since
$$ \vec w \cdot \vec e_1 = 3, \ \vec w \cdot \vec e_2 = 2,$$
we obtain
$$ \vec w = 3\vec e_1 + 2\vec e_2,$$
as we have seen in earlier chapters.

The power of the observation above is that it works for _any_ orthonormal basis. For instance, suppose we are given the same $\vec w$ and want to decompose it in terms of the basis $\vec v_1, \vec v_2$ given in ($\dag$). Since we have
$$ \vec w \cdot \vec v_1 = \begin{bmatrix} 3 \\ 2\end{bmatrix}\cdot \frac{1}{\sqrt{2}}\begin{bmatrix} 1 \\ 1 \end{bmatrix} = \frac{5}{\sqrt{2}};$$
$$ \vec w \cdot \vec v_2 = \begin{bmatrix} 3 \\ 2\end{bmatrix}\cdot \frac{1}{\sqrt{2}}\begin{bmatrix} -1 \\ 1 \end{bmatrix} = -\frac{1}{\sqrt{2}};$$
we obtain the decomposition
$$ \vec w = \frac{5}{\sqrt{2}}\vec v_1 - \frac{1}{\sqrt{2}}\vec v_2.$$
Note that for the same vector $\vec w$, we get different decompositions in terms of different bases. This is often described as _changing coordinate systems_.

It is not immediately clear why it is useful to have different orthonormal bases and their respective coordinate systems. In future chapters, we will learn that changing coordinates is immensely useful since matrices that look complicated in one coordinate system can be made very simple in another coordinate system, through a process called _diagonalization_.
