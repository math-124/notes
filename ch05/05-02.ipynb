# 5.2: Linear independence and invertibility

The column space tells us **which outputs a matrix can produce**. Now, let's ask a second question: when an output is possible, **how many inputs produce it?** This will help us understand when we can undo matrix-vector multiplication, even for matrices larger than $2\times2$.


---

## Counting inputs

Let's return to the input and output in the animation in [Chapter 5.1](05-01.ipynb), and compare the three matrices:

$$A=\begin{bmatrix}1&0\\1&1\\0&1\end{bmatrix},\qquad
B=\begin{bmatrix}1&0&2\\1&1&1\\0&1&-1\end{bmatrix},\qquad
C=\begin{bmatrix}1&0&2\\1&1&4\\0&1&3\end{bmatrix}.$$

All three can produce the output $\vec b=\begin{bmatrix}4\\9\\5\end{bmatrix}$. But knowing the output doesn't always tell us what the input was.

For $A$, the input has to be $\begin{bmatrix}4\\5\end{bmatrix}$: its two entries become the first and third entries of the output. So

$$A\begin{bmatrix}4\\5\end{bmatrix}=\begin{bmatrix}4\\9\\5\end{bmatrix}$$

is the only way to produce $\vec b$.

For $B$, there is more than one way:

$$B\begin{bmatrix}4\\5\\0\end{bmatrix}
=B\begin{bmatrix}2\\6\\1\end{bmatrix}
=\begin{bmatrix}4\\9\\5\end{bmatrix}.$$

In fact, there are infinitely many inputs of the form $\begin{bmatrix}4-2t\\5+t\\t\end{bmatrix}$. Knowing the output doesn't let us recover the input.

For $C$, the only input is $\begin{bmatrix}4\\5\\0\end{bmatrix}$. To see why, write out the equations for $C\vec v=\vec b$:

$$v_1+2v_3=4,\qquad v_1+v_2+4v_3=9,\qquad v_2+3v_3=5.$$

The first and third equations give $v_1=4-2v_3$ and $v_2=5-3v_3$. Substituting into the second gives $9-v_3=9$, so $v_3=0$. Then $v_1=4$ and $v_2=5$.

What is different about the columns of these matrices?


---

## Linear independence

Think of a collection of vectors as building blocks. If one of them can already be made using the others, it is redundant: removing it doesn't change the span.

:::{note} Definition: Linear independence
A collection of vectors $\vec v_1,\vec v_2,\ldots,\vec v_d$ is **linearly independent** if none of the vectors can be written as a linear combination of the other vectors in the collection.

If at least one vector can be written as a linear combination of the others, the collection is **linearly dependent**.
:::

### Dependent columns

For the matrix

$$B=\begin{bmatrix}1&0&2\\1&1&1\\0&1&-1\end{bmatrix},$$

call the columns $\vec a_1,\vec a_2,\vec a_3$. They satisfy

$$\vec a_3=2\vec a_1-\vec a_2.$$

So the three columns are linearly dependent. The third column doesn't add anything to the span of the first two.

This relationship is also why different inputs can produce the same output. Adding one copy of $\vec a_3$ is the same as adding two copies of $\vec a_1$ and subtracting one copy of $\vec a_2$. We can compensate by changing the other coefficients:

$$B\begin{bmatrix}p-2t\\q+t\\r+t\end{bmatrix}
=(p-2t)\vec a_1+(q+t)\vec a_2+(r+t)(2\vec a_1-\vec a_2)
=B\begin{bmatrix}p\\q\\r\end{bmatrix}.$$

This works for **any** starting input and **any** real number $t$. Every output in $\operatorname{colsp}(B)$ has infinitely many inputs.

We could remove the third column, but we could also remove the second, since $\vec a_2=2\vec a_1-\vec a_3$. There may be several choices of which vector to remove from a dependent collection.

:::{attention} Checking pairs is not enough
For two nonzero vectors, dependence means one is a scalar multiple of the other. For three or more vectors, a relationship can involve several vectors at once. No two columns of $B$ are scalar multiples, but the three columns are still dependent!
:::

### Independent columns

The columns of

$$A=\begin{bmatrix}1&0\\1&1\\0&1\end{bmatrix}$$

are independent: neither is a scalar multiple of the other. Each vector on their plane has just one representation using these two columns.

Now consider

$$C=\begin{bmatrix}1&0&2\\1&1&4\\0&1&3\end{bmatrix}.$$

The first two columns are the same as $A$'s. The third column is outside their plane, since $2-4+3\ne0$. It cannot be made using the first two. These three columns are independent, and every vector in their span has just one representation using them.

:::{important} Column space, independence, and solutions
For an $n\times d$ matrix $A$:

- $A\vec v=\vec b$ has a solution exactly when $\vec b\in\operatorname{colsp}(A)$.
- If the columns are independent, each $\vec b\in\operatorname{colsp}(A)$ has exactly one input.
- If the columns are dependent, each $\vec b\in\operatorname{colsp}(A)$ has infinitely many inputs.

An output outside the column space has no input in either case.
:::

## Invertibility

Recall that an inverse **undoes** multiplication by a matrix. For an $n\times n$ matrix $A$, its inverse $A^{-1}$ satisfies

$$A^{-1}A=AA^{-1}=I_n.$$

If we know the output $\vec b$, we can recover the input by computing $\vec v=A^{-1}\vec b$. So an invertible matrix must produce every vector in $\mathbb R^n$, and each output must come from exactly one input.

The following fact connects this to the columns. We'll use it without proof.

:::{important} Invertibility and columns
For an $n\times n$ matrix $A$, the following statements are equivalent:

1. $A$ is invertible.
2. For every $\vec b\in\mathbb R^n$, $A\vec v=\vec b$ has exactly one solution.
3. The columns of $A$ are linearly independent.
4. $\operatorname{colsp}(A)=\mathbb R^n$.
:::

We don't need a new inverse formula for each matrix size. We can often decide whether a matrix is invertible by looking at its columns!

For our three matrices:

| Matrix | Column space | Columns independent? | Invertible? |
| --- | --- | --- | --- |
| $A$ | The plane $x-y+z=0$ | Yes | No: $A$ is not square |
| $B$ | The plane $x-y+z=0$ | No | No |
| $C$ | All of $\mathbb R^3$ | Yes | Yes |

The distinction between $A$ and $C$ matters. Both have independent columns, and both produce $\begin{bmatrix}4\\9\\5\end{bmatrix}$ from a unique input. But $A$ only produces vectors on a plane, while $C$ produces every vector in $\mathbb R^3$. The equivalences above apply to **square matrices**.

$$H=\begin{bmatrix}0&2&0&0\\-3&0&0&0\\0&0&0&4\\0&0&5&0\end{bmatrix},\qquad
J=\begin{bmatrix}2&0&1&0\\0&-3&4&0\\0&0&0&0\\0&0&0&5\end{bmatrix}.$$

:::{tip} Solution
:class: dropdown

**$H$ is invertible.** Its columns point along four different coordinate directions, so none can be made using the others. More explicitly,

$$H\begin{bmatrix}x_1\\x_2\\x_3\\x_4\end{bmatrix}
=\begin{bmatrix}2x_2\\-3x_1\\4x_4\\5x_3\end{bmatrix}.$$

Every output $\begin{bmatrix}b_1\\b_2\\b_3\\b_4\end{bmatrix}$ has exactly one input:
$\begin{bmatrix}-b_2/3\\b_1/2\\b_4/5\\b_3/4\end{bmatrix}$.

**$J$ is not invertible.** Every output has third entry zero, so it cannot produce every vector in $\mathbb R^4$. Its third column is also $\frac12$ times its first minus $\frac43$ times its second, so the columns are dependent.

## More examples

A column relationship can save us a lot of work. For instance, a zero column is always redundant: adding any multiple of $\vec0$ doesn't change a linear combination. Changing the corresponding input entry therefore has no effect on the output.

### Three column patterns

Consider these three matrices. What can their columns tell us about invertibility?

$$D=\begin{bmatrix}2&0&4\\1&3&-1\\0&-1&1\end{bmatrix},\qquad
E=\begin{bmatrix}3&-2&1\\0&-1&4\\0&0&2\end{bmatrix},\qquad
F=\begin{bmatrix}1&0&2\\0&0&3\\2&0&-1\end{bmatrix}.$$

**$D$ is not invertible.** Its columns satisfy $\vec a_3=2\vec a_1-\vec a_2$. For example,

$$D\begin{bmatrix}2\\-1\\0\end{bmatrix}
=D\begin{bmatrix}0\\0\\1\end{bmatrix}
=\begin{bmatrix}4\\-1\\1\end{bmatrix}.$$

**$E$ is invertible.** For any output $\begin{bmatrix}b_1\\b_2\\b_3\end{bmatrix}$, the equations are

$$3v_1-2v_2+v_3=b_1,\qquad -v_2+4v_3=b_2,\qquad 2v_3=b_3.$$

Starting with the last equation, we can determine $v_3$, then $v_2$, then $v_1$. Each is uniquely determined, whatever $b_1,b_2,b_3$ are. This is back substitution; there is no need to compute an inverse.

**$F$ is not invertible.** Its second column is zero. For example,

$$F\begin{bmatrix}1\\0\\0\end{bmatrix}
=F\begin{bmatrix}1\\5\\0\end{bmatrix}
=\begin{bmatrix}1\\0\\2\end{bmatrix}.$$

The same reasoning works with more entries.

::::{tip} Activity 3

For which real values of $t$ is the following matrix invertible? Explain what happens to the inputs and outputs when it is not invertible.

$$D_t=\begin{bmatrix}2&0&0&0\\0&-3&0&0\\0&0&4&0\\0&0&0&t\end{bmatrix}.$$

:::{tip} Solution
:class: dropdown

Exactly when $t\ne0$. For any output $\begin{bmatrix}b_1\\b_2\\b_3\\b_4\end{bmatrix}$, the unique input is

$$\begin{bmatrix}b_1/2\\-b_2/3\\b_3/4\\b_4/t\end{bmatrix}.$$

When $t=0$, the fourth column is zero. Every output has fourth entry zero, so outputs with a nonzero fourth entry are impossible. For an output with fourth entry zero, the first three input entries are determined, but the fourth can be anything.
:::
::::

## Changing a column

::::{tip} Activity 4

Let

$$C_t=\begin{bmatrix}1&0&2\\1&1&t\\0&1&3\end{bmatrix}.$$

1. For which value of $t$ does the third column lie in the plane spanned by the first two columns?
2. For which values of $t$ is $C_t$ invertible? What is its column space in each case?
3. When $C_t$ is not invertible, find two different inputs that produce $\begin{bmatrix}2\\5\\3\end{bmatrix}$.

:::{tip} Solution
:class: dropdown

**1.** The first two columns span the plane $x-y+z=0$. The third column is on it exactly when $2-t+3=0$, or $t=5$. Then $\vec a_3=2\vec a_1+3\vec a_2$.

**2.** $C_t$ is invertible exactly when $t\ne5$. Its third column then adds a direction outside the plane, so the three columns are independent and their span is all of $\mathbb R^3$. When $t=5$, the column space is just the original plane.

**3.** The relationship gives

$$C_5\begin{bmatrix}2\\3\\0\end{bmatrix}
=C_5\begin{bmatrix}0\\0\\1\end{bmatrix}
=\begin{bmatrix}2\\5\\3\end{bmatrix}.$$
:::
::::
