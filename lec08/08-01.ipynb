# 8.1: Matrix multiplication

Last time, we've learned about the very important concept of multiplying a matrix by a vector. This operation is actually a special case of a more general operation: multiplying a matrix by another matrix. In this lecture, we will learn how to perform matrix multiplication and understand its properties.

## Definition

Given two matrices $A = (a_{ij})$ and $B = (b_{jk})$ of size $m \times n$ and $n \times p$ respectively, the **product** of $A$ and $B$, denoted as $C = AB$, is defined as a matrix of size $m \times p$ where each entry $c_{ik}$ is calculated as follows:
$$c_{ik} = a_{i1}b_{1k} + a_{i2}b_{2k} + \cdots + a_{in}b_{nk}.$$

Let's look at some computational examples to get a better grasp of this operation.
$$ \begin{bmatrix} 2 & 1 \\ 0 & 3 \end{bmatrix} \begin{bmatrix} 1 & 4 \\ 2 & 5 \end{bmatrix} = \begin{bmatrix} 2\cdot 1 + 1\cdot 2 & 2\cdot 4 + 1\cdot 5 \\ 0\cdot 1 + 3\cdot 2 & 0\cdot 4 + 3\cdot 5 \end{bmatrix} = \begin{bmatrix} 4 & 13 \\ 6 & 15 \end{bmatrix}. $$
$$ \begin{bmatrix} 1 \\ 2 \end{bmatrix} \begin{bmatrix} -1 & 3 & -2 \end{bmatrix} = \begin{bmatrix} 1\cdot (-1) & 1\cdot 3 & 1\cdot (-2) \\ 2\cdot (-1) & 2\cdot 3 & 2\cdot (-2) \end{bmatrix} = \begin{bmatrix} -1 & 3 & -2 \\ -2 & 6 & -4 \end{bmatrix}. $$
$$ \begin{bmatrix} 3&-1&2 \\ 0 & -4 & 1 \end{bmatrix} \begin{bmatrix} 1 \\ 3 \\ -2 \end{bmatrix} = \begin{bmatrix} 3\cdot 1 + (-1)\cdot 3 + 2\cdot (-2) \\ 0\cdot 1 + (-4)\cdot 3 + 1\cdot (-2) \end{bmatrix} = \begin{bmatrix} -4 \\ -14 \end{bmatrix}. $$

From the third example above, note that if the second matrix $B$ is a column vector, then the matrix multiplication operation is the exact same as the matrix-multiplication we've learned in the previous lecture. Hence we can say this new matrix multiplication operation is a generalization of the previous matrix-vector multiplication.


Based on these computations, let us give two interpretations of matrix multiplication that can be useful in carrying out computations and understanding the operation.


## Two interpretations

### Dot product of rows and columns
The first interpretation of matrix multiplication is to view the entry $c_{ik}$ of the product matrix $C$ as the dot product of the $i$-th row of $A$ and the $k$-th column of $B$. Let us denote the $A$ and $B$ in terms of their rows and column vectors as follows:
$$ A = \begin{bmatrix} \vec r_1^T \\ \vec r_2^T \\ \vdots \\ \vec r_m^T \end{bmatrix}, \quad B = \begin{bmatrix} \vec c_1 & \vec c_2 & \cdots & \vec c_p \end{bmatrix}. $$
Then their product can be calculated as
$$ C = \begin{bmatrix} \vec r_1^T \vec c_1 & \vec r_1^T \vec c_2 & \cdots & \vec r_1^T \vec c_p \\ \vec r_2^T \vec c_1 & \vec r_2^T \vec c_2 & \cdots & \vec r_2^T \vec c_p \\ \vdots & \vdots & \ddots & \vdots \\ \vec r_m^T \vec c_1 & \vec r_m^T \vec c_2 & \cdots & \vec r_m^T \vec c_p \end{bmatrix}
= \begin{bmatrix} \vec r_1 \cdot \vec c_1 & \vec r_1 \cdot \vec c_2 & \cdots & \vec r_1 \cdot \vec c_p \\ \vec r_2 \cdot \vec c_1 & \vec r_2 \cdot \vec c_2 & \cdots & \vec r_2 \cdot \vec c_p \\ \vdots & \vdots & \ddots & \vdots \\ \vec r_m \cdot \vec c_1 & \vec r_m \cdot \vec c_2 & \cdots & \vec r_m \cdot \vec c_p \end{bmatrix}. $$

### Matrix-vector multiplications
Another interpretation starts from our previous remark that multiplying a matrix by a column vector is a special case of matrix multiplication. In fact, we can view the multiplication of two matrices as a series of matrix-vector multiplications. Specifically, if we denote the columns of $B$ as $\vec c_1, \vec c_2, \ldots, \vec c_p$ as we did before, then we can write
$$ C = AB = \begin{bmatrix} A\vec c_1 & A\vec c_2 & \cdots & A\vec c_p \end{bmatrix}. $$
In other words, the $k$-th column of the product matrix $C$ is the result of multiplying the matrix $A$ by the $k$-th column of $B$.

## Basic properties

Just like we did for dot products and cross products, we also establish some basic properties of matrix multiplication with relation to addition and scalar multiplication of matrices. Let $A, B, C$ be matrices of sizes $m\times n$, $n\times k$, and $k\times l$ respectively. Let $t$ be a scalar. Then we have the following properties:

### Associative property
Matrix multiplication is associative. That is, we have
$$ A(BC) = (AB)C. $$
We also have associativity with scalar multiplication:
$$ t(AB) = (tA)B = A(tB). $$

### Distributive property
Matrix multiplication distributes over addition. That is, we have
$$ A(B + C) = AB + AC, \quad (A + B)C = AC + BC.$$

### Identity and zero

If we multiply a matrix $A$ by the identity matrix $I$ of appropriate size, we get back the same matrix. In other words, we have
$$ AI_n = I_m A = A.$$
If we multiply a matrix $A$ by the zero matrix of appropriate size, we get the zero matrix.


## Further properties

### Relation with transpose

Recall that the _transpose_ of a matrix $A$ is the matrix $A^T$ obtained by flipping the rows and columns of $A$. One might wonder how the transpose operation interacts with matrix multiplication. It turns out that the following property holds:
$$ (AB)^T = B^T A^T. $$

This relation leads to an interesting observation, namely that $A^T A$ and $AA^T$ are always symmetric matrices, since we have
$$ (A^T A)^T = A^T (A^T)^T = A^T A, \quad (AA^T)^T = (A^T)^T A^T = AA^T.$$
What are the entries of $A^T A$? Let $\vec r_i$ denote the $i$-th row vector of $A$ and $\vec c_j$ denote the $j$-th column vector of $A$. Note that $(A^T A)_{ij}$ is the dot product of the $i$-th row vector of $A^T$ and the $j$-th column vector of $A$. But the $i$-th row of $A^T$ is equal to the $i$-th column of $A$, so we have
$$ (A^T A)_{ij} = \vec c_i \cdot \vec c_j. $$
That is, the entries of $A^T A$ are the dot products of the column vectors of $A$. since $\vec c_i \cdot \vec c_j = \vec c_j \cdot \vec c_i$, we see that $(A^T A)_{ij} = (A^T A)_{ji}$, which is consistent with the fact that $A^T A$ is symmetric. Similarly, we can show that $(AA^T)_{ij} = \vec r_i \cdot \vec r_j$ consists of all dot products of the row vectors of $A$ and is also symmetric.

(*Optional*) Let's quickly justify the property above. Using the first interpretation of matrix multiplication as dot products of rows and columns, we see that the $(i,j)$-th entry of $(AB)^T$ is the $(j,i)$-th entry of $AB$, which is the dot product of the $j$-th row of $A$ and the $i$-th column of $B$, i.e.
$$ (AB)^T_{ij} = (AB)_{ji} = \vec r_j \cdot \vec c_i. $$
On the other hand, the $(i,j)$-th entry of $B^T A^T$ is the dot product of the $i$-th row of $B^T$ and the $j$-th column of $A^T$. Note that the $i$-th row vector of $B^T$ is precisely $\vec c_i^T$, and the $j$-th column vector of $A^T$ is $\vec r_j$. Hence we have
$$ (B^T A^T)_{ij} = \vec c_i \cdot \vec r_j$$
which is the same as what we obtained above. Since this holds for all $i$ and $j$, we conclude that $(AB)^T = B^T A^T$.


### Failure of commutativity

Unlike scalar multiplication and dot products, the matrix multiplication operation is _not_ commutative. That is, in general we have
$$ AB \neq BA.$$
We can't even say something like $AB = -BA$ as we did for cross products; the relation between the products $AB$ and $BA$ is much more unclear here.

Why does this commutativity not hold for matrix multiplication? There are many different explanations for this, but one thing to note is that $AB$ and $BA$ is not even defined for most pairs of matrices $A$ and $B$ due to size constraints. If $A$ is $m\times n$ and $B$ is $k\times l$, we need $n = k$ for $AB$ to be defined, and we need $l = m$ for $BA$ to be defined. Hence if $m \neq l$ or $n \neq k$, then one of the products is not even defined.

Even if both products are defined (so $A$, $B$ each have sizes $m\times n$ and $n\times m$ respectively), the products $AB$ and $BA$ have sizes $m\times m$ and $n \times n$ respectively, so we must have $m=n$ for them to be equal. In other words, $A$ and $B$ necessarily must be square matrices of the same size.

But even then, commutativity fails in general, as demonstrated by the following example:
$$ \begin{bmatrix} 2 & 1 \\ 1 & 0 \end{bmatrix}
\begin{bmatrix} 0 & -1 \\ 1 & 0 \end{bmatrix}
= \begin{bmatrix} 1 & -2 \\ 0 & -1 \end{bmatrix} $$
$$ \begin{bmatrix} 0 & -1 \\ 1 & 0 \end{bmatrix}
\begin{bmatrix} 2 & 1 \\ 1 & 0 \end{bmatrix}
= \begin{bmatrix} -1 & 0 \\ 2 & 1 \end{bmatrix} $$

This is to be expected, since the $(i,j)$-th entry of $AB$ is the dot product of the $i$-th row of $A$ and the $j$-th column of $B$, while the $(i,j)$-th entry of $BA$ is the dot product of the $i$-th row of $B$ and the $j$-th column of $A$. In general, these row and column vectors are unrelated.
