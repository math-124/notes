# 7.1: Basics of matrices

Let us briefly review what we have learned so far. We have learned the concept of a _vector_ as an ordered list of numbers that describe points (and displacements between points) in the $n$-dimensional space $\mathbb{R}^n$. We learned a good amount about the geometry of vectors -- how to find the length of a vector, how the _dot product_ measures the correlation of two vectors -- and this geometric language allowed us to describe affine lines and planes in two/three-dimensional spaces in terms of vectors.

In the last section, we looked at how systems of linear equations can be understood as finding intersections of lines and planes in two/three-dimensional space, and how _linear independence_ of the directional/normal vectors relates to the number of solutions of the system.

Solving systems of linear equations and understanding the number of solutions -- not only in two/three variables but in $n$ variables -- is an important objective in linear algebra with applications throughout physics, data science and more. To effectively study systems of linear equations in their full generality, we will need to introduce the concept of _matrices_.




## Definition of matrices

The formal definition of a matrix is very similar to that of a vector. A *matrix* is simply an ordered, rectangular array of numbers. Here are some examples of matrices:
$$ A = \begin{bmatrix} 2 & 1 \\ -1 & 3 \end{bmatrix},\ B = \begin{bmatrix} 1 \\ -1 \\ 2 \end{bmatrix}, \ C=\begin{bmatrix} 3.1 & \pi & -2.3 & 0.2  \\ -1/4 & 0 & 0 & 1 \end{bmatrix},\ D=\begin{bmatrix} 0 & 1 & -1 \end{bmatrix}\cdots $$

When representing matrices as a variable, we usually choose to denote it with a single capital letter as above. An arbitrary matrix will often be denoted like
$$ A = \begin{bmatrix} a_{11} & \cdots & a_{1n} \\ \vdots & \ddots & \vdots \\ a_{m1} & \cdots & a_{mn} \end{bmatrix},$$
where $a_{ij}$ denotes the entry in the $i$-th row and $j$-th column, i.e. the $(i,j)$-th entry. To indicate that $a_{ij}$ denotes the $(i,j)$-th entry of matrix $A$, we may sometimes write
$$ A = (a_{ij}).$$
Slightly confusingly, we may also choose to denote the $(i,j)$-th entry of $A$ as $A_{ij}$.

Note that a vector can also be viewed as a matrix. But unlike vectors, matrices can come in many different **shapes**. For matrix $A$ above which has $m$ rows and $n$ columns, and we say that $A$ is a $m\times n$ matrix ("$m$-by-$n$"). A vector always has only one column. If a matrix has the same number of rows adn columns, we say that it is a **square matrix**. For instance, $A$ above is a $2\times 2$ square matrix.

### Special matrices

The **identity matrix** of size $n$ is the $n\times n$ square matrix with $1$'s on the diagonal and $0$'s elsewhere. It is denoted as $I_n$. For instance, we have
$$ I_2 = \begin{bmatrix} 1 & 0 \\ 0 & 1 \end{bmatrix},\quad I_3 = \begin{bmatrix} 1 & 0 & 0 \\ 0 & 1 & 0 \\ 0 & 0 & 1 \end{bmatrix}.$$
If the size of the matrix is clear from context, we may simply write $I$ instead of $I_n$.

The **zero matrix** is a matrix (of any size) with all entries equal to $0$, and is denoted as $0$.

## Basic operations

Just like vectors, we can add matrices and multiply them with scalars.

### Addition

We add two matrices _with the same shape_ by adding each entry by entry. Abstractly, it is defined for $m\times n$ matrices $A = (a_{ij})$ and $B = (b_{ij})$ by
$$ A + B = \begin{bmatrix} a_{11} & \cdots & a_{1n} \\ \vdots & \ddots & \vdots \\ a_{m1} & \cdots & a_{mn} \end{bmatrix}
+ \begin{bmatrix} b_{11} & \cdots & b_{1n} \\ \vdots & \ddots & \vdots \\ b_{m1} & \cdots & b_{mn}\end{bmatrix}
= \begin{bmatrix} a_{11}+b_{11} & \cdots & a_{1n}+b_{1n} \\ \vdots & \ddots & \vdots \\ a_{m1}+b_{m1} & \cdots & a_{mn} + b_{mn}\end{bmatrix} . $$
Just like for vectors, we are not allowed to add two matrices with different shapes.

### Scalar multiplication

We can multiply a matrix with a scalar by multiplying the scalar to each entry of the matrix. Abstractly, it is defined by
$$ cA = \begin{bmatrix} ca_{11} & \cdots & ca_{1n} \\ \vdots & \ddots & \vdots \\ ca_{m1} & \cdots & ca_{mn} \end{bmatrix} .$$

### Transpose

Here is a new operation that we have not seen before for vectors. The **transpose** of matrix $A$, denoted $A^T$, is obtained by 'flipping the matrix along the diagonal'. More formally, we swap the $(i,j)$-th entry and the $(j,i)$-th entry to obtain the $n\times m$ matrix:
$$ A^{T} = \begin{bmatrix} a_{11} & \cdots & a_{m1} \\ \vdots & \ddots & \vdots \\ a_{n1} & \cdots & a_{nm} \end{bmatrix}.$$
It is important to note that taking the transpose of a matrix will change its shape unless the matrix was _square_, i.e. same number of rows and columns.

Let us look at some examples. The matrix
$$ A = \begin{bmatrix} 2 & 1 \\ -1 & 3 \end{bmatrix} $$
has transpose
$$ A^T = \begin{bmatrix} 2 & -1 \\ 1 & 3 \end{bmatrix}, $$
while the matrix
$$ B = \begin{bmatrix} 1 \\ -1 \\ 2 \end{bmatrix}$$
has transpose
$$ B^T = \begin{bmatrix} 1 & -1 & 2 \end{bmatrix}.$$

One property to note is that taking the transpose twice will return the original matrix, i.e. $(A^T)^T = A$.

If a matrix is equal to its transpose, we say that it is **symmetric**. Note that for size reasons, a symmetric matrix is necessarily square. For instance, the matrix
$$ C = \begin{bmatrix} 1 & 2 & 3 \\ 2 & 4 & 5 \\ 3 & 5 & 6 \end{bmatrix}$$
is symmetric, while the matrix
$$ D = \begin{bmatrix} 1 & 2 \\ 3 & 4 \end{bmatrix}$$
is not symmetric.

### Row and column vectors

We can similarly define transposes for vectors by viewing them as $n\times 1$ matrices, as we did for matrix $B$. More formally, the vector
$$ \vec x = \begin{bmatrix} x_1 \\ \vdots \\ x_n \end{bmatrix}$$
can be transposed to
$$ \vec x^T = \begin{bmatrix} x_1 & \cdots & x_n \end{bmatrix}.$$

Such $\vec x^T$, with only one row, is called a **row vector**. We will call the rows of a given matrix the **row vectors** of the matrix, and the columns of a given matrix the **column vectors** of the matrix. For example, for the matrix
$$ A = \begin{bmatrix} 2 & 1 \\ -1 & 3 \end{bmatrix},$$
the row vectors are
$$ \begin{bmatrix} 2 & 1 \end{bmatrix},\ \begin{bmatrix} -1 & 3 \end{bmatrix},$$
while the column vectors are
$$ \begin{bmatrix} 2 \\ -1 \end{bmatrix},\ \begin{bmatrix} 1 \\ 3 \end{bmatrix}.$$
