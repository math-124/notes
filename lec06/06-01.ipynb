# 6.1: Linear independence

## Review of linear combination and span

Let us recall the formal definitions of linear combination and span. Given vectors $\vec v_1, \ldots, \vec v_n$, a **linear combination** of these vectors is any sum of their scalar multiples, i.e. any vector of the form
$$ a_1 \vec v_1 + \ldots + a_n \vec v_n $$
for scalars $a_i$. The **span** of these vectors is the set of all linear combinations, which is denoted as
$$ \langle v_1, \ldots, v_n \rangle = \left\{ a_1 v_1 + \ldots + a_n v_n : a_1,\ldots, a_n \in \mathbb{R} \right\}.$$

In the previous lectures, we described lines (respectively planes) in $\mathbb{R}^3$ as the span of one (resp. two) vector using the vector-parametric form. For instance, the $xy$-plane in $\mathbb{R}^3$ can be described as the span of $\vec e_1$ and $\vec e_2$.

Sometimes, it happens that two vectors in $\mathbb{R}^3$ do not span a plane. For instance, consider the vectors
$$ \vec u = \begin{bmatrix} 1 \\ 1 \\ -2\end{bmatrix},\ \vec v = \begin{bmatrix} 2 \\ 2 \\ -4\end{bmatrix}.$$
These two vectors only span a line in $\mathbb{R}^3$. This is because they are parallel to each other; indeed notice that $\vec v = 2\vec u$. Let us interpret this as $\vec v$ belonging to the span of $\vec u$.

<!-- INSERT PICTURE OF VECTOR U AND V SPANNING ONLY A LINE -->


Recall also that $\mathbb{R}^3$ can be understood as the span of the standard basis vectors $\vec e_1, \vec e_2, \vec e_3$, since any vector with $n$ entries can be decomposed as
$$ \begin{bmatrix} x_1 \\ x_2 \\ x_3 \end{bmatrix} = x_1 \vec e_1 + x_2 \vec e_2 + x_3 \vec e_3.$$

But this is respect to a well-behaved choice of vectors $\vec e_1, \vec e_2, \vec e_3$. In general, it is possible that three vectors in $\mathbb{R}^3$ do not span the whole space. For instance, consider the following three vectors in $\mathbb{R}^3$:
$$ \vec u = \vec e_1,\ \vec v = \vec e_2,\ \vec w = 3\vec e_1 - 2\vec e_2.$$
These vectors only span the $xy$-plane. This is because $\vec u$ and $\vec v$ already spans the $xy$-plane, and $\vec w$ is contained in the span of $\vec u$ and $\vec v$ and thus the addition of $\vec w$ does not contribute anything new to the span. Note the similarly with the earlier interpretation of why two vectors did not span a plane and only spanned a line instead.

## Definition

The observations above are generalized into the very important concept of linear independence. Let us jump straight into the definition.

We say that a collection of vectors $\vec v_1, \ldots, \vec v_n$ is **linearly dependent** if one of the vectors $\vec v_i$ can be expressed as a linear combination of the other vectors, i.e. if $\vec v_i$ belongs to the span $\langle \vec v_1, \ldots, \vec v_{i-1}, \vec v_{i+1},\ldots, \vec v_n \rangle$. Otherwise, we say they are **linearly independent**.

Let us look at some examples.

1. The vectors $\vec e_1, \vec e_2, \vec e_3$ in $\mathbb{R}^3$ are linearly independent, since none of the vectors is contained in the span of the other two vectors. Similarly convince yourself that the standard basis vectors $\vec e_1, \ldots, \vec e_n$ in $\mathbb{R}^n$ are linearly independent.
2. The vectors $\vec e_1, \vec e_2$ in $\mathbb{R}^3$ are linearly independent. More generally, note that any two vectors are linearly independent if and only if they are not scalar multiples of each other.
3. The vectors
    $$ \vec u = \begin{bmatrix} 1 \\ 1 \\ -2\end{bmatrix},\ \vec v = \begin{bmatrix} 2 \\ 2 \\ -4\end{bmatrix}.$$
    are linearly dependent since $\vec v$ belongs to the span of $\vec u$, i.e. $\vec v = 2\vec u$.
4. The following vectors in $\mathbb{R}^3$:
    $$ \vec u = \vec e_1,\ \vec v = \vec e_2,\ \vec w = 3\vec e_1 - 2\vec e_2$$
    are linearly dependent since $\vec w$ is a linear combination of $\vec u$ and $\vec v$.

One thing to note is that if $\vec v_1, \ldots, \vec v_n$ is linearly independent, then any sub-collection of these $n$ vectors are also linearly independent. This is because if the sub-collection was linearly dependent, one of the vectors in $\vec v_1, \ldots, \vec v_n$ would be a linear combination of the other vectors, making $\vec v_1, \ldots, \vec v_n$ linearly dependent as well. Using this observation, the second example can also be deduced from the first example.

Another smaller thing to note is that the zero vector is always linearly dependent with any other vector, since it can be expressed as a linear combination of any vector with a coefficient of zero.


## Geometric interpretation

Let us focus on the case of $\mathbb{R}^3$ and visualize the geometric meaning of linear independence.

Recall that two vectors in $\mathbb{R}^3$ are linearly independent precisely when they are not parallel. This means that the two vectors are linearly independent precisely when they span a plane. If they are parallel, then they are linearly dependent and only span a line or the origin. Indeed, note from the example 2 that the linearly independent vectors $\vec e_1$ and $\vec e_2$ span the $xy$-plane, while in example 3 the two vectors only span the line given by
$$ \vec r = t\begin{bmatrix} 1 \\ 1 \\ -2\end{bmatrix} $$
in vector-parametric form.

Similarly, suppose we are given three vectors in $\mathbb{R}^3$. If the three vectors are linearly independent, then they span a three-dimensional space, i.e., the entire $\mathbb{R}^3$. If they are linearly dependent, then they span a plane, a line, or the origin. Indeed, note from the example 4 that the linearly dependent vectors $\vec u$, $\vec v$, and $\vec w$ span only the $xy$-plane, while in example 1 the linearly independent vectors $\vec e_1$, $\vec e_2$, and $\vec e_3$ span the entire $\mathbb{R}^3$.
