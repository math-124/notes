# 6.2: System of linear equations

The concept of linear independence is crucial in understanding the behavior of systems of linear equations. More precisely, they will tell us when a system has a unique solution, infinitely many solutions, or no solution. We will do this in more detail in future sections, but let us give a taste of how this works in two and three dimensions.



## Two-variable system of linear equations

### Solving by elimination
Let us briefly recall how to solve a two-variable system of linear equations. We will consider the following three examples:
$$ (A): \begin{cases} 3x + 2y = 13 \\ x - 2y = -1 \end{cases};\quad (B): \begin{cases} x-y =3 \\ -2x+2y = 0\end{cases};\quad (C): \begin{cases} x-y =3 \\ -2x+2y = -6\end{cases}.$$
Recall that we solve systems of linear equations by _elimination_, i.e. we express one variable in terms of the other and substitute it into the other equation. For instance, let us solve the system (A) by first rearranging the second equation for $x$:
$$ x = 2y-1 \tag{$\ast$}$$
and substitute this into the first equation:
$$ 3(2y-1) + 2y = 13.$$
Expanding the left side, we get
$$ 8y-3 = 13$$
and therefore $y = 2$. Substituting this back into the equation for $x$ labelled ($\ast$), we get $x = 2(2)-1 = 3$. Thus we obtain the unique solution for system (A) as:
$$x = 3,\ y = 2.$$



### A graphical interpretation

The system above can be solved more geometrically by representing the equations as graphs of linear functions. Drawing the graphs corresponding to $3x+2y = 13$ and $x-2y = -1$ on the Euclidean plane, we see that the two lines meet at exactly one point. That point corresponds to the unique solution of the system: $(x,y) = (3,2)$.

Recall that the line given by the graph $ax + by = c$ in $\mathbb{R}^2$ can be understood as pointing in the direction of, i.e. having directional vector
$$\vec v = \begin{bmatrix} b\\-a\end{bmatrix}.$$
The lines meeting at exactly one point is closely related to the fact that the lines are not parallel, i.e. that their directional vectors $\begin{bmatrix} 2 \\ -3\end{bmatrix}$ and $\begin{bmatrix} -2 \\ -1 \end{bmatrix}$ are linearly independent.

This graphical interpretation will help us understand the slightly more confusing examples of systems (B) and (C) given above. Let us first solve the system (B) using elimination: Solving the second equation for $x$, we obtain $x = y$, and plugging into the first equation gives us $0 = 3$ which is clearly incorrect! This indicates that the system has no solution at all.

Representing this system in terms of graphs as we did for the previous example, we see that the two equations represent parallel lines -- i.e. their directional vectors $\begin{bmatrix} -1 \\ -1\end{bmatrix}$ and $\begin{bmatrix} 2 \\ 2 \end{bmatrix}$ are linearly dependent -- that never meet.

Finally, let us consider the system (C). Solving by elimination, we obtain $x = y-3$ from the second equation which is identical to the first equation. Plugging into the first equation, we simply get $0 = 0$. What does this mean?

Plotting the two graphs of the equations, we see that they define the same line $y = x+3$, i.e. the two lines of the equation are identical. Hence all points on the line correspond to a solution of the system, i.e. there are _infinitely many_ solutions, which can be described by the scalar-parametric form of the line as follows:
$$ x = t,\ y = t + 3.$$
This was again only possible because the directional vectors $\begin{bmatrix} -1 \\ -1\end{bmatrix}$ and $\begin{bmatrix} 2 \\ 2 \end{bmatrix}$ were linearly dependent.
