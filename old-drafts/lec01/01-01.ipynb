
# 1.1: What is linear algebra?

**Linear algebra** is the study of linear objects in mathematics. "Objects" here can mean several different things; the goal of this class is to understand that they are intimately related to each other.

* Linear spaces: Lines or planes in our usual two or three-dimensional space are examples of what we'll call _linear subspaces_.
* Linear functions: Suppose an apple costs 2 dollars and a banana costs 1 dollar. The cost of buying $x$ apples and $y$ bananas is $2x+y$, which is what we call a _linear function_ in $x$ and $y$.
* Linear equations: If we have 12 dollars total to spend on fruits, we obtain a _linear equation_ $2x+y = 12$.


These three examples fit together in one. The equation $2x+y=12$ form a straight line on the $x$-$y$ plane, which is an example a simple linear space.


```python
import numpy as np
import plotly.graph_objects as go
import plotly.io as pio

x = np.linspace(-0.5, 6.5, 200)
y = 12 - 2 * x
integer_x = np.arange(1, 6)
integer_y = 12 - 2 * integer_x

fig = go.Figure()
fig.add_trace(go.Scatter(x=x, y=y, mode="lines", name=r"$2x+y=12$", line=dict(width=3)))
fig.add_trace(go.Scatter(
    x=integer_x, y=integer_y, mode="markers+text", name="positive integer points",
    text=[f"({px}, {py})" for px, py in zip(integer_x, integer_y)],
    textposition="top right", marker=dict(size=9),
))
fig.update_layout(
    title=r"The line $2x+y=12$", width=700, height=520,
    xaxis_title="$x$", yaxis_title="$y$", font_family="Palatino",
)
fig.update_xaxes(range=[-0.5, 6.5], dtick=1, zeroline=True)
fig.update_yaxes(range=[-1, 13], dtick=2, zeroline=True)
from pathlib import Path as _PlotPath
import sys as _plot_sys
_notes_root = next(path for path in [_PlotPath.cwd(), *_PlotPath.cwd().parents] if (path / "myst.yml").exists())
if str(_notes_root) not in _plot_sys.path:
    _plot_sys.path.insert(0, str(_notes_root))
from plot_style import style_plotly as _style_plotly

_style_plotly(fig, renderer="png")
fig.show(scale=2)
```
