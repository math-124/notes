# 3.1: Matrices

We can put several vectors together to form a matrix.


---

## Introduction

Suppose we have three vectors in $\mathbb R^2$:

$$\begin{bmatrix}2\\1\end{bmatrix},\qquad
\begin{bmatrix}-1\\4\end{bmatrix},\qquad
\begin{bmatrix}3\\0\end{bmatrix}.$$

Place them side by side, keeping their order:

$$A=\begin{bmatrix}2&-1&3\\1&4&0\end{bmatrix}.$$



:::{note} Definition: Matrix and shape
A **matrix** is a rectangular array of numbers, organized into rows and columns. A matrix with $n$ rows and $d$ columns has **shape** $n\times d$, read "$n$-by-$d$."

We write $A\in\mathbb R^{n\times d}$ when its entries are real numbers. **Rows come first; columns come second.**
:::

### Shape

Here $A$ has two rows and three columns, so $A\in\mathbb R^{2\times3}$. 

Here are several more matrices, with their shape underneath:

$$\underbrace{\begin{bmatrix}
1&0&2\\3&1&-1\\0&4&2\\2&-2&1\\5&0&3
\end{bmatrix}}_{5\times3}
\qquad
\underbrace{\begin{bmatrix}1&2&0&-1\\3&0&4&2\end{bmatrix}}_{2\times4}
\qquad
\underbrace{\begin{bmatrix}2&1&0\\0&3&-1\\4&0&2\end{bmatrix}}_{3\times3}.$$

A **square matrix** has the same number of rows and columns. Square matrices are special in many ways, and we will spend time discussing them in detail.


### Entries

We use capital letters such as $A$ and $B$ for matrices. The notation $a_{ij}$ denotes the entry of $A$ in **row $i$, column $j$**, counting from 1. In our original matrix,

$$a_{12}=-1,\qquad a_{21}=1,\qquad a_{23}=0.$$

For an arbitrary $n\times d$ matrix,

$$A=\begin{bmatrix}
a_{11}&a_{12}&\cdots&a_{1d}\\
a_{21}&a_{22}&\cdots&a_{2d}\\
\vdots&\vdots&\ddots&\vdots\\
a_{n1}&a_{n2}&\cdots&a_{nd}
\end{bmatrix}.$$

Two matrices are **equal** if they have the same shape and equal entries at every corresponding position. Merely containing the same numbers is not enough.

## Rows and columns as vectors

Return to

$$A=\begin{bmatrix}2&-1&3\\1&4&0\end{bmatrix}.$$

There are two groups of vectors embedded in this matrix:

- The **columns** make up **three** vectors in $\mathbb R^2$:

  $$\begin{bmatrix}2\\1\end{bmatrix},\qquad
  \begin{bmatrix}-1\\4\end{bmatrix},\qquad
  \begin{bmatrix}3\\0\end{bmatrix}.$$

- The **rows** make up **two** vectors in $\mathbb R^3$:

  $$\begin{bmatrix}2\\-1\\3\end{bmatrix},\qquad
  \begin{bmatrix}1\\4\\0\end{bmatrix}.$$

These are two ways to view the same matrix.

More generally, an $n\times d$ matrix contains $d$ columns, viewed as vectors in $\mathbb R^n$, and $n$ rows, viewed as vectors in $\mathbb R^d$.

A subtlety is that these are different objects:

$$\begin{bmatrix}2\\-1\\3\end{bmatrix}
\qquad\text{and}\qquad
\begin{bmatrix}2&-1&3\end{bmatrix}.$$

The first is a $3\times1$ matrix, which we call a vector in $\mathbb R^3$. The second is a $1\times3$ matrix, which we might call a **row vector**. But $\mathbb R^n$ always refers to ordinary column vectors. When we discuss the vectors formed by a matrix's rows, we write their entries vertically, as above.

Viewing a matrix as a collection of vectors lets us ask familiar questions:

- **What is the span of the columns of $A$?** These columns are vectors in $\mathbb R^2$, so their span is a subset of $\mathbb R^2$.
- **Are the rows of $A$ linearly independent?** Here we ask whether the two vectors in $\mathbb R^3$ formed from the rows are linearly independent.



### Example: A dataset

Suppose we survey five Math 124 students about their past week. For each student, we record hours spent studying, practice problems completed, and lectures attended. The following numbers are **made up**:

| Student | Study hours | Practice problems | Lectures attended |
| :--- | ---: | ---: | ---: |
| Student 1 | 4 | 12 | 2 |
| Student 2 | 6 | 18 | 2 |
| Student 3 | 3 | 8 | 1 |
| Student 4 | 5 | 15 | 2 |
| Student 5 | 2 | 6 | 1 |

Store the numerical values in a matrix, keeping the row and column order:

$$X=\underbrace{\begin{bmatrix}4&12&2\\6&18&2\\3&8&1\\5&15&2\\2&6&1\end{bmatrix}}_{5\times3}.$$

With **one row per person and one column per feature (or variable)**, the two views have concrete meanings:

- Each of the five rows is a vector in $\mathbb R^3$: three measurements about **one student**.
- Each of the three columns is a vector in $\mathbb R^5$: one measurement recorded across **all five students**. For example, the first column contains everyone's study hours.

We could instead choose **one row per feature and one column per person**:

$$\underbrace{\begin{bmatrix}4&6&3&5&2\\12&18&8&15&6\\2&2&1&2&1\end{bmatrix}}_{3\times5}.$$

Either arrangement works. We just need to say what the rows and columns represent.

## Addition and scalar multiplication

Addition and scalar multiplication work just as they did for vectors.

:::{note} Definition: Matrix addition
If $A$ and $B$ both have shape $n\times d$, their **sum** $A+B$ is the $n\times d$ matrix whose $(i,j)$ entry is $a_{ij}+b_{ij}$.

Matrix addition requires the **same shape**, so that every entry has a corresponding entry to add.
:::

For example,

$$\begin{bmatrix}2&-1&3\\1&4&0\end{bmatrix}
+\begin{bmatrix}-1&2&0\\3&-2&5\end{bmatrix}
=\begin{bmatrix}1&1&3\\4&2&5\end{bmatrix}.$$

Both inputs have shape $2\times3$, and so does their sum. A $2\times3$ matrix and a $3\times2$ matrix cannot be added, even though both contain six entries.

:::{note} Definition: Scalar multiplication
If $c$ is a real number and $A$ is a matrix, the **scalar multiple** $cA$ is obtained by multiplying every entry of $A$ by $c$. Its shape is the same as that of $A$.
:::

For example,

$$-2\begin{bmatrix}2&-1&3\\1&4&0\end{bmatrix}
=\begin{bmatrix}-4&2&-6\\-2&-8&0\end{bmatrix}.$$

We define subtraction by $A-B=A+(-1)B$. The **zero matrix** $0_{n\times d}$ has $n$ rows, $d$ columns, and every entry equal to zero; $A+0_{n\times d}=A$ for any $n\times d$ matrix $A$.
