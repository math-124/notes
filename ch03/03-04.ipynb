# 3.4: Geometry of linear transformations

A matrix can rotate a robot's displacement, stretch a drawing, describe a simple shear deformation, or extract a force component along a rail. We will construct these matrices using [Chapter 3.3](03-03.ipynb): **the columns are the images of the standard basis vectors**.

For each example, identify the input space, output space, and attainable outputs.

---

## Following the unit square

The **image** of a vector is its output. The image of a set consists of the outputs of every vector in that set.

The unit square has corners $\vec0$, $\vec e_1$, $\vec e_2$, and $\vec e_1+\vec e_2$. If $A$ has columns $\vec a_1,\vec a_2$, their images are

$$\vec0,\qquad\vec a_1,\qquad\vec a_2,\qquad\vec a_1+\vec a_2.$$

Every point in the square is $s\vec e_1+t\vec e_2$ with $0\le s,t\le1$. Its image is $s\vec a_1+t\vec a_2$. Thus the **filled square becomes a filled parallelogram**, possibly collapsed to a segment or a point. The columns form its edges starting at the origin.

Similarly, a line $\vec p+t\vec u$ becomes $A\vec p+tA\vec u$: another line if $A\vec u\ne\vec0$, or a point otherwise.

Transforming a drawing means transforming every point. For an image stored on a pixel grid, software must also decide how to sample and fill pixels; our focus here is the geometric coordinate transformation.

## Scaling and reflection

### Scaling coordinates

The matrix

$$D=\begin{bmatrix}a&0\\0&b\end{bmatrix}$$

sends $\begin{bmatrix}x\\y\end{bmatrix}$ to $\begin{bmatrix}ax\\by\end{bmatrix}$. With $a=2$ and $b=1/2$, the unit square becomes a rectangle twice as wide and half as tall.

This models resizing a drawing independently in its horizontal and vertical directions. Uniform scaling uses $a=b$. Converting displacement coordinates from meters to centimeters uses $100I_2$. A negative factor reverses a coordinate's direction; a zero factor collapses that direction.

### Reflection

Reflection across the horizontal axis fixes $\vec e_1$ and sends $\vec e_2$ to $-\vec e_2$, giving

$$F=\begin{bmatrix}1&0\\0&-1\end{bmatrix}.$$

Reflection across $y=x$ exchanges the standard basis vectors, giving

$$S=\begin{bmatrix}0&1\\1&0\end{bmatrix}.$$

Multiplication verifies the rules for every input: $F\begin{bmatrix}x\\y\end{bmatrix}=\begin{bmatrix}x\\-y\end{bmatrix}$ and $S\begin{bmatrix}x\\y\end{bmatrix}=\begin{bmatrix}y\\x\end{bmatrix}$. They are linear because they are fixed matrix maps.

In a mechanical drawing with a symmetry axis through the origin, reflection generates the mirrored part. Reflecting across an offset line generally requires an affine transformation.

## Rotation

Counterclockwise rotation through a **fixed** angle $\theta$ about the origin sends

$$\vec e_1\longmapsto\begin{bmatrix}\cos\theta\\\sin\theta\end{bmatrix},\qquad
\vec e_2\longmapsto\begin{bmatrix}-\sin\theta\\\cos\theta\end{bmatrix}.$$

Rotation preserves addition geometrically: rotating a vector-addition parallelogram rotates all its sides together. It also preserves scalar multiples, including negative ones. Thus it is linear, and the two basis images give

$$\boxed{R_\theta=\begin{bmatrix}\cos\theta&-\sin\theta\\\sin\theta&\cos\theta\end{bmatrix}.}$$

Positive angles give counterclockwise rotations, using the usual axes with the second coordinate increasing upward. For $90^\circ$,

$$R_{90^\circ}=\begin{bmatrix}0&-1\\1&0\end{bmatrix},\qquad R_{90^\circ}\begin{bmatrix}x\\y\end{bmatrix}=\begin{bmatrix}-y\\x\end{bmatrix}.$$

The unit square remains a square. Rotation preserves lengths and angles; these are special properties, not requirements for linearity.

### Example: Turning a robot's motion command

A robot's local forward and left directions initially align with the global coordinate axes. A displacement command $\begin{bmatrix}2\\1\end{bmatrix}$ means two meters forward and one meter left. If the robot turns $90^\circ$ counterclockwise, this command corresponds to global displacement

$$R_{90^\circ}\begin{bmatrix}2\\1\end{bmatrix}=\begin{bmatrix}-1\\2\end{bmatrix}\text{ meters}.$$

To obtain the endpoint's global **position**, add the robot's starting position as a separate affine offset. The rotation is linear in the displacement for a fixed angle; its dependence on $\theta$ is not linear.

### Rotation in three dimensions

A turntable rotates around the $z$-axis while keeping height fixed:

$$R_z=\begin{bmatrix}\cos\theta&-\sin\theta&0\\\sin\theta&\cos\theta&0\\0&0&1\end{bmatrix}.$$

Its three columns are the rotated standard basis vectors of $\mathbb R^3$.

## Shear

A horizontal shear has the rule

$$H_k\begin{bmatrix}x\\y\end{bmatrix}=\begin{bmatrix}x+ky\\y\end{bmatrix},\qquad H_k=\begin{bmatrix}1&k\\0&1\end{bmatrix}.$$

It fixes $\vec e_1$ and sends $\vec e_2$ to $\begin{bmatrix}k\\1\end{bmatrix}$. The unit square becomes a parallelogram with corners

$$\vec0,\quad\begin{bmatrix}1\\0\end{bmatrix},\quad\begin{bmatrix}k\\1\end{bmatrix},\quad\begin{bmatrix}1+k\\1\end{bmatrix}.$$

### Example: A simple model of material deformation

Imagine a rubber sheet whose bottom edge stays fixed while its top slides horizontally. In an idealized simple shear, each horizontal layer slides by an amount proportional to its height. For sheet height $h$ and top displacement $\delta$, $k=\delta/h$.

If $h=10$ cm and $\delta=2$ cm, a point initially at $\begin{bmatrix}3\\5\end{bmatrix}$ cm moves to

$$H_{1/5}\begin{bmatrix}3\\5\end{bmatrix}=\begin{bmatrix}4\\5\end{bmatrix}\text{ cm}.$$

This is a geometric model, not a claim that every material behaves this way. Unlike a translation, the displacement depends on height and is zero at the bottom.

## Orthogonal projection onto a line

Let $\ell$ pass through the origin with unit direction $\vec w=\begin{bmatrix}w_1\\w_2\end{bmatrix}$. Recall

$$P(\vec x)=(\vec x\cdot\vec w)\vec w.$$

This is linear by the dot product properties:

$$P(\vec u+\vec v)=((\vec u+\vec v)\cdot\vec w)\vec w=P(\vec u)+P(\vec v),\qquad P(c\vec u)=cP(\vec u).$$

Projecting the standard basis gives

$$P(\vec e_1)=w_1\vec w=\begin{bmatrix}w_1^2\\w_1w_2\end{bmatrix},\qquad P(\vec e_2)=w_2\vec w=\begin{bmatrix}w_1w_2\\w_2^2\end{bmatrix}.$$

Therefore the matrix is

$$\boxed{P=\begin{bmatrix}w_1^2&w_1w_2\\w_1w_2&w_2^2\end{bmatrix}.}$$

This derives the formula supplied in Lab 6. For that lab's vector $\vec w=\begin{bmatrix}3/5\\4/5\end{bmatrix}$,

$$P=\frac1{25}\begin{bmatrix}9&12\\12&16\end{bmatrix}.$$

Both columns lie on $\ell$, and the unit square collapses to the segment from zero to $\begin{bmatrix}21/25\\28/25\end{bmatrix}$.

### Example: Force along a guide rail

A rail points along $\vec w$. An applied force $\vec v=\begin{bmatrix}7\\1\end{bmatrix}$ newtons has component along the rail

$$P\vec v=\begin{bmatrix}3\\4\end{bmatrix}\text{ newtons}.$$

The remaining component is $\vec v-P\vec v=\begin{bmatrix}4\\-3\end{bmatrix}$, perpendicular to the rail. This separates the force along the permitted direction of travel from the transverse component. The rail direction stays fixed; the map is linear in the force.

Projection fixes vectors on the line, sends perpendicular vectors to zero, and satisfies $P(P(\vec x))=P(\vec x)$. The output space is still $\mathbb R^2$, although the attainable outputs form only a line.

:::{attention} Landing on a line is not enough
The matrix $\begin{bmatrix}3&6\\-2&-4\end{bmatrix}$ from Chapter 3.2 sends every output to a line but does not fix vectors on that line. Orthogonal projection must land on the target and leave a perpendicular difference between input and output.
:::

## Orthogonal projection onto a plane

The plane $W$ with equation $x+y=0$ has unit normal $\vec n=\frac1{\sqrt2}\begin{bmatrix}1\\1\\0\end{bmatrix}$. Subtract the normal component:

$$\begin{aligned}
P_W\left(\begin{bmatrix}x\\y\\z\end{bmatrix}\right)
&=\begin{bmatrix}x\\y\\z\end{bmatrix}-\left(\begin{bmatrix}x\\y\\z\end{bmatrix}\cdot\vec n\right)\vec n\\
&=\begin{bmatrix}(x-y)/2\\(y-x)/2\\z\end{bmatrix}.
\end{aligned}$$

Applying this to the three standard basis vectors produces the columns of

$$\boxed{P_W=\begin{bmatrix}1/2&-1/2&0\\-1/2&1/2&0\\0&0&1\end{bmatrix}.}$$

For example,

$$P_W\begin{bmatrix}3\\1\\2\end{bmatrix}=\begin{bmatrix}1\\-1\\2\end{bmatrix}.$$

The output lies in $W$; the removed vector $\begin{bmatrix}2\\2\\0\end{bmatrix}$ is normal to $W$.

### Example: Tangential and normal forces

For a flat surface parallel to $W$, this extracts the **tangential component** of a force. The removed component is its normal component. The same decomposition finds the part of a velocity parallel to the surface.

Here the inputs represent forces or velocities. For projecting **positions** onto an offset plane $\vec n\cdot\vec x=c$ with unit normal $\vec n$, the formula becomes $\vec x-(\vec n\cdot\vec x-c)\vec n$, which is affine when $c\ne0$.

The input and output spaces of $P_W$ are both $\mathbb R^3$. Attainable outputs form a plane. Vectors on the plane stay fixed, normal vectors go to zero, and repeating the projection changes nothing.

## Different input and output dimensions

### A surface parameterization

To position a tool on $z=x+y$, choose horizontal coordinates $s,t$ and set its height to $s+t$:

$$E:\mathbb R^2\to\mathbb R^3,\qquad E\begin{bmatrix}s\\t\end{bmatrix}=\begin{bmatrix}s\\t\\s+t\end{bmatrix},\qquad E=\begin{bmatrix}1&0\\0&1\\1&1\end{bmatrix}.$$

The columns describe changing each input coordinate separately. The unit square becomes a parallelogram on the plane, with corners

$$\vec0,\quad\begin{bmatrix}1\\0\\1\end{bmatrix},\quad\begin{bmatrix}0\\1\\1\end{bmatrix},\quad\begin{bmatrix}1\\1\\2\end{bmatrix}.$$

Every point on the plane can be produced; points off it cannot. Three output coordinates do not imply three independent input controls.

### A top view

An orthographic top view in an engineering drawing discards height:

$$C:\mathbb R^3\to\mathbb R^2,\qquad C\begin{bmatrix}x\\y\\z\end{bmatrix}=\begin{bmatrix}x\\y\end{bmatrix},\qquad C=\begin{bmatrix}1&0&0\\0&1&0\end{bmatrix}.$$

Changing only height has no effect, explaining the zero third column. This is an orthographic coordinate map, not a perspective camera model with division by depth.

Compare it with projection onto the $xy$-plane inside $\mathbb R^3$:

$$Q:\mathbb R^3\to\mathbb R^3,\qquad Q\begin{bmatrix}x\\y\\z\end{bmatrix}=\begin{bmatrix}x\\y\\0\end{bmatrix},\qquad Q=\begin{bmatrix}1&0&0\\0&1&0\\0&0&0\end{bmatrix}.$$

$C$ returns two coordinates; $Q$ returns three. Their output spaces differ.

### A non-geometric example: Sensor summaries

Suppose $\begin{bmatrix}x_1\\x_2\\x_3\end{bmatrix}$ records three temperature changes in degrees Celsius. To report the mean change and the difference between the first and third sensors, use

$$M=\begin{bmatrix}1/3&1/3&1/3\\1&0&-1\end{bmatrix}.$$

Input $\begin{bmatrix}3\\6\\9\end{bmatrix}$ produces $\begin{bmatrix}6\\-6\end{bmatrix}$. This is another $\mathbb R^3\to\mathbb R^2$ linear transformation; the outputs are two summaries rather than coordinates of a point.

:::{admonition} Solution
:class: dropdown
$E$ takes $\mathbb R^2$ to $\mathbb R^3$ and reaches exactly $z=x+y$. Both $C$ and $Q$ send $\begin{bmatrix}1\\2\\0\end{bmatrix}$ and $\begin{bmatrix}1\\2\\5\end{bmatrix}$ to the same output.
:::

## Looking ahead: Applying two transformations

Taking a top view of a point on the tool surface gives

$$C\left(E\begin{bmatrix}s\\t\end{bmatrix}\right)=\begin{bmatrix}s\\t\end{bmatrix}.$$

In the reverse order,

$$E\left(C\begin{bmatrix}x\\y\\z\end{bmatrix}\right)=\begin{bmatrix}x\\y\\x+y\end{bmatrix}.$$

The second sequence moves a point to $z=x+y$ by changing only its height. It fixes points already on the plane, but is **not orthogonal projection**: a vertical direction is not normal to this plane.

To apply transformations in sequence, the first output must be an allowed input to the second. The two orders here even have different input and output spaces.

Composition preserves linearity: applying each map to a linear combination successively preserves its coefficients. Can a single matrix describe each sequence? We can find it by following the standard basis vectors through both operations and collecting the final outputs as columns. This leads to matrix multiplication.
