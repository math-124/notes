# 3.7: Transposes

So far, we've added matrices, scaled them, multiplied them, and found inverses. Here, we'll introduce another operation: the **transpose**, which exchanges a matrix's rows and columns.

The operation itself has a straightforward definition, but has a lot of power!

---

## Transposing a matrix

Let's start with an example:

$$A=\begin{bmatrix}1&2\\0&3\\-1&1\end{bmatrix}.$$

The first column has entries $1,0,-1$, and the second has entries $2,3,1$. To **transpose** $A$, we turn those columns into rows, keeping their order:

$$A^T=\begin{bmatrix}1&0&-1\\2&3&1\end{bmatrix}.$$

Read $A^T$ as "$A$ transpose." $A$ has shape $3\times2$, while $A^T$ has shape $2\times3$. Each entry is still there; it has just changed position.

:::{note} Definition: Transpose
If $A$ has shape $n\times d$, its **transpose**, $A^T$, has shape $d\times n$, with entries

$$(A^T)_{ij}=A_{ji}.$$

In words, entry $(i,j)$ of $A^T$ is entry $(j,i)$ of $A$. The rows of $A$ become the columns of $A^T$, and the columns of $A$ become the rows of $A^T$.
:::

For a square matrix, we can picture this as reflecting its entries across the main diagonal. The diagonal entries stay where they are:

$$\begin{bmatrix}a&b\\c&d\end{bmatrix}^T
=\begin{bmatrix}a&c\\b&d\end{bmatrix}.$$

What happens if we transpose twice? We return each entry to its original position, so

$$\boxed{(A^T)^T=A.}$$

Transposing also works nicely with addition and scalar multiplication:

$$(A+B)^T=A^T+B^T,\qquad (cA)^T=cA^T.$$

For the first identity, $A$ and $B$ must have the same shape. Both identities follow from looking at what happens to each entry. The product of two matrices will require a little more care.

## Vectors, transposes, and the dot product

Remember, when we say "vector," we mean a **column vector**. A vector $\vec u\in\mathbb R^d$ has shape $d\times1$, so its transpose has shape $1\times d$:

$$\vec u=\begin{bmatrix}u_1\\\vdots\\u_d\end{bmatrix},\qquad
\vec u^T=\begin{bmatrix}u_1&\cdots&u_d\end{bmatrix}.$$

$\vec u^T$ is a one-row matrix, also called a **row vector**. This gives us a new way to write the dot product. If $\vec u,\vec v\in\mathbb R^d$, then

$$\vec u^T\vec v
=\begin{bmatrix}u_1&\cdots&u_d\end{bmatrix}
\begin{bmatrix}v_1\\\vdots\\v_d\end{bmatrix}
=u_1v_1+\cdots+u_dv_d
=\vec u\cdot\vec v.$$

The shapes are $(1\times d)(d\times1)$, so the product is a $1\times1$ matrix. We usually identify this with the scalar it contains and write

$$\boxed{\vec u\cdot\vec v=\vec u^T\vec v.}$$

For example, if

$$\vec u=\begin{bmatrix}1\\2\\-1\end{bmatrix},\qquad
\vec v=\begin{bmatrix}3\\0\\4\end{bmatrix},$$

then $\vec u^T\vec v=1(3)+2(0)+(-1)(4)=-1$, exactly as with the dot product. In particular,

$$\vec u^T\vec u=\|\vec u\|^2.$$

### Reversing the shapes

What about $\vec u\vec v^T$? Its shapes are $(d\times1)(1\times d)$, so it produces a **$d\times d$ matrix**, whose $(i,j)$ entry is $u_i v_j$. This is called an **outer product**. It is a different operation from the dot product!

For instance,

$$\begin{bmatrix}1\\2\end{bmatrix}\begin{bmatrix}3&4\end{bmatrix}
=\begin{bmatrix}3&4\\6&8\end{bmatrix},$$

while $\begin{bmatrix}1&2\end{bmatrix}\begin{bmatrix}3\\4\end{bmatrix}=11$. **Checking the shapes tells us which type of answer to expect.**

## Transposing a product

Here's the main algebraic identity for transposes:

:::{important} Transpose of a product
Whenever $AB$ is defined,

$$\boxed{(AB)^T=B^TA^T.}$$

Transpose each factor **and reverse their order**.
:::

### A concrete example

Use our earlier $A$ and let

$$B=\begin{bmatrix}2&0\\-1&1\end{bmatrix}.$$

Then

$$AB=\begin{bmatrix}0&2\\-3&3\\-3&1\end{bmatrix},\qquad
(AB)^T=\begin{bmatrix}0&-3&-3\\2&3&1\end{bmatrix}.$$

Transposing the factors first gives the same result:

$$B^TA^T
=\begin{bmatrix}2&-1\\0&1\end{bmatrix}
\begin{bmatrix}1&0&-1\\2&3&1\end{bmatrix}
=\begin{bmatrix}0&-3&-3\\2&3&1\end{bmatrix}.$$

The same rule extends to more factors: $(ABC)^T=C^TB^TA^T$.

## Interpreting $A^TA$ and $AA^T$

For any matrix $A$ of shape $n\times d$, both of these products are defined:

$$\underbrace{A^T}_{d\times n}\;\underbrace{A}_{n\times d}
=\underbrace{A^TA}_{d\times d},\qquad
\underbrace{A}_{n\times d}\;\underbrace{A^T}_{d\times n}
=\underbrace{AA^T}_{n\times n}.$$

**Both products are square, even when $A$ isn't.** What do their entries tell us?

### $A^TA$: Dot products of columns

Use our earlier matrix:

$$A=\begin{bmatrix}1&2\\0&3\\-1&1\end{bmatrix},\qquad A^T=\begin{bmatrix}1&0&-1\\2&3&1\end{bmatrix}.$$

Multiplying gives

$$A^TA=\begin{bmatrix}1&0&-1\\2&3&1\end{bmatrix}\begin{bmatrix}1&2\\0&3\\-1&1\end{bmatrix}=\begin{bmatrix}2&1\\1&14\end{bmatrix}.$$

Each entry is a dot product of two columns of $A$. For example, the top-right entry is the dot product of the first and second columns:

$$\begin{bmatrix}1\\0\\-1\end{bmatrix}\cdot\begin{bmatrix}2\\3\\1\end{bmatrix}=(1)(2)+(0)(3)+(-1)(1)=1.$$

The diagonal entries are the squared lengths of the columns: $1^2+0^2+(-1)^2=2$ and $2^2+3^2+1^2=14$. The two off-diagonal entries are equal because reversing the order of a dot product doesn't change its value.

### $AA^T$: Dot products of rows

Multiplying in the other order gives

$$AA^T=\begin{bmatrix}1&2\\0&3\\-1&1\end{bmatrix}\begin{bmatrix}1&0&-1\\2&3&1\end{bmatrix}=\begin{bmatrix}5&6&1\\6&9&3\\1&3&2\end{bmatrix}.$$

Now each entry is a dot product of two rows of $A$. For example, entry $(1,3)$ uses the first and third rows. Writing their entries as column vectors, this dot product is

$$\begin{bmatrix}1\\2\end{bmatrix}\cdot\begin{bmatrix}-1\\1\end{bmatrix}=(1)(-1)+(2)(1)=1.$$

The diagonal entries are the squared lengths of the rows: $1^2+2^2=5$, $0^2+3^2=9$, and $(-1)^2+1^2=2$.

These products are called **Gram matrices**. $A^TA$ is the Gram matrix of the columns; $AA^T$ is the Gram matrix of the rows. They usually have different shapes and contain different information.

As transformations, $A^TA$ applies $A$ first and then $A^T$, taking a vector in $\mathbb R^d$ back to $\mathbb R^d$. $AA^T$ applies them in the other order, taking a vector in $\mathbb R^n$ back to $\mathbb R^n$. Returning to the same set doesn't mean returning to the same vector! Our computed products are not identity matrices.

## Symmetric matrices

<!--
Look at $A^TA$ and $AA^T$ above. In each one, entries on opposite sides of the main diagonal agree. This isn't a coincidence: reversing the order of a dot product doesn't change its value, whether we use columns or rows.
-->

Some square matrices have matching entries on opposite sides of the main diagonal.

:::{note} Definition: Symmetric matrix
A square matrix $S$ is **symmetric** if

$$S^T=S.$$

Equivalently, $S_{ij}=S_{ji}$ for every $i,j$.
:::

For example, $\begin{bmatrix}2&-1\\-1&3\end{bmatrix}$ is symmetric, while $\begin{bmatrix}2&-1\\0&3\end{bmatrix}$ isn't. Every diagonal matrix is symmetric, including the identity matrix.

\end{aligned}$$

Notice that $A$ itself need not be square or symmetric. The way we form these two products guarantees their symmetry.

## Inverses and transposes

If $A$ is invertible, is $A^T$ also invertible? Yes, and its inverse comes from transposing $A^{-1}$:

$$\boxed{(A^T)^{-1}=(A^{-1})^T.}$$

To see why, transpose each of the two inverse identities. Since $I_n^T=I_n$,

$$\begin{aligned}
(AA^{-1})^T=I_n&\quad\Longrightarrow\quad (A^{-1})^TA^T=I_n,\\
(A^{-1}A)^T=I_n&\quad\Longrightarrow\quad A^T(A^{-1})^T=I_n.
\end{aligned}$$

These are precisely the identities that make $(A^{-1})^T$ the inverse of $A^T$. Here, transposing and inverting can be done in either order, even though the factors inside a product must still be reversed.

One consequence: **the inverse of an invertible symmetric matrix is also symmetric**. If $A^T=A$, the boxed identity gives $(A^{-1})^T=(A^T)^{-1}=A^{-1}$.

For a rotation, the inverse equals the transpose:

$$R_\theta^{-1}=R_{-\theta}=\begin{bmatrix}\cos\theta&\sin\theta\\-\sin\theta&\cos\theta\end{bmatrix}=R_\theta^T.$$

That isn't a general rule: for $H=\begin{bmatrix}2&0\\0&1\end{bmatrix}$, we have $H^T=H$, while $H^{-1}\ne H$.

### Combining inverses and transposes

Suppose $A$ is invertible and we'd like to solve

$$AX^T=C.$$

First, isolate $X^T$ by multiplying on the left:

$$X^T=A^{-1}C.$$

Then transpose both sides, reversing the order of the factors:

$$\boxed{X=(A^{-1}C)^T=C^T(A^{-1})^T.}$$

To check it, transpose our answer to get $X^T=A^{-1}C$ and substitute: $AX^T=AA^{-1}C=C$.
