# 6.1: Orthonormal bases and QR factorization

In [**Chapter 5.3**](../ch05/05-03.ipynb), we saw that a subspace can have many different bases. Our focus in Chapter 6 will be finding a special type of basis – an **orthonormal basis** – for a given subspace. Orthonormal bases are not new; we studied them in Chapter 2, and they are convenient to work with. We'll recap their usefulness as we go.


---

## Orthonormal bases

Let's start with the plane

$$P=\left\{\begin{bmatrix}x\\y\\z\end{bmatrix}\in\mathbb R^3:x+y-z=0\right\}.$$

This is a subspace: it contains $\vec0$, and adding vectors or taking scalar multiples preserves the equation $x+y-z=0$.

One basis of $P$ is

$$\vec v_1=\begin{bmatrix}1\\0\\1\end{bmatrix},\qquad
\vec v_2=\begin{bmatrix}1\\1\\2\end{bmatrix}.$$

Both vectors lie on the plane, and neither is a multiple of the other. They span the plane: if $z=x+y$, then

$$\begin{bmatrix}x\\y\\z\end{bmatrix}
=(x-y)\begin{bmatrix}1\\0\\1\end{bmatrix}
+y\begin{bmatrix}1\\1\\2\end{bmatrix}.$$

But these two basis vectors aren't perpendicular: their dot product is $3$. Let's find a basis that's more convenient for calculating lengths and coordinates.

:::{note} Definition: Orthonormal basis
An **orthonormal basis** of a subspace is a basis whose vectors are mutually perpendicular and each have length one. If the basis is $\vec u_1,\ldots,\vec u_d$, then

$$\vec u_i\cdot\vec u_i=1,\qquad
\vec u_i\cdot\vec u_j=0\quad\text{when }i\ne j.$$
:::

We'll use $\vec v_1,\vec v_2,\ldots$ for the original basis vectors and $\vec u_1,\vec u_2,\ldots$ for the orthonormal ones.

## The Gram-Schmidt process

Return to our basis of the plane $x+y-z=0$:

$$\vec v_1=\begin{bmatrix}1\\0\\1\end{bmatrix},\qquad
\vec v_2=\begin{bmatrix}1\\1\\2\end{bmatrix}.$$

We'll keep the direction of the first vector and adjust the second to make it perpendicular to the first.

- **Normalize the first vector:** Dividing a nonzero vector by its length is called **normalizing** it. Since $\|\vec v_1\|=\sqrt2$, set

  $$\vec u_1=\frac{\vec v_1}{\|\vec v_1\|}
  =\frac1{\sqrt2}\begin{bmatrix}1\\0\\1\end{bmatrix}.$$

  This vector has length one and points in the same direction as $\vec v_1$.

- **Subtract the projection:** Let $\ell=\operatorname{span}(\vec u_1)$. Recall from [Chapter 2.7](../ch02/02-07.ipynb) that $\vec v_2$ splits into a component along $\ell$ and a component perpendicular to $\ell$:

  $$\vec v_2=
  \underbrace{\operatorname{proj}_{\ell}(\vec v_2)}_{\text{along }\ell}
  +\underbrace{\bigl(\vec v_2-\operatorname{proj}_{\ell}(\vec v_2)\bigr)}_{\text{perpendicular to }\ell}.$$

  Since $\vec u_1$ has length one, the projection formula simplifies to

  $$\operatorname{proj}_{\ell}(\vec v_2)
  =(\vec v_2\cdot\vec u_1)\vec u_1
  =\frac3{\sqrt2}\frac1{\sqrt2}\begin{bmatrix}1\\0\\1\end{bmatrix}
  =\frac32\begin{bmatrix}1\\0\\1\end{bmatrix}.$$

  We want the perpendicular component, so subtract this projection:

  $$\vec w_2=\vec v_2-\operatorname{proj}_{\ell}(\vec v_2)
  =\begin{bmatrix}1\\1\\2\end{bmatrix}
  -\frac32\begin{bmatrix}1\\0\\1\end{bmatrix}
  =\frac12\begin{bmatrix}-1\\2\\1\end{bmatrix}.$$

- **Normalize the remaining vector:** The remaining vector is perpendicular to $\vec u_1$, but it doesn't yet have length one. Its length is

  $$\|\vec w_2\|=\sqrt{\frac14+1+\frac14}=\frac{\sqrt6}{2}.$$

  Dividing by this length gives

  $$\vec u_2=\frac{\vec w_2}{\|\vec w_2\|}
  =\frac1{\sqrt6}\begin{bmatrix}-1\\2\\1\end{bmatrix}.$$

Let's check:

$$\vec u_1\cdot\vec u_2=\frac{-1+0+1}{\sqrt2\sqrt6}=0,
\qquad
\|\vec u_1\|^2=\frac{1+0+1}{2}=1,
\qquad
\|\vec u_2\|^2=\frac{1+4+1}{6}=1.$$

The span hasn't changed. Each new vector is a combination of the old ones, and we can recover the old vectors by undoing the subtraction and rescaling. Thus $\vec u_1,\vec u_2$ form an orthonormal basis of $P$.

This is the idea behind the **Gram-Schmidt process**: subtract the projections along the directions we've already found, then normalize what remains.

## Coordinates

Let's use our new basis to describe another vector on the plane:

$$\vec b=\begin{bmatrix}3\\2\\5\end{bmatrix}.$$

It lies on the plane $x+y-z=0$ because $3+2-5=0$. We want coefficients $c_1,c_2$ such that

$$\vec b=c_1\vec u_1+c_2\vec u_2.$$

We could solve a system of equations, but orthonormality gives us a quicker way. Take the dot product with $\vec u_1$:

$$\vec b\cdot\vec u_1
=c_1(\vec u_1\cdot\vec u_1)+c_2(\vec u_2\cdot\vec u_1)
=c_1.$$

The first dot product is $1$ and the second is $0$, so only $c_1$ remains. Taking the dot product with $\vec u_2$ similarly gives $c_2$. Therefore,

$$c_1=\frac{3+0+5}{\sqrt2}=4\sqrt2,
\qquad
c_2=\frac{-3+4+5}{\sqrt6}=\sqrt6.$$

Substituting the vectors back in checks our answer:

$$4\sqrt2\vec u_1+\sqrt6\vec u_2
=4\begin{bmatrix}1\\0\\1\end{bmatrix}
+\begin{bmatrix}-1\\2\\1\end{bmatrix}
=\begin{bmatrix}3\\2\\5\end{bmatrix}=\vec b.$$

:::{important} Finding coefficients using dot products
If $\vec u_1,\ldots,\vec u_d$ form an orthonormal basis of $V$ and $\vec b\in V$, then

$$\vec b=(\vec b\cdot\vec u_1)\vec u_1+\cdots+(\vec b\cdot\vec u_d)\vec u_d.$$

The coefficient of each basis vector is just a dot product!
:::

## QR factorization

Return to our worked example with $\vec v_1=\begin{bmatrix}1\\0\\1\end{bmatrix}$ and $\vec v_2=\begin{bmatrix}1\\1\\2\end{bmatrix}$. We now have two bases of the same plane. How are they related?

For the first vector, we divided by $\sqrt2$. For the second, we subtracted $\frac3{\sqrt2}\vec u_1$ and divided the remainder by $\frac{\sqrt6}2$. Undoing these steps gives

$$\vec v_1=\sqrt2\vec u_1,\qquad
\vec v_2=\frac3{\sqrt2}\vec u_1+\frac{\sqrt6}2\vec u_2.$$

Put the original vectors in the columns of $A$ and the orthonormal vectors in the columns of $Q$. These two equations become one matrix equation:

$$\underbrace{\begin{bmatrix}1&1\\0&1\\1&2\end{bmatrix}}_A
=\underbrace{\begin{bmatrix}
1/\sqrt2&-1/\sqrt6\\
0&2/\sqrt6\\
1/\sqrt2&1/\sqrt6
\end{bmatrix}}_Q
\underbrace{\begin{bmatrix}\sqrt2&3/\sqrt2\\0&\sqrt6/2\end{bmatrix}}_R.$$

Remember how matrix multiplication works column by column. The first column of $QR$ is $\sqrt2\vec u_1+0\vec u_2=\vec v_1$. The second is $\frac3{\sqrt2}\vec u_1+\frac{\sqrt6}2\vec u_2=\vec v_2$.

:::{note} Definition: QR factorization
A **QR factorization** of a matrix $A$ of shape $n\times d$ writes

$$A=QR,$$

where $Q$ has shape $n\times d$ with orthonormal columns, and $R$ has shape $d\times d$ and is **upper triangular**: all entries below its diagonal are zero.
:::

In our example, $Q$ has shape $3\times2$ and $R$ has shape $2\times2$. Each column of $R$ records the coefficients for one original basis vector. The zero appears because recovering $\vec v_1$ only requires $\vec u_1$; no $\vec u_2$ is needed.

We can start with a matrix instead of a named subspace. If the columns of $A$ are linearly independent, they form a basis of $\operatorname{colsp}(A)$. Applying the Gram-Schmidt process gives an orthonormal basis of that same column space. Those vectors become the columns of $Q$, and expressing the original columns using them gives $R$.

For example, let's find a QR factorization of

$$A=\begin{bmatrix}3&-1\\4&7\end{bmatrix}.$$

Its columns $\vec v_1=\begin{bmatrix}3\\4\end{bmatrix}$ and $\vec v_2=\begin{bmatrix}-1\\7\end{bmatrix}$ aren't multiples of one another, so they form a basis of $\mathbb R^2$.

- **Normalize the first vector:** The first column has length $5$, giving

  $$\vec u_1=\frac15\begin{bmatrix}3\\4\end{bmatrix}.$$

- **Subtract the projection:** For the second column, calculate the projection coefficient and subtract the projection:

  $$\vec v_2\cdot\vec u_1=\frac{-3+28}{5}=5,
  \qquad
  \vec w_2=\vec v_2-5\vec u_1
  =\begin{bmatrix}-1\\7\end{bmatrix}-\begin{bmatrix}3\\4\end{bmatrix}
  =\begin{bmatrix}-4\\3\end{bmatrix}.$$

- **Normalize the remaining vector:** The remaining vector has length $5$. We still need to divide by that length:

  $$\vec u_2=\frac15\begin{bmatrix}-4\\3\end{bmatrix}.$$

Both vectors now have length one, and their dot product is $(-12+12)/25=0$.

We don't need a new calculation to find $R$. Put back what we subtracted and record the lengths:

$$\vec v_1=5\vec u_1,\qquad \vec v_2=5\vec u_1+5\vec u_2.$$

Therefore,

$$\begin{bmatrix}3&-1\\4&7\end{bmatrix}
=\underbrace{\begin{bmatrix}3/5&-4/5\\4/5&3/5\end{bmatrix}}_Q
\underbrace{\begin{bmatrix}5&5\\0&5\end{bmatrix}}_R.$$

Here $Q$ is square: its two columns span all of $\mathbb R^2$. In our plane example, $Q$ had two columns with three entries each, spanning a plane inside $\mathbb R^3$.

## The general process

Suppose we start with three linearly independent vectors $\vec v_1,\vec v_2,\vec v_3$ in $\mathbb R^n$. At each step, we'll show the **current original vector in blue** and the **previously constructed unit vectors in orange**.

- **First vector:** Divide by its length:

  $$\vec u_1=\frac{\textcolor{#3d81f6}{\vec v_1}}{\|\textcolor{#3d81f6}{\vec v_1}\|}.$$

- **Second vector:** Subtract its projection onto the line along $\vec u_1$, then normalize:

  $$\vec w_2=\textcolor{#3d81f6}{\vec v_2}-(\textcolor{#3d81f6}{\vec v_2}\cdot\textcolor{orange}{\vec u_1})\textcolor{orange}{\vec u_1},
  \qquad \vec u_2=\frac{\vec w_2}{\|\vec w_2\|}.$$

- **Third vector:** We now need a vector perpendicular to **both** $\vec u_1$ and $\vec u_2$. Subtract both projections:

  $$\vec w_3=\textcolor{#3d81f6}{\vec v_3}-(\textcolor{#3d81f6}{\vec v_3}\cdot\textcolor{orange}{\vec u_1})\textcolor{orange}{\vec u_1}-(\textcolor{#3d81f6}{\vec v_3}\cdot\textcolor{orange}{\vec u_2})\textcolor{orange}{\vec u_2}.$$

  Because $\vec u_1$ and $\vec u_2$ are perpendicular, subtracting a component along one doesn't change the component along the other. For example,

  $$\vec w_3\cdot\vec u_1
  =\vec v_3\cdot\vec u_1-(\vec v_3\cdot\vec u_1)(1)-(\vec v_3\cdot\vec u_2)(0)=0.$$

  Similarly, $\vec w_3\cdot\vec u_2=0$. Now normalize:

  $$\vec u_3=\frac{\vec w_3}{\|\vec w_3\|}.$$

Why can we divide by these lengths? If $\vec w_3$ were zero, then $\vec v_3$ would be a combination of $\vec u_1,\vec u_2$, and hence of $\vec v_1,\vec v_2$. That would contradict the independence of the original vectors. The same reasoning ensures $\vec w_2\ne\vec0$.

The same procedure works for any number of linearly independent vectors.

:::{note} Definition: Gram-Schmidt process
Given $d$ linearly independent vectors $\vec v_1,\ldots,\vec v_d$ in $\mathbb R^n$, the **Gram-Schmidt process** constructs an orthonormal basis $\vec u_1,\ldots,\vec u_d$ of their span.

For each $i=1,\ldots,d$, use the current vector $\textcolor{#3d81f6}{\vec v_i}$ and the previously constructed unit vectors $\textcolor{orange}{\vec u_1,\ldots,\vec u_{i-1}}$:

- **Subtract the projections:** Remove the component along each previous unit vector:

  $$\vec w_i=\textcolor{#3d81f6}{\vec v_i}-(\textcolor{#3d81f6}{\vec v_i}\cdot\textcolor{orange}{\vec u_1})\textcolor{orange}{\vec u_1}-\cdots-(\textcolor{#3d81f6}{\vec v_i}\cdot\textcolor{orange}{\vec u_{i-1}})\textcolor{orange}{\vec u_{i-1}}.$$

- **Normalize:** Divide the remaining vector by its length:

  $$\vec u_i=\frac{\vec w_i}{\|\vec w_i\|}.$$

When $i=1$, there are no projections to subtract, so $\vec w_1=\vec v_1$. Linear independence guarantees that every $\vec w_i$ is nonzero.
:::

Each new unit vector is a combination of the original vectors, and putting back the projections expresses each original vector using the unit vectors. Their spans therefore agree. To form $R$, put the projection coefficients above the diagonal and the lengths of the remaining vectors on the diagonal.


Let's apply the procedure to

$$A=\begin{bmatrix}1&2&3\\1&0&1\\1&1&-1\end{bmatrix}.$$

Call its columns $\vec v_1,\vec v_2,\vec v_3$, in that order.

- **First column:** Its length is $\sqrt3$, so

  $$\vec u_1=\frac1{\sqrt3}\begin{bmatrix}1\\1\\1\end{bmatrix}.$$

- **Second column:** Find its component along $\textcolor{orange}{\vec u_1}$ and subtract:

  $$\textcolor{#3d81f6}{\vec v_2}\cdot\textcolor{orange}{\vec u_1}=\frac{2+0+1}{\sqrt3}=\sqrt3,\qquad
  \vec w_2=\begin{bmatrix}2\\0\\1\end{bmatrix}-\begin{bmatrix}1\\1\\1\end{bmatrix}
  =\begin{bmatrix}1\\-1\\0\end{bmatrix}.$$

  The remaining vector has length $\sqrt2$, so

  $$\vec u_2=\frac1{\sqrt2}\begin{bmatrix}1\\-1\\0\end{bmatrix}.$$

- **Third column:** This time, we need two coefficients:

  $$\textcolor{#3d81f6}{\vec v_3}\cdot\textcolor{orange}{\vec u_1}=\frac{3+1-1}{\sqrt3}=\sqrt3,\qquad
  \textcolor{#3d81f6}{\vec v_3}\cdot\textcolor{orange}{\vec u_2}=\frac{3-1+0}{\sqrt2}=\sqrt2.$$

  Subtract both components:

  $$\vec w_3=\begin{bmatrix}3\\1\\-1\end{bmatrix}
  -\begin{bmatrix}1\\1\\1\end{bmatrix}
  -\begin{bmatrix}1\\-1\\0\end{bmatrix}
  =\begin{bmatrix}1\\1\\-2\end{bmatrix}.$$

  The remaining vector has length $\sqrt6$, so

  $$\vec u_3=\frac1{\sqrt6}\begin{bmatrix}1\\1\\-2\end{bmatrix}.$$

The three unnormalized directions have pairwise dot products $1-1=0$, $1+1-2=0$, and $1-1=0$. After normalization, they form an orthonormal basis of $\mathbb R^3$.

We have already calculated every entry of $R$:

$$A=
\underbrace{\begin{bmatrix}
1/\sqrt3&1/\sqrt2&1/\sqrt6\\
1/\sqrt3&-1/\sqrt2&1/\sqrt6\\
1/\sqrt3&0&-2/\sqrt6
\end{bmatrix}}_Q
\underbrace{\begin{bmatrix}\sqrt3&\sqrt3&\sqrt3\\0&\sqrt2&\sqrt2\\0&0&\sqrt6\end{bmatrix}}_R.$$

For example, the last column says

$$\vec v_3=\sqrt3\vec u_1+\sqrt2\vec u_2+\sqrt6\vec u_3,$$

which puts back the two projections and the remaining vector.

## Existence and uniqueness

There is one simple choice we could change: reverse the sign of a column of $Q$ and reverse the sign of the corresponding row of $R$. The product stays the same. For example,

$$\begin{bmatrix}3&-1\\4&7\end{bmatrix}
=\begin{bmatrix}3/5&-4/5\\4/5&3/5\end{bmatrix}\begin{bmatrix}5&5\\0&5\end{bmatrix}
=\begin{bmatrix}-3/5&-4/5\\-4/5&3/5\end{bmatrix}\begin{bmatrix}-5&-5\\0&5\end{bmatrix}.$$

Both choices of $Q$ have orthonormal columns. But the second $R$ has a negative diagonal entry. The Gram-Schmidt process gives **positive** diagonal entries because they are lengths of nonzero vectors. Requiring those entries to be positive removes the ambiguity.

:::{important} Existence and uniqueness of QR factorization
Let $A$ be a real matrix of shape $n\times d$ whose columns are linearly independent. There is a unique factorization $A=QR$ in which:

- $Q$ has shape $n\times d$ and has orthonormal columns.
- $R$ has shape $d\times d$, is upper triangular, and has positive diagonal entries.

The columns of $Q$ form an orthonormal basis of $\operatorname{colsp}(A)$.
:::

To see why the factorization is unique, start with its first column. Since $R$ is upper triangular with positive diagonal entries, $\vec v_1$ must be a positive multiple of the unit vector $\vec u_1$. This forces

$$\vec u_1=\frac{\vec v_1}{\|\vec v_1\|}.$$

For the next column, orthonormality forces the coefficient along $\vec u_1$ to be $\vec v_2\cdot\vec u_1$. Subtracting this component leaves a positive multiple of $\vec u_2$, so

$$\vec w_2=\vec v_2-(\vec v_2\cdot\vec u_1)\vec u_1,
\qquad \vec u_2=\frac{\vec w_2}{\|\vec w_2\|}.$$

The same argument continues for each column: dot products determine its components along the previous unit vectors, and normalizing what remains determines the next unit vector. Independence ensures that the remaining vector is nonzero. Once $Q$ is fixed, the coefficients in $R$ are fixed as well. Every factorization satisfying the theorem is forced to follow the steps of the Gram-Schmidt process!

This does **not** mean that a subspace has only one orthonormal basis. Choosing a different starting basis, or changing its order, can change the orthonormal basis we get. Uniqueness applies once we fix $A$, including its column order, and require positive diagonal entries in $R$.
