# 6.1: Orthonormal Bases and QR Factorization

Last time, we conclude by briefly mentioning how orthonormal bases will be useful in the future? The natural question to ask following this discussion is: How do we _find_ orthonormal bases of subspaces?

We will often be given a basis of subspaces -- for instance, the column space of matrix $A$ has a spanning set given by its column vectors, and by removing the redundant column vectors we obtain a basis. The question then turns to the following: How do we _convert_ an arbitrary basis into an orthonormal basis?

This question has a very pleasant, geometric answer which has many useful applications and interpretations in linear algebra.

## Projections and orthogonal decomposition

Before we jump into answering this question, we need some warm-up. Suppose we are given vectors $\vec v$ and $\vec w$. The key idea will turn out to be decomposing $\vec v$ into a sum of two vectors, one which is parallel to $\vec w$ and one which is orthogonal to $\vec w$. In other words, we want to decompose $\vec v$ as the sum:
$$ \vec v = c\vec w + \vec v_{\perp}, \tag{$*$}$$
where $c$ is some scalar and $\vec v_{\perp}$ is orthogonal to $\vec w$. This is called the **orthogonal decomposition** of $\vec v$ with respect to $\vec w$.

<!-- Picture or orthogonal decomposition of v wrt. w -->

An orthogonal decomposition always exists, granted that $\vec w$ is nonzero, and can be found explicitly in terms of $\vec v$ and $\vec w$. Let us derive it now: The only thing to be determined in ($*$) is the scalar $c$, so that
$$ \vec v_{\perp} = \vec v - c\vec w$$
is orthogonal to $\vec w$. Orthogonality can be detected by the dot product:
$$ 0 = \vec w \cdot \vec v_{\perp} = \vec w \cdot (\vec v - c\vec w) = \vec w \cdot \vec v - c(\vec w\cdot \vec w) $$
Solving for $c$, we obtain:
$$ c = \frac{\vec w\cdot \vec v}{\vec w \cdot \vec w}.$$
Hence in the orthogonal decomposition of $\vec v$ with respect to $\vec w$, the component parallel to $\vec w$ is equal to
$$ c\vec w = \frac{\vec w\cdot \vec v}{\vec w \cdot \vec w}\vec w $$
and is called the **projection** of $\vec v$ onto $\vec w$. We often denote it as $\operatorname{proj}_{\vec w}{\vec v}$, so
$$ \operatorname{proj}_{\vec w}{\vec v} = \frac{\vec w\cdot \vec v}{\vec w \cdot \vec w}\vec w.$$

The terminology makes sense here since we are essentially 'projecting' the vector $\vec v$ onto vector $\vec w$ to find the component parallel to $\vec w$.

The component orthogonal to $\vec w$ is
$$ \vec v_{\perp} = \vec v - \operatorname{proj}_{\vec w}{\vec v} = \vec v - \frac{\vec w\cdot \vec v}{\vec w \cdot \vec w}\vec w.$$

One thing to note is that the projection is another natural way to motivate the concept of dot products. Indeed, suppose that the vector $\vec w$ we're projecting onto is a unit vector. Then the projection becomes
$$ \operatorname{proj}_{\vec w}{\vec v} = \frac{\vec w\cdot \vec v}{\vec w \cdot \vec w}\vec w = (\vec w\cdot \vec v) \vec w.$$
Note further that the length of this projection is precisely $\vec w\cdot \vec v$. This matches again with our previous interpretation of the dot product as measuring `how aligned' two vectors are.


### Projections with respect to an orthogonal basis

We have just seen the concept of an orthogonal decomposition of the vector $\vec v$ with respect to another vector $\vec w$. For our purposes, we will need a slightly more sophisticated version of this: the orthogonal decomposition of $\vec v$ with respect to an orthogonal basis $\vec w_1,\ldots,\vec w_k$.

Let us focus on the simple case of decomposing $\vec v$ with respect to the orthogonal basis $\vec w_1,\vec w_2$. The orthogonal decomposition we are looking for is of the form:
$$ \vec v = c_1 \vec w_1 + c_2 \vec w_2 + \vec v_{\perp} $$
for scalars $c_1,c_2$ and a vector $\vec v_{\perp}$ perpendicular to both $\vec w_1$ and $\vec w_2$. Note that all three vectors in the summand of the right hand side are mutually orthogonal to each other.

We can detect orthogonality by taking dot products. Taking the dot product of both sides with $\vec w_1$, we obtain
$$ \vec v \cdot \vec w_1 = c_1 (\vec w_1 \cdot \vec w_1)$$
and so we find the value of $c_1$ as
$$ c_1 = \frac{\vec v \cdot \vec w_1}{\vec w_1 \cdot \vec w_1}.$$
Similarly, taking the dot product of each side by $\vec w_2$, we can find the value of $c_2$ as
$$ c_2 = \frac{\vec v \cdot \vec w_2}{\vec w_2 \cdot \vec w_2}.$$

Therefore the orthogonal decomposition becomes
$$
\vec v = \frac{\vec v \cdot \vec w_1}{\vec w_1 \cdot \vec w_1}\vec w_1 + \frac{\vec v \cdot \vec w_2}{\vec w_2 \cdot \vec w_2} \vec w_2 + \vec v_{\perp},$$
which is precisely equal to
$$ \vec v = \operatorname{proj}_{\vec w_1}{\vec v} + \operatorname{proj}_{\vec w_2}{\vec v} + \vec v_{\perp}. $$


This can be generalized to orthogonal bases of any size. Indeed, if we have an orthogonal basis $\vec w_1, \cdots, \vec w_k$ of some subspace, we can orthogonally decompose $\vec v$ with respect to this basis as
$$ \vec v = \operatorname{proj}_{\vec w_1}{\vec v} + \operatorname{proj}_{\vec w_2}{\vec v} + \cdots + \operatorname{proj}_{\vec w_k}{\vec v} + \vec v_{\perp}.$$

## Gram-Schmidt process

The Gram-Schmidt process takes a basis $S = \left\{ \vec v_1, \ldots, \vec v_p\right\}$ of some subspace and converts it into an orthonormal basis of the same subspace. Let us demonstrate how the process works through some examples.

### Example 1. Basis of $\mathbb{R}^2$

Consider the following basis of $\mathbb{R}^2$:
$$ \vec v_1 = \begin{bmatrix} 2 \\ 1 \end{bmatrix}, \vec v_2 = \begin{bmatrix} 1 \\ 2 \end{bmatrix}.$$
Note that this is indeed a basis of $\mathbb{R}^2$ since it spans the whole space and there are no redundant vectors.

The idea is the first create an _orthogonal_ basis $\vec w_1, \vec w_2$ that spans the same subspace as $S$, and then scale the vectors to be unit size to form an orthonormal basis. 

We first take
$$\vec w_1 = \vec v_1 = \begin{bmatrix} 2 \\ 1 \end{bmatrix},$$
and then take the orthogonal decomposition of $\vec v_2$ with respect to $\vec w_1$ and let $\vec w_2$ be the orthogonal component:
$$ \vec v_2 = \operatorname{proj}_{\vec w_1}{\vec v_2} + \vec w_2.$$
This guarantees that $\vec w_2$ is orthogonal to $\vec w_1$. Let us calculate the projection:
$$ \operatorname{proj}_{\vec w_1}{\vec v_2} = \frac{\vec v_2 \cdot \vec w_1}{\vec w_1 \cdot \vec w_1} \vec w_1 = \frac{4}{5}\vec w_1 = \frac{4}{5}\begin{bmatrix} 2 \\ 1 \end{bmatrix}.$$
From here we obtain
$$\vec w_2 = \vec v_2 - \operatorname{proj}_{\vec w_1}{\vec v_2} = \begin{bmatrix} 1 \\ 2 \end{bmatrix} - \frac{4}{5}\begin{bmatrix} 2 \\ 1 \end{bmatrix} = \frac{3}{5}\begin{bmatrix} -1 \\ 2 \end{bmatrix}.$$

Pictorially, this is what is happening:
<!-- Picture of basis v_1=w_1, v_2 and the orthogonal decomposition giving w_2 -->

We finish by scaling down the vectors $\vec w_1,\vec w_2$ to be unit length:
$$ \| \vec w_1\| = \sqrt{5},\ \| \vec w_2 \| = \frac{4\sqrt{5}}{5};$$
$$ \vec u_1 = \vec w_1 / \| \vec w_1 \| = \frac{1}{\sqrt{5}}\begin{bmatrix} 2 \\ 1 \end{bmatrix},\ \vec u_2 = \vec w_2 / \|\vec w_2\| = \frac{1}{\sqrt{5}}\begin{bmatrix} -1 \\ 2 \end{bmatrix}.$$

Hence we obtain an orthonormal basis $\vec u_1, \vec u_2$ for $\mathbb{R}^2$.



### Example 2. Basis of $\mathbb{R}^3$

Consider the following basis of $\mathbb{R}^3$:
$$ \vec v_1 = \begin{bmatrix} 1 \\ 1 \\ 1 \end{bmatrix},\qquad \vec v_2 = \begin{bmatrix} 2 \\ 0 \\ 1 \end{bmatrix},\qquad \vec v_3 = \begin{bmatrix} 3 \\ 1 \\ -1 \end{bmatrix}.$$

We first take
$$\vec w_1 = \vec v_1 = \begin{bmatrix} 1 \\ 1 \\ 1 \end{bmatrix}.$$
Next, we take the orthogonal decomposition of $\vec v_2$ with respect to $\vec w_1$ and let $\vec w_2$ be the orthogonal component. The projection is
$$ \operatorname{proj}_{\vec w_1}{\vec v_2} = \frac{\vec v_2 \cdot \vec w_1}{\vec w_1 \cdot \vec w_1}\vec w_1 = \frac{3}{3}\vec w_1 = \vec w_1.$$
Therefore,
$$\vec w_2 = \vec v_2 - \operatorname{proj}_{\vec w_1}{\vec v_2} = \begin{bmatrix} 2 \\ 0 \\ 1 \end{bmatrix} - \begin{bmatrix} 1 \\ 1 \\ 1 \end{bmatrix} = \begin{bmatrix} 1 \\ -1 \\ 0 \end{bmatrix}.$$

For the third vector, we take the orthogonal decomposition of $\vec v_3$ with respect to the orthogonal basis $\vec w_1,\vec w_2$:
$$ \vec v_3 = \operatorname{proj}_{\vec w_1}{\vec v_3} + \operatorname{proj}_{\vec w_2}{\vec v_3} + \vec w_3.$$
The two projections are
$$ \operatorname{proj}_{\vec w_1}{\vec v_3} = \frac{3}{3}\vec w_1 = \vec w_1,\qquad \operatorname{proj}_{\vec w_2}{\vec v_3} = \frac{2}{2}\vec w_2 = \vec w_2.$$
It follows that
$$\vec w_3 = \vec v_3 - \operatorname{proj}_{\vec w_1}{\vec v_3} - \operatorname{proj}_{\vec w_2}{\vec v_3} = \begin{bmatrix} 3 \\ 1 \\ -1 \end{bmatrix} - \begin{bmatrix} 1 \\ 1 \\ 1 \end{bmatrix} - \begin{bmatrix} 1 \\ -1 \\ 0 \end{bmatrix} = \begin{bmatrix} 1 \\ 1 \\ -2 \end{bmatrix}.$$

We finish by scaling the vectors $\vec w_1,\vec w_2,\vec w_3$ to be unit length. Since
$$ \|\vec w_1\|=\sqrt{3},\qquad \|\vec w_2\|=\sqrt{2},\qquad \|\vec w_3\|=\sqrt{6},$$
we obtain
$$ \vec u_1 = \frac{1}{\sqrt{3}}\begin{bmatrix} 1 \\ 1 \\ 1 \end{bmatrix},\qquad \vec u_2 = \frac{1}{\sqrt{2}}\begin{bmatrix} 1 \\ -1 \\ 0 \end{bmatrix},\qquad \vec u_3 = \frac{1}{\sqrt{6}}\begin{bmatrix} 1 \\ 1 \\ -2 \end{bmatrix}.$$
Hence $\vec u_1,\vec u_2,\vec u_3$ form an orthonormal basis of $\mathbb{R}^3$.

### Example 3. Basis of a hyperplane in $\mathbb{R}^3$

Gram-Schmidt does not only apply to bases of the entire ambient space. It applies equally well to a basis of any subspace. Consider the hyperplane
$$ H = \left\{\begin{bmatrix} x \\ y \\ z \end{bmatrix} \in \mathbb{R}^3 : x+y+z=0\right\}.$$
One basis of $H$ is
$$ \vec v_1 = \begin{bmatrix} 1 \\ -1 \\ 0 \end{bmatrix},\qquad \vec v_2 = \begin{bmatrix} 1 \\ 0 \\ -1 \end{bmatrix}.$$

As before, we begin by taking
$$ \vec w_1 = \vec v_1 = \begin{bmatrix} 1 \\ -1 \\ 0 \end{bmatrix}.$$
We then subtract from $\vec v_2$ its projection onto $\vec w_1$:
$$ \operatorname{proj}_{\vec w_1}{\vec v_2} = \frac{\vec v_2\cdot \vec w_1}{\vec w_1\cdot \vec w_1}\vec w_1 = \frac{1}{2}\begin{bmatrix} 1 \\ -1 \\ 0 \end{bmatrix}.$$
This gives the second orthogonal vector
$$ \vec w_2 = \vec v_2 - \operatorname{proj}_{\vec w_1}{\vec v_2} = \begin{bmatrix} 1 \\ 0 \\ -1 \end{bmatrix} - \frac{1}{2}\begin{bmatrix} 1 \\ -1 \\ 0 \end{bmatrix} = \frac{1}{2}\begin{bmatrix} 1 \\ 1 \\ -2 \end{bmatrix}.$$

Finally, we normalize the two vectors:
$$ \vec u_1 = \frac{1}{\sqrt{2}}\begin{bmatrix} 1 \\ -1 \\ 0 \end{bmatrix},\qquad \vec u_2 = \frac{1}{\sqrt{6}}\begin{bmatrix} 1 \\ 1 \\ -2 \end{bmatrix}.$$
Hence $\vec u_1,\vec u_2$ form an orthonormal basis of the hyperplane $H$.

The Gram-Schmidt process preserves the subspace spanned by the original basis. Here $\vec w_1=\vec v_1$ and $\vec w_2$ is a linear combination of $\vec v_1,\vec v_2$, while $\vec v_2=\operatorname{proj}_{\vec w_1}{\vec v_2}+\vec w_2$ is a linear combination of $\vec w_1,\vec w_2$. Therefore
$$ \langle \vec u_1,\vec u_2\rangle = \langle\vec w_1,\vec w_2\rangle = \langle \vec v_1,\vec v_2\rangle = H.$$

## The QR decomposition

The Gram-Schmidt process is useful in not only its original purpose of converting bases of subspaces to orthogonal bases, but also in decomposing matrices. By analyzing the Gram-Schmidt process, we will obtain the **QR decomposition** of matrices. This decomposition is similar to the LU decomposition in that it express the matrix as a product of two matrices with nice properties, and helps us understand the structure of the matrix (in particular the column space) much better.

For simplicity, let us focus on the basis $\vec v_1, \ldots, \vec v_n$ of $\mathbb{R}^n$. Note that $\mathbb{R}^n$ has dimension $n$, so any basis of it must have size $n$. We discuss by examples how applying the Gram-Schmidt process gives us a decomposition of the matrix
$$ A = \begin{bmatrix} \vec v_1 & \cdots & \vec v_n \end{bmatrix},$$
where the column vectors are the basis elements.

### Example 1 revisited
In Example 1, we started with the basis $\vec v_1, \vec v_2$ of $\mathbb{R}^2$. Going back through our calculations in the Gram-Schmidt process, we see that
$$ \vec w_1 = \vec v_1;$$
$$ \vec w_2 = \vec v_2 - \frac{4}{5}\vec w_1.$$
Re-organizing everything so that the original basis elements come to the left hand side, we get
$$ \vec v_1 = \vec w_1;$$
$$ \vec v_2 = \frac{4}{5}\vec w_1 + \vec w_2.$$

Recall also that we scaled the orthogonal basis $\vec w_1, \vec w_2$ to form an orthonormal basis $\vec u_1, \vec u_2$ as follows:
$$ \vec u_1 = \vec w_1 / \| \vec w_1 \| = \frac{1}{\sqrt{5}}\vec w_1,\quad \vec u_2 = \vec w_2 / \|\vec w_2\| = \frac{5}{4\sqrt{5}}\vec w_2.$$
Re-organizing this in terms of the $\vec w_i$'s, we get
$$ \vec w_1 = \sqrt{5}\vec u_1,\quad \vec w_2 = \frac{4\sqrt{5}}{5}\vec u_2.$$
Substituting these into the previous equations, we obtain:
$$ \vec v_1 = \sqrt{5}\vec u_1;$$
$$ \vec v_2 = 4\vec u_1 + \frac{4\sqrt{5}}{5}\vec u_2.$$

This can be rephrased in terms of matrix multiplication as
$$ \begin{bmatrix} \vec v_1  & \vec v_2 \end{bmatrix} = \begin{bmatrix} \vec u_1 & \vec u_2 \end{bmatrix} \begin{bmatrix} \sqrt{5} & 4 \\ 0 &  4\sqrt{5}/5\end{bmatrix}.$$

Let us name the two matrices on the right hand side as $Q$ and $R$. This gives us the decomposition:
$$ A = QR,$$
where $R$ is an upper triangular matrix, and $Q$ is a square matrix whose column vectors form an orthonormal basis of $\mathbb{R}^2$. Writing down the entries, we have a decomposition
$$ \begin{bmatrix} 2  & 1 \\ 1 & 2 \end{bmatrix} = \begin{bmatrix} 2/\sqrt{5} & -1/\sqrt{5} \\ 1/\sqrt{5} & 2/\sqrt{5} \end{bmatrix} \begin{bmatrix} \sqrt{5} & 4 \\ 0 & 4\sqrt{5}/5\end{bmatrix} $$
which can be checked by hand.

### Example 2 revisited

In Example 2, we started with the basis $\vec v_1,\vec v_2,\vec v_3$ of $\mathbb{R}^3$. Going back through our calculations in the Gram-Schmidt process, we have
$$ \vec w_1 = \vec v_1;$$
$$ \vec w_2 = \vec v_2 - \vec w_1;$$
$$ \vec w_3 = \vec v_3 - \vec w_1 - \vec w_2.$$
Re-organizing these equations so that the original basis elements appear on the left hand side, we get
$$ \vec v_1 = \vec w_1;$$
$$ \vec v_2 = \vec w_1 + \vec w_2;$$
$$ \vec v_3 = \vec w_1 + \vec w_2 + \vec w_3.$$

Recall that we scaled the orthogonal basis $\vec w_1,\vec w_2,\vec w_3$ to form the orthonormal basis $\vec u_1,\vec u_2,\vec u_3$. Re-organizing the normalization equations in terms of the $\vec w_i$'s gives
$$ \vec w_1=\sqrt{3}\vec u_1,\qquad \vec w_2=\sqrt{2}\vec u_2,\qquad \vec w_3=\sqrt{6}\vec u_3.$$
Substituting these into the previous equations, we obtain
$$ \vec v_1 = \sqrt{3}\vec u_1;$$
$$ \vec v_2 = \sqrt{3}\vec u_1 + \sqrt{2}\vec u_2;$$
$$ \vec v_3 = \sqrt{3}\vec u_1 + \sqrt{2}\vec u_2 + \sqrt{6}\vec u_3.$$

This can be rephrased in terms of matrix multiplication as
$$ \begin{bmatrix} \vec v_1 & \vec v_2 & \vec v_3 \end{bmatrix} = \begin{bmatrix} \vec u_1 & \vec u_2 & \vec u_3 \end{bmatrix}\begin{bmatrix} \sqrt{3} & \sqrt{3} & \sqrt{3} \\ 0 & \sqrt{2} & \sqrt{2} \\ 0 & 0 & \sqrt{6} \end{bmatrix}.$$
Thus we obtain the QR decomposition $A=QR$, where
$$ Q=\begin{bmatrix} 1/\sqrt{3} & 1/\sqrt{2} & 1/\sqrt{6} \\ 1/\sqrt{3} & -1/\sqrt{2} & 1/\sqrt{6} \\ 1/\sqrt{3} & 0 & -2/\sqrt{6} \end{bmatrix},\qquad R=\begin{bmatrix} \sqrt{3} & \sqrt{3} & \sqrt{3} \\ 0 & \sqrt{2} & \sqrt{2} \\ 0 & 0 & \sqrt{6} \end{bmatrix}.$$
Writing out $A$, the decomposition is
$$ \begin{bmatrix} 1 & 2 & 3 \\ 1 & 0 & 1 \\ 1 & 1 & -1 \end{bmatrix} = \begin{bmatrix} 1/\sqrt{3} & 1/\sqrt{2} & 1/\sqrt{6} \\ 1/\sqrt{3} & -1/\sqrt{2} & 1/\sqrt{6} \\ 1/\sqrt{3} & 0 & -2/\sqrt{6} \end{bmatrix}\begin{bmatrix} \sqrt{3} & \sqrt{3} & \sqrt{3} \\ 0 & \sqrt{2} & \sqrt{2} \\ 0 & 0 & \sqrt{6} \end{bmatrix}.$$
Note again that the columns of $Q$ are orthonormal, and that $R$ is upper triangular.

In general, whenever we are given an $n\times n$ square matrix $A$ whose columns form a basis of $\mathbb{R}^n$, we can always find the QR decomposition
$$ A = QR,$$
where the columns of $Q$ are orthonormal and $R$ is upper triangular. This can be done precisely by running the Gram-Schmidt process on the column vectors of matrix $A$. The new orthonormal basis vectors form the columns of $Q$, and the entries of $R$ can be read off from the calculation as we have seen from the examples above.

In fact, looking back at our derivation of the QR decomposition, we realize that there is no need to focus on the case of square matrices $A$. One can equally carry out this derivation for a basis $\vec v_1,\ldots,\vec v_m$ of some subspace in $\mathbb{R}^n$ to obtain the decomposition
$$ A = QR,$$
where $A$ is an $m\times n$ matrix whose columns $\vec v_1,\ldots,\vec v_m$ form a basis of some subspace (in particular, they are linearly independent), the $m\times n$ matrix $Q$ are the orthonormal basis vectors obtained by Gram-Schmidt, and $R$ is an $n\times n$ upper triangular matrix. For instance, carrying out the same analysis for Example 3 gives us the QR decomposition:

$$ \begin{bmatrix} 1 & 1 \\ -1 & 0 \\ 0 & -1 \end{bmatrix} = \begin{bmatrix} 1/\sqrt{2} & 1/\sqrt{6} \\ -1/\sqrt{2} & 1/\sqrt{6} \\ 0 & -2/\sqrt{6} \end{bmatrix}\begin{bmatrix} \sqrt{2} & 1/\sqrt{2} \\ 0 & \sqrt{6}/2 \end{bmatrix}. \tag{$*$}$$
Here $A$ and $Q$ are $3\times 2$ matrices, while $R$ is a $2\times 2$ upper triangular matrix. The columns of $Q$ form the orthonormal basis of the hyperplane obtained in Example 3.

## Interpretation of the QR decomposition

How does one interpret the QR decomposition of a matrix $A$? Let us view the column vectors of $A$ as a basis of the column space of $A$ -- we are again stipulating that the column vectors of $A$ are linearly independent.

From the viewpoint of the Gram-Schmidt process, we are simply replacing the basis with an orthonormal basis provided by the column vectors of $Q$. In other words, _the columns of $Q$ form an orthonormal basis of the column space of_ $A$.

Recall from the last chapter that one reason we consider orthonormal bases is that they provide a nice coordinate system. Let's try to demonstrate this with Example 3 and the associated QR decomposition obtained in ($*$).

The column vectors $\vec v_1, \vec v_2$ of the original matrix
$$ \begin{bmatrix} 1 & 1 \\ -1 & 0 \\ 0 & -1 \end{bmatrix} $$
form a basis of the hyperplane $H$. Let us take the following two vectors on the hyperplane:
$$ \vec a = \begin{bmatrix} 2 \\ -1 \\ -1\end{bmatrix} = \vec v_1 + \vec v_2, $$
$$ \vec b = \begin{bmatrix} 2 \\ -2 \\ 0 \end{bmatrix} = 2\vec v_1.$$
We can think of the basis $\vec v_1, \vec v_2$ as providing the coordinates $(1,1)$ and $(2,0)$ to the vectors $\vec a$ and $\vec b$. But one has to be careful here since this coordinate system does not behave well with the dot product, namely
$$ (1,1) \cdot (2,0) = 2 \neq 6 = \vec a \cdot \vec b.$$

This sort of bad behavior disappears once we convert to the orthonormal basis $\vec u_1,\vec u_2$ provided by the columns of $Q$. How do we express $\vec a$ as a combination of those two vectors? Note that by the interpretation of matrix-vector multiplication as linear combinations of the column vectors, we can write
$$ \vec a = \vec v_1 + \vec v_2 = \begin{bmatrix} 1 & 1 \\ -1 & 0 \\ 0 & -1 \end{bmatrix} \begin{bmatrix} 1 \\ 1 \end{bmatrix}.$$
Now substituting the QR decomposition of the matrix $A$, we have
$$ \vec a = \begin{bmatrix} 1/\sqrt{2} & 1/\sqrt{6} \\ -1/\sqrt{2} & 1/\sqrt{6} \\ 0 & -2/\sqrt{6} \end{bmatrix}\begin{bmatrix} \sqrt{2} & 1/\sqrt{2} \\ 0 & \sqrt{6}/2 \end{bmatrix}\begin{bmatrix} 1 \\ 1 \end{bmatrix} = \begin{bmatrix} 1/\sqrt{2} & 1/\sqrt{6} \\ -1/\sqrt{2} & 1/\sqrt{6} \\ 0 & -2/\sqrt{6} \end{bmatrix}\begin{bmatrix} 3\sqrt{2}/2 \\ \sqrt{6}/2 \end{bmatrix} = \frac{3\sqrt 2}{2}\vec u_1 + \frac{\sqrt{6}}{2} \vec u_2.$$
Similarly, we can convert $\vec b$ by calculating as follows:
$$ \begin{bmatrix} \sqrt{2} & 1/\sqrt{2} \\ 0 & \sqrt{6}/2 \end{bmatrix}\begin{bmatrix} 2 \\ 0 \end{bmatrix} = \begin{bmatrix} 2\sqrt 2 \\ 0 \end{bmatrix},$$
$$ \vec b = 2\sqrt{2}\vec u_1.$$
Thus we can think of $\vec u_1, \vec u_2$ as providing coordinates $(\frac{3\sqrt 2}{2},\frac{\sqrt{6}}{2})$ and $(2\sqrt 2, 0)$ to $\vec a$ and $\vec b$. These coordinates are nicer since they behave well with the dot product. Indeed,
$$ (\frac{3\sqrt 2}{2},\frac{\sqrt{6}}{2}) \cdot (2\sqrt 2, 0) = 6 = \vec a \cdot \vec b.$$
We can also measure length using these coordinates:
$$ (\frac{3\sqrt 2}{2},\frac{\sqrt{6}}{2}) \cdot (\frac{3\sqrt 2}{2},\frac{\sqrt{6}}{2}) = \frac{9}{2} + \frac{3}{2} = 6 = \vec a \cdot \vec a$$
so $\|\vec a\| = 6$.


Essentially, we can think of the new basis $\vec u_1,\vec u_2$ for the hyperplane $H$ as being an analogue of the standard basis vectors $\vec e_1,\cdot, \vec e_n$ for $\mathbb{R}^n$, in that they provide a coordinate system where one can compute dot products in the usual way of multiplying coordinate-by-coordinate and adding. We will explore this viewpoint in more detail in the next lecture.
