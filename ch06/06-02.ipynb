# 6.2: Orthogonal Matrices and Orthogonal Transformations

Last time, we learned about the QR decomposition of an $m\times n$ matrix $A$:
$$ A = QR$$
where $Q$ is an $m\times n$ matrix with orthonormal columns and $R$ is an $n\times n$ upper triangular matrix.

Let us now narrow down to the case of when $A$ is square, in other words when $m=n$. In this case, the matrix $Q$ is an $n\times n$ matrix with orthonormal columns. Note that those column vectors form an orthonormal basis for $\mathbb{R}^n$, i.e. writing $Q$ as
$$ Q = \begin{bmatrix} \vec q_1 & \vec q_2 & \cdots & \vec q_n\end{bmatrix},$$
the vectors $\vec q_1, \ldots, \vec q_n$ form an orthonormal basis for $\mathbb{R}^n$. In this case, we say that $Q$ is an **orthogonal matrix**.

What are some examples of orthogonal matrices? Let us recall the QR decompositions we obtained last time:

$$ \begin{bmatrix} 2  & 1 \\ 1 & 2 \end{bmatrix} = \begin{bmatrix} 2/\sqrt{5} & -1/\sqrt{5} \\ 1/\sqrt{5} & 2/\sqrt{5} \end{bmatrix} \begin{bmatrix} \sqrt{5} & 4 \\ 0 & 4\sqrt{5}/5\end{bmatrix} $$


$$ \begin{bmatrix} 1 & 2 & 3 \\ 1 & 0 & 1 \\ 1 & 1 & -1 \end{bmatrix} = \begin{bmatrix} 1/\sqrt{3} & 1/\sqrt{2} & 1/\sqrt{6} \\ 1/\sqrt{3} & -1/\sqrt{2} & 1/\sqrt{6} \\ 1/\sqrt{3} & 0 & -2/\sqrt{6} \end{bmatrix}\begin{bmatrix} \sqrt{3} & \sqrt{3} & \sqrt{3} \\ 0 & \sqrt{2} & \sqrt{2} \\ 0 & 0 & \sqrt{6} \end{bmatrix}.$$

In both decompositions, the first matrix on the right hand side is an orthogonal matrix. Indeed, note that the column vectors of the $Q$ matrix in the first example:
$$ \vec q_1 = \begin{bmatrix} 2/\sqrt{5} \\ 1/\sqrt{5}\end{bmatrix}, \quad \vec q_2 = \begin{bmatrix}-1/\sqrt{5} \\ 2/\sqrt{5}\end{bmatrix}$$
are of unit length and are orthogonal to each other, hence forming an orthonormal basis for $\mathbb{R}^2$.

## Basic properties of orthogonal matrices

### The transpose is its inverse

Let us write down the column vectors of an orthogonal matrix $Q$ as follows:
$$ Q = \begin{bmatrix} \vec q_1 & \vec q_2 & \cdots & \vec q_n\end{bmatrix}.$$

Multiplying the transpose of $Q^T$ with $Q$, we obtain:
$$ Q^T Q = \begin{bmatrix} \vec q_1^T \\ \vec q_2^T \\ \vdots \\ \vec q_n^T\end{bmatrix}\begin{bmatrix} \vec q_1 & \vec q_2 & \cdots & \vec q_n\end{bmatrix} = \begin{bmatrix} \vec q_1 \cdot \vec q_1 & \vec q_1\cdot \vec q_2 & \cdots & \vec q_1\cdot \vec q_n \\ \vec q_2\cdot \vec q_1 & \vec q_2\cdot \vec q_2 & \cdots & \vec q_2\cdot \vec q_n \\ \vdots & \vdots & \ddots & \vdots \\ \vec q_n\cdot \vec q_1 & \vec q_n\cdot \vec q_2 & \cdots & \vec q_n\cdot \vec q_n\end{bmatrix}, \tag{$*$}$$
by recalling the interpretation of matrix multiplication as taking dot products of the row vectors on the left and the column vectors on the right.

But note that since the column vectors form an orthonormal basis of $\mathbb{R}^n$, we must have that
$$ \vec q_i \cdot \vec q_i = 1 $$
for all $i$, and that
$$ \vec q_i \cdot \vec q_i = 0 $$
for all distinct $i$ and $j$. Hence the right hand side of the equation above becomes
$$ Q^T Q = I_n,$$
where $I_n$ is the $n\times n$ identity matrix. Since $Q$ is a square matrix, this identity implies $Q$ is invertible and that the transpose $Q^T$ is equal to the inverse $Q^{-1}$:
$$ Q^T = Q^{-1}.$$

Moreover, the opposite is also true: If a square matrix $Q$ satisfies $Q^T = Q^{-1}$, then $Q$ is an orthogonal matrix. This is because the equation $Q^T Q = I_n$ can be rewritten as ($*$), which in turn implies that the column vectors satisfy
$$ \vec q_i \cdot \vec q_i = 1, \vec q_i \cdot \vec q_i = 0, $$
thus telling us that the columns form an orthonormal basis.

Let us check this property with our previous example:
$$ Q = \begin{bmatrix} 2/\sqrt{5} & -1/\sqrt{5} \\ 1/\sqrt{5} & 2/\sqrt{5} \end{bmatrix}.$$
Calculating by hand, we indeed see that:
$$ Q^T Q = \begin{bmatrix} 2/\sqrt{5} & 1/\sqrt{5} \\ -1/\sqrt{5} & 2/\sqrt{5} \end{bmatrix} \begin{bmatrix} 2/\sqrt{5} & -1/\sqrt{5} \\ 1/\sqrt{5} & 2/\sqrt{5} \end{bmatrix} = \begin{bmatrix} 1 & 0 \\ 0 & 1\end{bmatrix} = I_2.$$


### The row vectors also form an orthonormal basis

The property of matrix $Q$ being orthogonal can be characterized precisely by
$$ Q^T = Q^{-1}. $$
Note that taking the transpose on both sides, we obtain
$$ (Q^T)^T = (Q^{-1})^T = (Q^T)^{-1}.$$

Hence the matrix $Q^T$ is also orthogonal, since its transpose is equal to its inverse. In other words, the row vectors of an orthogonal matrix also form an orthonormal basis for $\mathbb{R}^n$.

Let us bring back our previous example again:
$$ Q = \begin{bmatrix} 2/\sqrt{5} & -1/\sqrt{5} \\ 1/\sqrt{5} & 2/\sqrt{5} \end{bmatrix}.$$

Note that the row vectors
$$ \begin{bmatrix} 2/\sqrt{5} \\ -1/\sqrt{5}\end{bmatrix}, \quad \begin{bmatrix} 1/\sqrt{5} \\ 2/\sqrt{5}\end{bmatrix}$$
are also of unit length and orthogonal to each other, hence forming an orthonormal basis of $\mathbb{R}^2$.

### A product of orthogonal matrices is orthogonal

Suppose that $Q_1$ and $Q_2$ are both orthogonal matrices of the same size. Then the product $Q_1 Q_2$ is also an orthogonal matrix. Indeed, note that
$$ (Q_1 Q_2)^T = Q_2^T Q_1^T = Q_2^{-1} Q_1^{-1} = (Q_1 Q_2)^{-1},$$
where we used the property that the transpose of a product is the product of the transposes in reverse order, and that the transpose of an orthogonal matrix is equal to its inverse.

## Orthogonal matrices as rigid transformations

Recall that matrices can also be viewed as linear transformations. From this viewpoint, orthogonal matrices admit a very nice interpretation as _rigid transformations_, in other words transformations that preserve both lengths and angles. Let us see why this is the case.

Lengths of vectors and angles between them are calculated in terms of the dot product. More concretely, recall that given two vectors $\vec x$ and $\vec y$, we calculate their lengths by
$$ \| \vec x \| = \sqrt{\vec x \cdot \vec x},\ \| \vec y \| = \sqrt{\vec y \cdot \vec y} $$
and the angle $\theta$ between them by
$$ \cos \theta = \frac{\vec x \cdot \vec y}{\| \vec x \| \| \vec y \|}.$$
Hence a matrix $A$ preserves lengths and angles if and ony if it does not change the dot product. On other words, we want that
$$ (A\vec x)\cdot (A\vec y) = \vec x \cdot \vec y$$
for all vectors $\vec x$ and $\vec y$. Recall that the dot product can be expressed in terms of matrix multiplication:
$$ (A\vec x)\cdot (A\vec y) = (A\vec x)^T (A\vec y) = \vec x^T A^T A \vec y.$$

Now suppose that our matrix $A$ is orthogonal, and let us re-write it as $A = Q$. Then we have
$$ (Q\vec x)\cdot (Q\vec y) = \vec x^T Q^T Q \vec y = \vec x^T I_n \vec y = \vec x^T \vec y = \vec x \cdot \vec y.$$
Hence we see that orthogonal matrices preserve lengths and angles.

### Example 1: A rotation in $\mathbb{R}^2$

As a first example, consider the linear transformation which rotates every vector counterclockwise by $90^\circ$. Let us denote its matrix by $U$:
$$ U = \begin{bmatrix} 0 & -1 \\ 1 & 0 \end{bmatrix}. $$
For instance, this rotation sends the vector $\vec x = \begin{bmatrix} 3 \\ -2 \end{bmatrix}$ to
$$ U\vec x = \begin{bmatrix} 0 & -1 \\ 1 & 0 \end{bmatrix}\begin{bmatrix} 3 \\ -2 \end{bmatrix} = \begin{bmatrix} 2 \\ 3 \end{bmatrix}. $$
Geometrically, the vector has turned by $90^\circ$, but its length has not changed: both $\vec x$ and $U\vec x$ have length $\sqrt{3^2+(-2)^2}=\sqrt{13}$.

Let us verify directly that $U$ is orthogonal. We have
$$ U^T U = \begin{bmatrix} 0 & 1 \\ -1 & 0 \end{bmatrix}\begin{bmatrix} 0 & -1 \\ 1 & 0 \end{bmatrix} = \begin{bmatrix} 1 & 0 \\ 0 & 1 \end{bmatrix} = I_2. $$
Thus $U$ is an orthogonal matrix, so the corresponding rotation is a rigid transformation. Notice that the calculation $U^T U=I_2$ verifies not merely the length preservation in our particular example, but the preservation of dot products, lengths, and angles for every pair of vectors in $\mathbb{R}^2$.

The matrix which we have been using throughout this section gives a slightly more complicated example:
$$ Q = \begin{bmatrix} 2/\sqrt{5} & -1/\sqrt{5} \\ 1/\sqrt{5} & 2/\sqrt{5} \end{bmatrix}. $$
This is also a rotation matrix. Recall that the matrix for counterclockwise rotation through an angle $\theta$ in $\mathbb{R}^2$ has the form
$$ \begin{bmatrix} \cos\theta & -\sin\theta \\ \sin\theta & \cos\theta \end{bmatrix}. $$
Our matrix $Q$ has exactly this form with $\cos\theta=2/\sqrt{5}$ and $\sin\theta=1/\sqrt{5}$. Thus $\theta=\tan^{-1}(1/2)$, which is approximately $26.6^\circ$, and $Q$ rotates every vector counterclockwise through this angle. The calculation $Q^TQ=I_2$ above confirms that this rotation is a rigid transformation.

### Example 2: A reflection in $\mathbb{R}^2$

As a second example, consider reflection across the line $y=x$. Reflecting across this line interchanges the two coordinates, so its matrix is
$$ S = \begin{bmatrix} 0 & 1 \\ 1 & 0 \end{bmatrix}. $$
For example, if $\vec x = \begin{bmatrix} 3 \\ -2 \end{bmatrix}$, then
$$ S\vec x = \begin{bmatrix} 0 & 1 \\ 1 & 0 \end{bmatrix}\begin{bmatrix} 3 \\ -2 \end{bmatrix} = \begin{bmatrix} -2 \\ 3 \end{bmatrix}. $$
The points represented by $\begin{bmatrix} 3 \\ -2 \end{bmatrix}$ and $\begin{bmatrix} -2 \\ 3 \end{bmatrix}$ are mirror images across $y=x$. Once again, both vectors have length $\sqrt{13}$.

The reflection matrix is symmetric, so $S^T=S$. Consequently,
$$ S^T S = \begin{bmatrix} 0 & 1 \\ 1 & 0 \end{bmatrix}\begin{bmatrix} 0 & 1 \\ 1 & 0 \end{bmatrix} = \begin{bmatrix} 1 & 0 \\ 0 & 1 \end{bmatrix} = I_2. $$
Therefore $S$ is also orthogonal, and reflection across the line $y=x$ is a rigid transformation. More generally, rotations and reflections in higher-dimensional spaces $\mathbb{R}^n$ are represented by orthogonal matrices as well, and hence they are also rigid transformations: they may change the position or orientation of a geometric object, but they do not change its lengths or angles.

### Example 3: The minus map

Consider the **minus map** which sends every vector to its negative:
$$ \vec x \text{ to } -\vec x. $$
In $\mathbb{R}^2$, this transformation has matrix $-I_2$. We can understand it geometrically as a composition of two familiar reflections. Reflection across the $x$-axis and reflection across the $y$-axis have matrices
$$ F_x=\begin{bmatrix} 1 & 0 \\ 0 & -1 \end{bmatrix}, \qquad F_y=\begin{bmatrix} -1 & 0 \\ 0 & 1 \end{bmatrix}, $$
respectively. Both matrices are orthogonal, since
$$ F_x^T F_x=I_2 \qquad \text{and} \qquad F_y^T F_y=I_2. $$
If we first reflect across the $x$-axis and then across the $y$-axis, the matrix of the resulting composition is
$$ F_yF_x=\begin{bmatrix} -1 & 0 \\ 0 & 1 \end{bmatrix}\begin{bmatrix} 1 & 0 \\ 0 & -1 \end{bmatrix}=\begin{bmatrix} -1 & 0 \\ 0 & -1 \end{bmatrix}=-I_2. $$
Thus the composition sends $\begin{bmatrix} x_1 \\ x_2 \end{bmatrix}$ to $\begin{bmatrix} -x_1 \\ -x_2 \end{bmatrix}$, exactly as the minus map does. We saw above that a product of orthogonal matrices is orthogonal. Equivalently, a composition of rigid linear transformations is again rigid. Since $F_x$ and $F_y$ are orthogonal, their product $-I_2$ is orthogonal as well.

This conclusion is also what we should expect geometrically. For example, the minus map sends $\begin{bmatrix} 3 \\ -2 \end{bmatrix}$ to $\begin{bmatrix} -3 \\ 2 \end{bmatrix}$, and both vectors have length $\sqrt{13}$. More generally, for any vectors $\vec x$ and $\vec y$,
$$ (-\vec x)\cdot(-\vec y)=\vec x\cdot\vec y. $$
Therefore the minus map preserves dot products, and hence preserves both lengths and angles. In $\mathbb{R}^n$ its matrix is $-I_n$, and we can also verify its orthogonality directly:
$$ (-I_n)^T(-I_n)=I_n. $$
In higher dimensions, $-I_n$ can similarly be obtained by composing the reflections which reverse the individual coordinate directions. Thus the minus map is an orthogonal map, and consequently a rigid transformation, in every dimension.

### A cautionary note

There needs to be some caution in thinking about orthogonal matrices as rigid transformations. The reason is very simple: as with any other linear transformation, an orthogonal matrix sends the origin to the origin. This means that orthogonal matrices cannot represent translations, which are also rigid in the sense that they preserve lengths and angles.
