# 8.2: Computing determinants

Suppose we want the determinant of $A=\begin{bmatrix}2&1&3\\0&-1&4\\1&0&2\end{bmatrix}$. We already know how to compute a $2\times2$ determinant. Let's turn this into a problem we know how to solve.


---

## Cofactor expansion

Pick a row or column. For each entry, cross out its row and column. The remaining entries form a smaller square matrix. Multiply its determinant by the entry you picked and the sign in that position in this checkerboard:

$$\begin{bmatrix}+&-&+\\-&+&-\\+&-&+\end{bmatrix}.$$

Add the contributions. This is called **cofactor expansion**. For our matrix, let's use the second row. The first entry is zero, so we only need two smaller determinants:

$$
\begin{aligned}
\det(A)
&=(-1)\det\begin{bmatrix}2&3\\1&2\end{bmatrix}
-4\det\begin{bmatrix}2&1\\1&0\end{bmatrix}\\
&=(-1)(4-3)-4(0-1)=3.
\end{aligned}
$$

Where did the minus sign in front of the $4$ come from? The $4$ is in row $2$, column $3$, a negative position in the checkerboard. Try the first column instead. The two contributions are $2(-2)$ and $1(7)$, giving $3$ again.

In general, write $A_{ij}$ for the matrix obtained by deleting row $i$ and column $j$. Expanding along row $i$ gives

$$
\det(A)=\sum_{j=1}^n a_{ij}(-1)^{i+j}\det(A_{ij}).
$$

We can expand along **any** row or column. Each step takes us down one size, until we reach $\det([a])=a$. Look for zeros when you choose where to expand. In an upper triangular matrix, the first column has only one possible nonzero entry, and deleting its row and column leaves another upper triangular matrix. Repeating gives the **product of the diagonal entries**. For a lower triangular matrix, the same argument starts with the first row.


---

## What happens when we change a matrix?

The determinant is **linear in each column separately**. For example, if we hold $\vec u$ fixed, then

$$
\det[\vec u\;\;s\vec v+t\vec w]
=s\det[\vec u\;\;\vec v]+t\det[\vec u\;\;\vec w].
$$

Swapping two columns reverses orientation, so it changes the sign. This means two equal columns give determinant zero: swapping them leaves the matrix unchanged and negates its determinant. Together these facts tell us what happens during elimination. The rules apply to rows too, since $\det(A^T)=\det(A)$.

| Operation on rows (or columns) | Effect on the determinant |
| --- | --- |
| Exchange two rows | Multiply by $-1$ |
| Multiply one row by $t$ | Multiply by $t$ |
| Add a multiple of one row to another | Leave it unchanged |

Why does the last rule work? Replace $\vec v$ by $\vec v+t\vec u$ in $[\vec u\;\;\vec v]$. Linearity gives an extra term of $t\det[\vec u\;\;\vec u]$, which is zero. Also notice that scaling an *entire* $n\times n$ matrix by $t$ scales each of its $n$ columns, so $\det(tA)=t^n\det(A)$.


---

## Products, invertibility, and LU

When we apply $B$ and then $A$, their volume scale factors multiply. Their signs multiply too: reversing orientation twice preserves it. For square matrices of the same size, **$\det(AB)=\det(A)\det(B)$**.

For example, an orthogonal matrix $Q$ satisfies $Q^TQ=I$. Taking determinants gives $\det(Q)^2=1$, so $\det(Q)$ is either $1$ or $-1$. And if $A$ is invertible, $AA^{-1}=I$ gives $\det(A)\det(A^{-1})=1$. In particular, $\det(A^{-1})=1/\det(A)$. If $A$ is singular, its columns span a lower-dimensional space, so the volume is zero. Together, these give another test for invertibility:

$$A\text{ is invertible}\quad\Longleftrightarrow\quad\det(A)\ne0.$$

Finally, suppose we've already factored $A$:

$$
A=LU=
\begin{bmatrix}1&0&0\\2&1&0\\-1&3&1\end{bmatrix}
\begin{bmatrix}2&1&-1\\0&-3&2\\0&0&4\end{bmatrix}.
$$

Can we find $\det(A)$ without multiplying out $LU$? Both factors are triangular. The diagonal of $L$ consists entirely of ones, so $\det(L)=1$. That leaves $\det(A)=\det(U)=2(-3)(4)=-24$.

If elimination takes $s$ row swaps, our factorization is $PA=LU$. Account for each swap with a factor of $-1$:

$$
\det(A)=(-1)^s u_{11}u_{22}\cdots u_{nn}.
$$

This brings us back to pivots. A pivot in every diagonal position gives a nonzero product. A zero row in echelon form gives a determinant of zero.
