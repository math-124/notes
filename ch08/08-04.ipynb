# 8.4: Properties of determinants

Suppose someone tells us a matrix has determinant $6$, but they don’t tell us its entries. How much can we figure out about the determinant after changing one of its columns? What if we multiply the matrix by itself? We can answer both questions without a single cofactor expansion.

---

## Changing one column at a time

The determinant is **linear in each column separately**, provided all the other columns are held fixed. This property is called *multilinearity*. For example,

$$\begin{bmatrix}|&|&|\\\vec u&\vec v&\vec w\\|&|&|\end{bmatrix}$$

has columns $\vec u$, $\vec v$, and $\vec w$, in that order. Here’s what happens when the first column is a linear combination:

$$
\begin{aligned}
&\det\begin{bmatrix}|&|&|\\s\vec u+t\vec v&\vec p&\vec q\\|&|&|\end{bmatrix}\\
&\quad=s\det\begin{bmatrix}|&|&|\\\vec u&\vec p&\vec q\\|&|&|\end{bmatrix}+t\det\begin{bmatrix}|&|&|\\\vec v&\vec p&\vec q\\|&|&|\end{bmatrix}.
\end{aligned}
$$

Why does that work? Expand along the column you’re changing. The cofactors only involve entries in the *other* columns. Holding those fixed makes the determinant a fixed linear combination of the entries in your column.

Here’s a numerical example. The first column below is $2\begin{bmatrix}1\\0\\2\end{bmatrix}+3\begin{bmatrix}2\\1\\1\end{bmatrix}$. Keeping the other two columns fixed gives

$$
\begin{vmatrix}8&1&0\\3&2&1\\7&0&3\end{vmatrix}
=2\begin{vmatrix}1&1&0\\0&2&1\\2&0&3\end{vmatrix}
+3\begin{vmatrix}2&1&0\\1&2&1\\1&0&3\end{vmatrix}
=2(8)+3(10)=46.
$$

The determinant is also **alternating**. If two columns are equal, the determinant is zero. We saw the geometric reason in two and three dimensions. Repeated sides give zero area or volume. Together with multilinearity, this explains why exchanging any two columns changes the sign. For example, start with two copies of $\vec u+\vec v$ and expand in both columns:

$$
\begin{aligned}
0&=\det\begin{bmatrix}|&|&|\\\vec u+\vec v&\vec u+\vec v&\vec w\\|&|&|\end{bmatrix}\\
&=\det\begin{bmatrix}|&|&|\\\vec u&\vec u&\vec w\\|&|&|\end{bmatrix}+\det\begin{bmatrix}|&|&|\\\vec u&\vec v&\vec w\\|&|&|\end{bmatrix}
\\
&\quad+\det\begin{bmatrix}|&|&|\\\vec v&\vec u&\vec w\\|&|&|\end{bmatrix}+\det\begin{bmatrix}|&|&|\\\vec v&\vec v&\vec w\\|&|&|\end{bmatrix}.
\end{aligned}
$$

The first and last terms vanish. The two remaining terms must be negatives of each other. What about adding a multiple of one column to another? For example,

$$
\begin{aligned}
&\det\begin{bmatrix}|&|&|\\\vec u&\vec v+t\vec u&\vec w\\|&|&|\end{bmatrix}\\
&\quad=\det\begin{bmatrix}|&|&|\\\vec u&\vec v&\vec w\\|&|&|\end{bmatrix}+t\det\begin{bmatrix}|&|&|\\\vec u&\vec u&\vec w\\|&|&|\end{bmatrix}.
\end{aligned}
$$

The extra term is zero, so the determinant stays the same. Here are the three operations we know from elimination. Each rule works for rows as well as columns.

| Operation | Effect on the determinant |
| --- | --- |
| Exchange two columns (or rows) | Changes the sign |
| Multiply one column (or row) by $t$ | Multiplies the determinant by $t$ |
| Add a multiple of one column (or row) to another | Leaves the determinant unchanged |

What if we multiply the **whole matrix** by $t$? For an $n\times n$ matrix, we’ve scaled $n$ columns, so $\det(tA)=t^n\det(A)$. This also shows why we can’t distribute a determinant over a matrix sum. For example, $\det(I_2+I_2)=4$, while $\det(I_2)+\det(I_2)=2$.

## Composition and transpose

Remember that $AB$ means “first apply $B$, then apply $A$.” Think about what happens to the unit cube. Applying $B$ changes its oriented volume from $1$ to $\det(B)$. Applying $A$ multiplies that oriented volume by $\det(A)$. In particular, reversing orientation twice preserves it. This gives a geometric way to remember the **product rule**.

:::{note} Products and transposes
For square matrices $A$ and $B$ of the same shape,
$$\det(AB)=\det(A)\det(B),\qquad \det(A^T)=\det(A).$$
:::

Why does transposing preserve the determinant? A row expansion for $A^T$ is the corresponding column expansion for $A$, with the smaller matrices transposed too. Repeat this down to $1\times1$ determinants and the calculations agree. This is also why our column-operation rules work for rows.

Recall from [Section 6.2](../ch06/06-02.ipynb) that an orthogonal matrix satisfies $Q^TQ=I$. Taking determinants gives $1=\det(Q^T)\det(Q)=\det(Q)^2$. So $\det(Q)$ is either $1$ or $-1$. Both preserve volume. The sign tells us whether $Q$ preserves or reverses orientation.

---

## When is a matrix invertible?

If $A$ is invertible, then $AA^{-1}=I$. By the product rule, $\det(A)\det(A^{-1})=1$. This tells us two things at once. The determinant of $A$ is nonzero, and the determinant of its inverse is $1/\det(A)$.

What about the other direction? If $A$ is singular, its columns are linearly dependent. Write one column as a linear combination of the others and expand using multilinearity in that column. Every term has a repeated column, so every term is zero. A zero column also gives zero directly. Geometrically, a singular transformation collapses space into a lower-dimensional subspace.

:::{note} Determinants and invertibility
A square matrix $A$ is invertible exactly when $\det(A)\ne0$. Equivalently, its columns form a basis. In this case, $\det(A^{-1})=1/\det(A)$.
:::

## Back to LU factorization

Cofactor expansion quickly becomes a lot of work for a large matrix. What if we already have an LU factorization? If $A=LU$, then $\det(A)=\det(L)\det(U)$. Our $L$ has ones on the diagonal, so $\det(L)=1$. All we need is the product of the diagonal entries of $U$.

If elimination uses row swaps, write the factorization as $PA=LU$. Starting from the identity, each swap changes the sign of the determinant, so $\det(P)=(-1)^s$ for $s$ swaps. Taking determinants of both sides gives

$$
\det(A)=(-1)^s u_{11}u_{22}\cdots u_{nn}.
$$

This brings us back to pivots. For an invertible square matrix, there’s a pivot in every diagonal position of $U$, so the product is nonzero. For a singular square matrix, the echelon form has a zero row, so the product is zero.

::::{tip} Activity 4: Determinants from LU
Suppose $PA=LU$, where
$$P=\begin{bmatrix}0&1&0\\1&0&0\\0&0&1\end{bmatrix},\quad
L=\begin{bmatrix}1&0&0\\-2&1&0\\3&-1&1\end{bmatrix},\quad
U=\begin{bmatrix}3&-2&5\\0&2&-4\\0&0&-2\end{bmatrix}.$$

**a)** Compute $\det(P)$, $\det(L)$, and $\det(U)$. Use $\det(PA)=\det(LU)$ to find $\det(A)$ without multiplying out the matrices. Is $A$ invertible?

**b)** Suppose the bottom-right entry of $U$ were changed to zero, with $P$ and $L$ unchanged. Would the corresponding $A$ be invertible?

:::{tip} Solution
:class: dropdown

**a)** $P$ swaps the first two rows, so $\det(P)=-1$. The triangular factors give $\det(L)=1$ and $\det(U)=3\cdot2\cdot(-2)=-12$. Thus $(-1)\det(A)=1\cdot(-12)$, so $\det(A)=12$ and $A$ is invertible.

**b)** Now the product of the diagonal entries of $U$ is zero. Since $\det(P)=-1$, the corresponding $A$ has determinant zero and is singular.
:::
