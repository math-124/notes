# 8.1: Determinants as Area and Volume Scale Factors

Determinants are a fundamental concept in linear algebra. Given a square matrix $A$, its determinant $\det A$ is some scalar number that wil depend on the entries of $A$. They not only determine whether a square matrix is invertible, but provide a lot of additional geometric information about the linear transformation represented by the matrix. We will not spend much time on how to actually calculate it for matrices bigger than $3 \times 3$ because it is rather complicated, but we will instead focus on its geometric meaning, properties, and applications. 

## Determinants of $2\times 2$ matrices

Given a $2\times 2$ matrix $A = \begin{bmatrix} a & b \\ c & d \end{bmatrix}$, the determinant of $A$ is defined as:
$$\det A = \det \begin{bmatrix} a & b \\ c& d \end{bmatrix} = ad - bc.
$$

Let's look at some examples.

### Examples

#### Example 1: A positive determinant

Let
$$A = \begin{bmatrix} 3 & 2 \\ 1 & 4 \end{bmatrix}. $$
Using $\det A = ad-bc$, we get
$$\det A = 3\cdot 4-2\cdot 1=12-2=10. $$

#### Example 2: A negative determinant

Let
$$B = \begin{bmatrix} 2 & 3 \\ 4 & 1 \end{bmatrix}. $$
Then
$$\det B = 2\cdot 1-3\cdot 4=2-12=-10. $$
A determinant doesn't always have to be a positive number; it can be positive, negative, or zero.

#### Example 3: An upper triangular matrix

Let
$$C = \begin{bmatrix} 5 & 1 \\ 0 & -2 \end{bmatrix}. $$
Then
$$\det C = 5\cdot(-2)-1\cdot 0=-10. $$
Notice that in general for an upper triangular $2\times 2$ matrix, the determinant is simply the product of its diagonal entries. This is true in general for any square upper/lower-triangular matrix; we will repeat this fact later.

#### Example 4: Parallel columns give determinant zero

Consider
$$D = \begin{bmatrix} 2 & 6 \\ -1 & -3 \end{bmatrix}. $$
The second column is three times the first column:
$$\begin{bmatrix} 6 \\ -3 \end{bmatrix}=3\begin{bmatrix} 2 \\ -1 \end{bmatrix}. $$
Thus, the two columns are parallel. Calculating directly,
$$\det D=2\cdot(-3)-6\cdot(-1)=-6+6=0. $$

This is a general fact: if the columns of a $2\times 2$ matrix are parallel, then the determinant is zero. What is even better is that the opposite is also true: if the determinant is zero, then the columns are parallel.

This is a special case of a more general fact that we will discuss later: the determinant of a square matrix is zero precisely when the columns are linearly dependent and thus the matrix is not invertible.

### Determinants as signed area scale factors

Before we move onto determinants for $3\times 3$ matrices, let's take a moment to interpret the determinant of a $2\times 2$ matrix geometrically. This will not only help us understand the meaning of the determinant formula for $3\times 3$ matrices much better, but also provide meaning for determinants of matrices of any size.

The determinant
$$ \det A = \det \begin{bmatrix} a & b \\ c & d \end{bmatrix} = ad - bc $$
is equal to the **signed area** of the parallelogram spanned by the two column vectors $ \begin{bmatrix} a \\ c \end{bmatrix},\begin{bmatrix} b \\ d \end{bmatrix}$ of $A$:

The bounding rectangle has area $(a+b)(c+d)$. The four triangles have total area $ac+bd$, and the two rectangles have total area $2bc$. Therefore, the area of the parallelogram is
$$
(a+b)(c+d)-(ac+bd+2bc)=ad-bc=\det A.
$$

### Orientation

Let's get back to what we mean by 'signed area'. Notice that the parallelogram drawn in the example matches the case of Example 1, where we were given the matrix 
$$ A = \begin{bmatrix} 3 & 2 \\ 1 & 4 \end{bmatrix}. $$

An observation here is that in Example 2, we were given the matrix:
$$ B = \begin{bmatrix} 2 & 3 \\ 4 & 1 \end{bmatrix},$$
which can be obtained from $A$ by simply swapping the two columns. The parallelogram spanned by the columns of $B$ is exactly the same one spanned by the columns of $A$ -- which means they must have the same area -- but the determinants differ by a sign:
$$ \det A = 10,\quad \det B = -10.$$

This is because the determinant also captures the **orientation** of the parallelogram. We say the ordered pair of vectors $\vec v_1, \vec v_2$ is _positively oriented_ if we can wrap our right hand around from $\vec v_1$ to $\vec v_2$ while keeping our thumb pointing up out of the paper. Otherwise, we say they are _negatively oriented_. In Example 1, the two column vectors of $A$, in their given order, is positively oriented. By swapping the order of the two vectors, the column vectors of $B$ become negatively oriented.

The determinant of a $2\times 2$ matrix is positive if the two column vectors are positively oriented, and negative if they are negatively oriented. This explains why the determinant is called the _signed area_: by its size it captures the area of the parallelogram, and by its sign it detects the orientation of the two vectors.


### Parallel columns give zero determinant

Another thing to note -- which we've already seen in Example 4 -- is that if the two column vectors are parallel, then the determinant is zero. This can be explained by the fact that the parallelogram spanned by the two vectors has zero area. But it can also be proven symbolically: if the two vectors are parallel, then our matrix must be of the form:
$$A = \begin{bmatrix} a & ka \\ c & kc \end{bmatrix}$$
for some scalar $k$. Calculating the determinant, we obtain
$$ \det A = a(kc)- (ka)c  = 0.$$

### Why 'scale factor'?

We've just described the determinant of a $2\times 2$ matrix as the signed area of the parallelogram spanned by its two column vectors. There is a more mature interpretation of the determinant that is based off this description as the **signed area scale factor**, which was what we alluded to in the title of this section.

Let us take the square matrix $A$ and view it as a linear transformation. Recall that the two column vectors can be viewed as the images $A\vec e_1, A\vec e_2$ of the standard basis vectors $\vec e_1, \vec e_2$ under the linear transformation. Since the parallelogram spanned by $\vec e_1, \vec e_2$ is the unit square of signed area 1, and the parallelogram spanned by $A\vec e_1, A\vec e_2$ has signed area $\det A$, we can view the determinant as the scale factor of the linear transformation $A$ on areas. In other words, if we take any shape in the plane and apply the linear transformation $A$ to it, the area of the resulting shape will be $\det A$ times the area of the original shape.

In this interpretation, the sign of the determinant detects whether the linear transformation changes the orientation of the shape or not. If $\det A$ is positive, then the orientation is preserved; if $\det A$ is negative, it is reserved.

An interesting fact is that this is true for any shape -- like triangles, ovals, or any blob -- not just parallelograms.

## Determinants of $3\times 3$ matrices

Suppose we are given the $3\times 3$ matrix
$$ A = \begin{bmatrix} a & b & c \\ d & e & f \\ g & h & i \end{bmatrix}. $$
The determinant of $A$ is defined by
$$ \begin{align*}\det A &= a \det \begin{bmatrix} e & f \\ h & i \end{bmatrix} - b \det \begin{bmatrix} d & f \\ g & i \end{bmatrix} + c \det \begin{bmatrix} d & e \\ g & h \end{bmatrix}\\
&= a(ei-fh) - b(di-fg) + c(dh-eg).
\end{align*}$$

A good way to remember this formula is to note that the coefficients of the three $2\times 2$ determinants are the entries of the first row of $A$ with alternating sign $(+,-,+)$, and that each of the $2\times 2$ determinants are obtained by deleting the row and column of the entry in question.





### Examples

#### Example 1: Expanding along a row with two zeros

Let
$$E=\begin{bmatrix} 2 & 0 & 0 \\ 1 & 3 & 1 \\ 0 & 2 & 4 \end{bmatrix}. $$
Because the first row contains two zeros, only its first entry contributes to the determinant:
$$\begin{align*}
\det E
&=2\det\begin{bmatrix}3&1\\2&4\end{bmatrix}\\
&=2\cdot(3\cdot4-1\cdot2)\\
&=2\cdot10=20.
\end{align*}$$


#### Example 2: An upper triangular matrix

Let
$$F=\begin{bmatrix} 2 & 1 & -1 \\ 0 & 3 & 2 \\ 0 & 0 & -2 \end{bmatrix}. $$
Expanding along the first row, we get
$$\begin{align*}
\det F
&=2\det\begin{bmatrix}3&2\\0&-2\end{bmatrix}
-1\det\begin{bmatrix}0&2\\0&-2\end{bmatrix}
+(-1)\det\begin{bmatrix}0&3\\0&0\end{bmatrix}\\
&=2\cdot(3\cdot(-2)-2\cdot0)
-1\cdot(0\cdot(-2)-2\cdot0)
+(-1)\cdot(0\cdot0-3\cdot0)\\
&= 2\cdot 3\cdot (-2) = -12.
\end{align*}$$
The last two cofactors are zero because of the zeros below the diagonal. Note that the determinant ends up being the product of the diagonal entries. This is something we have remarked before when discussing the determinant of upper/lower-triangular matrices.

#### Example 3: Linearly dependent rows give determinant zero

Consider
$$G=\begin{bmatrix} 1 & 2 & 0 \\ 2 & 4 & 0 \\ 0 & 1 & 1 \end{bmatrix}. $$
Note that the second row is twice the first row plus the third row, so the rows are linearly dependent. Calculating directly,
$$\begin{align*}
\det G
&=1\cdot(4\cdot1-0\cdot1)-2\cdot(2\cdot1-0\cdot0)+0\\
&=4-4=0.
\end{align*}$$
This is the $3\times3$ version of parallel columns giving determinant zero: whenever the columns of a square matrix are linearly dependent, its determinant is zero and the matrix is not invertible.

### Determinants as signed volume scale factors

Just as the determinant of $2\times 2$ matrices could be interpreted as the signed area of the parallelogram spanned by its two column vectors, the determinant of $3\times 3$ matrices turns out to be equal to the **signed volume** of the **parallelepiped** spanned by its three column vectors. A parallelepiped is a skewed three-dimensional cuboid, just like how a parallelogram is a skewed two-dimensional rectangle.

We won't be able to verify the fact that the volume of the parallelepiped is equal to the determinant as this is quite a bit more complicated than the $2\times 2$ case, but we will take it as a fact. What is worth talking about is what the _sign_ of the determinant means here.

For the $2\times 2$ case, the sign told us about the orientation of the two vectors. This is the same in the $3\times 3$ case: the sign of the determinant tells us about the orientation of the three column vectors of the matrix (in the order given by the matrix).

What is the orientation of three ordered vectors $\vec v_1, \vec v_2, \vec v_3$? We say they are _positively oriented_ if when we wrap our right hand around from $\vec v_1$ to $\vec v_2$, the thumb is pointing in the same direction as $\vec v_3$. Otherwise, we say they are _negatively oriented_. The determinant is positive if the three column vectors are positively oriented, and negative if they are negatively oriented.

Let's verify this in the very simple case of diagonal matrices. Suppose we are given
$$ A = \begin{bmatrix} 1 & 0 & 0 \\ 0 & 2 & 0 \\ 0 & 0 & 3 \end{bmatrix}. $$
Then the three column vectors are positively oriented, and the parallelepiped spanned by them is a rectangle cuboid of volume $1 \cdot 2 \cdot 3 = 6$, which is equal to the determinant of $A$. Now consider
$$ B = \begin{bmatrix} 0 & 0 & 1 \\ 0 & 2 & 0 \\ 3 & 0 & 0 \end{bmatrix}. $$
The columns defines the same rectangle cuboid, but the swapped orders make them negatively oriented. The determinant of $B$ when calculated is indeed $-6$.

Finally, the $3\times 3$ determinant can be interpreted as the **signed volume scale factor**, just as we did for the $2\times 2$ case. Under the linear transformation defined by $3\times 3$ matrix $A$, any shape -- a cube, sphere, or just about any blob -- gets its volume multiplied by $\det A$. The sign detects the change in orientation.

## Determinant and invertibility

Recall that a square matrix $A$ is invertible precisely when the columns of $A$ are linearly independent. We have seen from Example 4 of the $2\times 2$ case that when the two columns of a $2\times 2$ matrix are parallel (so linearly dependent), then the determinant is zero. This could be understood as the two column vectors spanning a degenerate parallelogram -- lying on some 1-dimensional line -- with zero area.

Similarly, we saw in Example 3 of the $3\times 3$ case that when the columns are linearly dependent -- in that example we had
$$ \vec v_2 = 2\vec v_1 + \vec v_3$$
-- then the determinant is zero. This can again be understood as the column vectors $\vec v_1, \vec v_2, \vec v_3$ spanning a degenerate parallelepiped of zero volume, since $\vec v_2$ lies on the plane spanned by $\vec v_1 $ and $\vec v_3$, hence forcing the parallelepiped to lie on that 2-dimensional plane.

These observations are true in bigger generality; if the column vectors of an $n\times n$ matrix $A$ are linearly dependent, then the $n$-dimensional parallelepiped spanned by the columns lie in some lower dimensional subspace and thus becomes degenerate. This implies that the volume is zero and therefore $\det A = 0$.

The opposite is true; if $\det A = 0$, this would mean that the $n$-dimensional parallelepiped has zero volume and hence lies on some lower-dimensional subspace. Since the $n$ column vectors of $A$ span a subspace that has dimension lower than $n$, this can only mean that the column vectors are linearly dependent.

Therefore we have the following remarkably simple equivalence: **A square matrix $A$ is invertible precisely when its column vectors are linearly independent, which in turn is precisely when $\det A\neq 0$**.

## Algebraic facts about determinants

So far, we've focused on the geometric aspects of determinants as the signed area/volume scale factor. In practice, we will need to know how to calculate the determinant given the entries of a square matrix. While we won't cover how this can be done in general  -- as this is complicated and not intuitive -- and use software instead, we will list some algebraic facts about determinants (without proof) that will be useful for us and motivate the various decompositions we've learned.

### Triangular matrices

If a square matrix $A$ is upper or lower-triangular, then $\det A$ is simply the multiple of all of its diagonal entries. We have seen this for various examples previously.

### Multiplicity

If $A$ and $B$ are square matrices of the same size, then the determinant is _multiplicative_, that is
$$ \det AB = \det A \det B.$$

### Relation with LU and QR decompositions

We can synthesize the two results with the LU and QR decompositions to efficiently calculate determinants. Suppose for an $n\times n$ matrix $A$ we are given the LU decomposition:
$$ A = LU.$$
Then by multiplicity, we have
$$ \det A = \det L \det U.$$
But since both $L$ and $U$ are triangular matrices, we can easily calculate the determinant by taking the product of their diagonal entries.

Suppose now we are given the QR decomposition:
$$ A = QR.$$
By multiplicity, we have
$$ \det A =\det Q \det R.$$
Here the interpretation of determinants as scaled volume scale factor helps us. An orthogonal matrix maps the unit vectors rigidly, i.e. by preserving lengths and angles. This means that it necessarily preserves volumes too. The only caveat is that it may reverse the orientation, which tells us that:
$$ \det Q = \pm 1.$$
Hence we know that
$$ \det A = \pm \det R,$$
and the determinant of $R$ can again be found by multiplying all of its diagonal entries.
