# 8.1: Determinants, area, and volume

We have already seen several ways to decide whether a square matrix is invertible. We can look at its columns: do they form a basis? We can try to solve a system. We can compute an LU factorization and examine the pivots.

Today we will attach a single number to a square matrix that answers the same question. This number is called its **determinant**. It also tells us something geometric: how the matrix changes areas and volumes.


---

## Start with a parallelogram

Take $A=\begin{bmatrix}a&b\\c&d\end{bmatrix}$. Recall that $A$ sends $\vec e_1$ and $\vec e_2$ to its first and second columns. What happens to the unit square? Its four corners land at the origin, $(a,c)$, $(b,d)$, and $(a+b,c+d)$. Connect them in the order you would walk around the square, and you get a parallelogram.

![The parallelogram with column vectors as its sides, inside a rectangle cut into the parallelogram, four triangles, and two smaller rectangles.](parallelogram-area.png)

How much area does the parallelogram take up? In the configuration above, the enclosing rectangle has width $a+b$ and height $c+d$. Subtract the pieces outside the parallelogram. There are two triangles of area $ac/2$, two of area $bd/2$, and two rectangles of area $bc$. This leaves

$$
(a+b)(c+d)-ac-bd-2bc=ad-bc.
$$

Now exchange the columns. The parallelogram stays put, but the calculation gives $bc-ad$. The **determinant** keeps track of the sign; taking its absolute value gives the area in either order:

$$
\det(A)=ad-bc, \qquad \text{area}=\lvert\det(A)\rvert.
$$

For example, take $A=\begin{bmatrix}3&1\\1&2\end{bmatrix}$. Its determinant is $3\cdot 2-1\cdot 1=5$. The unit square becomes a parallelogram of area $5$. More generally, this transformation multiplies the area of *any* region by $5$.


---

## What does the sign tell us?

Look at the two columns in the picture. The shorter turn from the first column to the second is counterclockwise, so the determinant is positive. Exchange the columns and the turn is clockwise, so the determinant is negative. This is what we mean by **signed area**. A rotation preserves area and orientation, so its determinant is $1$. A reflection preserves area and reverses orientation, so its determinant is $-1$.

What would a determinant of $0$ look like? Take $\begin{bmatrix}2&6\\1&3\end{bmatrix}$. The second column is three times the first. The entire parallelogram fits on a line! Its area is zero, just as the formula predicts: $2\cdot3-6\cdot1=0$. The same picture explains why a $2\times2$ matrix is invertible exactly when its determinant is nonzero: its columns must span the plane.


---

## From area to volume

In $\mathbb R^3$, the unit cube becomes a **parallelepiped** with the three columns $\vec v_1,\vec v_2,\vec v_3$ as edges. To find its volume, we need a base area and a height. The vector $\vec v_1\times\vec v_2$ gives us both a direction perpendicular to the base and a length equal to the base area. Taking its dot product with the third column multiplies that area by the signed height:

$$
\det(A)=(\vec v_1\times\vec v_2)\cdot\vec v_3.
$$

Try this with $\vec e_1,\vec e_2,\vec e_3$. Since $\vec e_1\times\vec e_2=\vec e_3$, the determinant is $1$. Replace $\vec e_3$ with $-\vec e_3$ and the volume is still $1$, but the determinant is $-1$. In general, its sign tells us which side of the base the third column points toward. Positive means the ordered basis is right-handed, negative means it is left-handed, and zero means the columns fit in a plane.

The same interpretation works in every dimension: $\lvert\det(A)\rvert$ is the factor by which $A$ scales volume. How can we find that factor from the entries of a larger matrix?
