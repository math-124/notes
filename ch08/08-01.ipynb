# 8.1: Determinants of 2 × 2 matrices

We’ve already seen several ways to decide whether a square matrix is invertible. We can look at its columns. Do they form a basis? We can try to solve a system, or we can compute a PLU factorization and examine the pivots.

In this chapter, we’ll attach a single number to a square matrix that answers the same question. This number is called its **determinant**. It also tells us something geometric. If we apply the matrix to a square, how does its area change?

---

## What’s a determinant?

Let’s start with the smallest possible square matrix. A $1\times1$ matrix has just one entry, and its determinant is that entry. For example, $\det([5])=5$ and $\det([-3])=-3$. For a $2\times2$ matrix, we need all four entries.

:::{note} Definition: Determinant of a $2\times2$ matrix
If $A=\begin{bmatrix}a&b\\c&d\end{bmatrix}$, its **determinant** is
$$\det(A)=ad-bc.$$
We also write $\begin{vmatrix}a&b\\c&d\end{vmatrix}$ for the same number. The vertical bars around a matrix mean “take its determinant.”
:::

For example, if $A=\begin{bmatrix}3&1\\1&2\end{bmatrix}$, then $\det(A)=3\cdot2-1\cdot1=5$. Where does the $5$ show up in a picture?

---

## What happens to the unit square?

The transformation $\vec x\mapsto A\vec x$ sends the standard basis vectors to the columns of $A$. Draw those columns from the origin. Where does the far corner of the unit square go? Since $A(\vec e_1+\vec e_2)=A\vec e_1+A\vec e_2$, it lands at the sum of the columns. Connecting the four corners gives a **parallelogram**.

![The blue parallelogram lies inside a rectangle. Its four surrounding triangles and two small rectangles are labeled with their areas.](parallelogram-area.png)

How much area does the blue parallelogram take up? For the configuration in the picture, enclose it in a rectangle of width $a+b$ and height $c+d$. Outside the parallelogram are two right triangles of area $ac/2$, two right triangles of area $bd/2$, and two small rectangles of area $bc$. Subtracting these pieces leaves

$$
\begin{aligned}
\text{area}&=(a+b)(c+d)-2\left(\tfrac12 ac\right)-2\left(\tfrac12 bd\right)-2bc\\
&=ad-bc.
\end{aligned}
$$

For our numerical example, the unit square goes from area $1$ to area $5$. Now exchange the columns. We get the *same parallelogram*, but the determinant becomes $1\cdot1-3\cdot2=-5$. Taking the absolute value recovers the area in either order.

:::{note} Area scaling
The parallelogram spanned by the columns of $A$ has area $|\det(A)|$. More generally, the transformation $\vec x\mapsto A\vec x$ multiplies the area of any region in the plane by $|\det(A)|$.
:::

---

## Why keep the sign?

Look again at the two columns in the picture. The shorter turn from the first column to the second is **counterclockwise**. Exchange the columns and the turn is **clockwise**. For two linearly independent columns, the first order gives a positive determinant and the second gives a negative determinant. This is what we mean by **oriented area**. A rotation preserves area and orientation, so its determinant is $1$. A reflection preserves area and reverses orientation, so its determinant is $-1$.

What would a determinant of zero look like? If the columns lie on one line, the parallelogram collapses to a line segment, or even a point. Conversely, two independent columns span a parallelogram of positive area. So a $2\times2$ matrix has nonzero determinant exactly when its columns are linearly independent, which is exactly when it is invertible.

::::{tip} Activity 1: Area and orientation
Let
$$A=\begin{bmatrix}4&-1\\1&3\end{bmatrix},\qquad
B=\begin{bmatrix}-1&4\\3&1\end{bmatrix},\qquad
C=\begin{bmatrix}2&6\\1&3\end{bmatrix}.$$

**a)** Compute $\det(A)$ and $\det(B)$.

**b)** Draw the parallelogram spanned by the columns of $A$ and find its area from your drawing. How does that area relate to the two determinants?

**c)** Compute $\det(C)$. What happens to its parallelogram? Are its columns linearly independent? Is $C$ invertible?

:::{tip} Solution
:class: dropdown

**a)** $\det(A)=4\cdot3-(-1)\cdot1=13$. Exchanging the columns gives $\det(B)=-13$.

**b)** Join $(0,0)$, $(4,1)$, $(3,4)$, and $(-1,3)$ in that order. The enclosing rectangle runs from $x=-1$ to $x=4$ and from $y=0$ to $y=4$. Cut off its four corners. Two triangles have legs $4$ and $1$, and two have legs $1$ and $3$. The area left is $5\cdot4-2(4\cdot1/2)-2(1\cdot3/2)=13$. Both column orders give this area.

**c)** $\det(C)=2\cdot3-6\cdot1=0$. Its second column is three times its first, so the parallelogram collapses to a segment. The columns are linearly dependent and $C$ is singular.
:::
