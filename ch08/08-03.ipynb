# 8.3: Determinants of square matrices

In [Section 8.2](08-02.ipynb), we found a $3\times3$ determinant by computing $2\times2$ determinants. What happens if we try the same rule on a $2\times2$ matrix? Expanding along the first row gives

$$\det\begin{bmatrix}a&b\\c&d\end{bmatrix}
=a\det([d])-b\det([c])=ad-bc.$$

There’s our original formula! Now we can run the idea in the other direction. To compute a $4\times4$ determinant, we use $3\times3$ determinants. Each expansion takes us down one size until we reach an entry we can read off.

## Reuse a smaller determinant

Here’s the optional expansion from Lecture 24. Look at the last column before doing any arithmetic. Which smaller determinant will we actually need?

$$E=\begin{bmatrix}
3&2&2&0\\-1&4&1&0\\2&3&-1&0\\5&-2&6&2
\end{bmatrix}.$$

The checkerboard signs down that column are $-,+,-,+$. The first three entries are zero. Deleting the last row and last column leaves exactly the matrix $D$ from Activity 2, so

$$
\det(E)=2\det\begin{bmatrix}3&2&2\\-1&4&1\\2&3&-1\end{bmatrix}
=2(-41)=-82.
$$

We didn’t use the $5,-2,6$ at all. Those entries were in the row we crossed out.

## Cofactor expansion in any dimension

Let $A$ have shape $n\times n$, and write $a_{ij}$ for its entry in row $i$, column $j$. Let $A_{ij}$ be the matrix obtained by deleting that row and column. Its determinant is called the **minor** of $a_{ij}$. How do we keep track of the checkerboard in a big matrix? The sign is $(-1)^{i+j}$, positive when $i+j$ is even and negative when it’s odd.

:::{note} Definition: Cofactor expansion
Start with $\det([a])=a$. For $n\geq2$, the **cofactor** of entry $a_{ij}$ is $C_{ij}=(-1)^{i+j}\det(A_{ij})$.

Choosing a row $i$ and moving across its columns gives
$$\det(A)=\sum_{j=1}^n a_{ij}C_{ij}.$$
Choosing a column $j$ and moving down its rows gives
$$\det(A)=\sum_{i=1}^n a_{ij}C_{ij}.$$
:::

This is a **recursive definition**. Each cofactor asks for the determinant of a smaller square matrix. We’ll use without proof the fact that all choices of row or column give the same answer. The geometric interpretation continues, too. The absolute value of the determinant is the factor by which the transformation scales $n$-dimensional volume.

---

## Triangular matrices are easy

Suppose $U$ is upper triangular. Expand along its first column. There’s only one possibly nonzero term, and the smaller matrix is still upper triangular.

$$
\det\begin{bmatrix}
u_{11}&*&*\\0&u_{22}&*\\0&0&u_{33}
\end{bmatrix}
=u_{11}\begin{vmatrix}u_{22}&*\\0&u_{33}\end{vmatrix}
=u_{11}u_{22}u_{33}.
$$

The same argument works in any dimension. For a lower triangular matrix, expand along the first **row** instead. Either way, **the determinant of a triangular matrix is the product of its diagonal entries**. For example,

$$
\det\begin{bmatrix}2&7&-4\\0&-3&5\\0&0&4\end{bmatrix}
=2(-3)(4)=-24.
$$

In particular, $\det(I_n)=1$, since the identity matrix is triangular and every diagonal entry is $1$. We’ll use this observation when we return to LU factorization in [Section 8.4](08-04.ipynb).
