# 4.3: PLU decompositions in general

## Non-square matrices

So far, we have focused on the case of linear equations
$$ A\vec x  = \vec b$$
where the coefficient matrix $A$ is square. This means that the number of equations is equal to the number of unknowns. However in many practical equations, the number of equations may not be equal to the number of unknowns. This corresponds to when the matrix $A$ is not square; an $m\times n$ matrix $A$ has $m$ equations and $n$ unknowns.

It turns out that we can still consider the PLU decomposition for non-square matrices $A$. In particular, for an $m\times n$ matrix $A$, we always have
$$ PA = LU$$
for $m\times m$ permutation matrix $P$, $m\times m$ lower triangular matrix $L$, and $m\times n$ upper triangular matrix $U$. This is obtained in the exact same way as for square matrices, by first swapping around the rows of $A$ and then performing row operations to reduce it to upper triangular form. The only difference is that the resulting $U$ matrix will be the same shape as $A$ and hence not necessarily square.

Let's look at one example. Suppose we are given
$$ A = \begin{bmatrix} 1 & 2 & -2 \\ 2 & 5 & -3 \end{bmatrix}.$$
To put $A$ in upper triangular form, we subtract 2 times the first row from the second row, which amounts to multiplying a lower triangular matrix on the left:
$$ \begin{bmatrix} 1 & 0 \\ -2 & 1 \end{bmatrix} \begin{bmatrix} 1 & 2 & -2 \\ 2 & 5 & -3 \end{bmatrix} = \begin{bmatrix} 1 & 2 & -2 \\ 0 & 1 & 1 \end{bmatrix}.$$
Multiplying the inverse of that lower triangular matrix on the left, we obtain the LU decomposition:
$$ A = \begin{bmatrix} 1 & 0 \\ 2 & 1 \end{bmatrix} \begin{bmatrix} 1 & 2 & -2 \\ 0 & 1 & 1 \end{bmatrix}.$$

## Pivot and free variables

We can similarly use the PLU decomposition $PA = LU$ to solve systems of equations $A\vec x = \vec b$ when $A$ is not square. Just as we did for the square case, let us first multiply both sides by $P$:
$$ PA\vec x = LU\vec x = P\vec b.$$
We set $U\vec x = \vec y$ and solve $L\vec y = P\vec b$ using forward-substitution. But something interesting (that we haven't seen for square matrices) happens when we try to solve the system $U\vec x = \vec y$ using back-substitution.

### Underdetermined systems

Let us first look at examples where $A$ has more columns than rows. This is when there are more variables than equations, and we call these systems **underdetermined**.

For the first example, suppose we are given $A$ as above, and consider the equation involving the upper-triangular matrix obtained from its PLU decomposition:
$$ \begin{bmatrix} 1& 2 & -2 \\ 0 & 1 & 1 \end{bmatrix} \begin{bmatrix} x \\ y \\ z \end{bmatrix} = \begin{bmatrix} 0 \\ 1 \end{bmatrix}.$$

For square matrices, what would've usually happened is that the last equation determines the value of $z$. But here the last equation is
$$ y+z = 1$$
which does not determine $z$. Instead all we can do here is express $y$ in terms of $z$:
$$ y = 1-z.$$
Carrying on, the equation on the first row is
$$x + 2y -2z = 0.$$
Substituting our expression $y = 1-z$, we obtain
$$ x + 2(1-z) - 2z = x -4z + 2 = 0,\quad x = 4z-2.$$
Thus the solutions are expressed as
$$x = 4z-2,\ y = 1-z.$$
We can check that for any value of $z$, the corresponding values of $x,y,z$ are all solutions to the system of equations. For instance, let us arbitrarily choose $z = 1$. Then the corresponding values are $x = 2,y=0$, and one can check that this is a valid solution. This is a case in which we have _infinitely many solutions_, where $x$ and $y$ are determined by $z$, while $z$ can be any value we want.

Let us look at one more example before discussing the general approach. Consider the system
$$ \begin{bmatrix} 1 & -2 & 2 & 1 \\ 0 & 0 & 1 & -1 \end{bmatrix}\begin{bmatrix} a \\ b\\ c\\ d\end{bmatrix} = \begin{bmatrix} 2 \\ -1\end{bmatrix}.$$
The second row gives us $c + d = -1$, which does not determine the value of the last variable $d$. The best we can do here is to express $c$ in terms of $d$ as $c = -d-1$.

The first row gives us $a - 2b + 2c + d = 2$. Substituting our previous expression, we get
$$ a - 2b + 2(-d-1) + d = a-2b-2d-2 = 2,$$
which allows us to express $a$ as:
$$ a = 2b + 2d + 4.$$

Therefore the solutions to the system are
$$ a = 2b + 2d + 4,\ c = -d-1.$$
Again we have infinitely many solutions, where $a$ and $c$ are determined by $b$ and $d$ which can be any value we want.


This is what happens for underdetermined systems $U\vec x = \vec y$ in general ($U$ is upper triangular with more columns than rows); the system will have infinitely many solutions, where some variables are determined by the other variables, and the other variables can take on arbitrary values.

Let us make this observation precise. **Pivot columns** are columns that contain an entry that is the first non-zero entry of some row. In our first example:
$$ \begin{bmatrix} 1& 2 & -2 \\ 0 & 1 & 1 \end{bmatrix}, $$
the pivot columns are the first and second columns. In our second example:
$$ \begin{bmatrix} 1 & -2 & 2 & 1 \\ 0 & 0 & 1 & -1 \end{bmatrix},$$
the pivot columns are the first and third columns. Any column that is not a pivot column is called a **free column**.

Note that each column corresponds to a variable, in the sense that entries of a column correspond to coefficients of a certain variable in the system. Indeed, if we expand out our first example
$$ \begin{bmatrix} 1& 2 & -2 \\ 0 & 1 & 1 \end{bmatrix} \begin{bmatrix} x \\ y \\ z\end{bmatrix} = \begin{bmatrix} 0 \\ 1\end{bmatrix},$$
we obtain
$$ \begin{cases} x + 2y - 2z = 0 \\ y + z =1\end{cases}$$
and note that the first, second and third column each correspond to coefficients of $x,y,z$ respectively. The variables corresponding to the pivot columns are called **pivot variables**. Similarly, those corresponding to free columns are called **free variables**.

In the first example, the free variables are $x$ and $y$, while $z$ is a free variable. In the second example, the pivot variables are $a$ and $c$, while the free variables are $b$ and $d$. The upshot is that once we go through back-substitution on $U\vec x = \vec y$, we can deduce that **free variables can take any value**, while **pivot variables are determined by values of the free variable**. Note that this is indeed the case for the two examples demonstrated above.

Since the matrix $U$ is wide, free columns/variables must always exist. This means that **in most cases, underdetermined systems have infinitely many solutions**. The reason for 'in most cases' will be clarified once we cover overdetermined systems.

### Overdetermined systems

Now let us move onto **overdetermined systems**, where the matrix $A$ has more rows than columns, i.e. $A$ is 'tall'. Taking the PLU decomposition, we again study the system $U\vec x = \vec y$ for upper-triangular $U$ with the same shape as $A$.

An observation here is that a tall, upper-triangular matrix $U$ must always have zeros filled in the bottom portion of the matrix. More precisely, if $U$ is an $m\times n$ matrix where $m > n$, then the bottom $m-n$ rows must all be zero. Here are two examples:
$$ \begin{bmatrix} 2 & 1 \\ 0 & 3 \\ 0 & 0 \end{bmatrix};\ \begin{bmatrix} 1 & -1 & 0 \\ 0 & 0 & 2 \\ 0 & 0 & 0 \\ 0 & 0 & 0 \end{bmatrix}.$$

These rows with zeros are usually problematic. Suppose we are given the system
$$ \begin{bmatrix} 1 & -1 & 0 \\ 0 & 0 & 2 \\ 0 & 0 & 0 \\ 0 & 0 & 0 \end{bmatrix}\begin{bmatrix} x\\ y \\ z\end{bmatrix} = \begin{bmatrix} 1 \\ 2 \\ 0 \\ -2\end{bmatrix}.$$

Note that the equation of the final row is
$$ 0x + 0y + 0z = -2$$
which cannot be true for any values of $x,y,z$. This means that the system has _no solutions_, and explains that **in most cases, overdetermined systems have no solutions**.

Why did we need to say the ambiguous phrase 'most cases' again? This is because sometimes the zero rows are not a problem. Suppose the vector on the right-hand side is given this time as
$$ \begin{bmatrix} 1 & -1 & 0 \\ 0 & 0 & 2 \\ 0 & 0 & 0 \\ 0 & 0 & 0 \end{bmatrix}\begin{bmatrix} x\\ y \\ z\end{bmatrix} = \begin{bmatrix} 1 \\ 2 \\ 0 \\ 0\end{bmatrix}.$$
Then the equations of the last two rows are
$$ 0x + 0y + 0z  = 0$$
which is satisfied for any values of $x,y,z$ and hence is not a problem! Carrying out back-substitution as usual, we see that the second row gives us
$$ z = 1$$
and the first row gives us
$$ x = y+1.$$
Hence we have infinitely many solution $x = y+1$, $z=1$. Note that the previous argument for underdetermined systems applies here too: the pivot variables are $x,z$ and the free variable is $y$.

Consider another example
$$ \begin{bmatrix} 2 & 1 \\ 0 & 3 \\ 0 & 0 \end{bmatrix}\begin{bmatrix} x \\ y \end{bmatrix} = \begin{bmatrix} 1 \\ 3 \\ 6 \end{bmatrix}.$$
Since the zero row corresponds to $0x + 0y = 6$, there are no solutions. If the vector on the right-hand side is instead given as
$$ \begin{bmatrix} 2 & 1 \\ 0 & 3 \\ 0 & 0 \end{bmatrix}\begin{bmatrix} x \\ y \end{bmatrix} = \begin{bmatrix} 1 \\ 3 \\ 0 \end{bmatrix},$$
then the zero row is a no longer an issue, and we can carry out back-substitution as though the coefficient matrix was square and obtain $y = 1$, $x = 0$.

We can now correct our observation and say that: **in most cases overdetermined systems have no solutions**, but they may have **one or infinitely many solutions depending on the constant vector**.

This sort of exception also occurs for underdetermined systems and is why we included the 'in most cases' ambiguity as well. If the underdetermined, wide upper triangular matrix $U$ happens to have zero rows on the bottom, there may not be any solutions. An example would be
$$ \begin{bmatrix} 1 & 2 & 0 & 1 \\ 0 & 1 & 1 & 0 \\ 0 & 0 & 0 & 0\end{bmatrix} \begin{bmatrix} x\\y\\z\\w\end{bmatrix} = \begin{bmatrix} 0 \\1 \\ 1\end{bmatrix},$$
for which the last row translates to $0 = 1$ again, hence there being no solutions. But if we are given the constant vector as
$$ \begin{bmatrix} 1 & 2 & 0 & 1 \\ 0 & 1 & 1 & 0 \\ 0 & 0 & 0 & 0\end{bmatrix} \begin{bmatrix} x\\y\\z\\w\end{bmatrix} = \begin{bmatrix} 0 \\ 1 \\ 0\end{bmatrix},$$
then the last row is redundant and we can solve the system as usual and get infinitely many solutions:
$$ x = 2z-w-2,\ y = -z+1.$$
Note indeed that $x,y$ are the pivot variables and $z,w$ are the free variables. Hence we can correct our observation as: **in most cases underdetermined systems have infinitely many solutions**, but they may have **no solutions depending on the constant vector**.


## Geometric interpretation of under/overdetermined systems

One can give a very nice geometric interpretation of these linear systems, especially in low dimensions. Let us go back to the example of the first underdetermined system:
$$ \begin{bmatrix} 1& 2 & -2 \\ 0 & 1 & 1 \end{bmatrix} \begin{bmatrix} x \\ y \\ z \end{bmatrix} = \begin{bmatrix} 0 \\ 1 \end{bmatrix}.$$
Solving this system is equivalent to finding the intersection of the following two planes in $\mathbb{R}^3$.
$$x + 2y - 2z = 0,\ y+ z =1$$

Note that the two planes intersect at a line, meaning that there are infinitely many solutions. This is the almost always the case for two planes in $\mathbb{R}^3$, i.e. systems with three variables and two equations. This is indeed the case for the example above as well.

The only time this fails is when the two planes are parallel, for instance
$$ x + 2y -2z = 5,\ x + 2y - 2 = 2.$$
Translating this into a matrix-vector equation
$$ \begin{bmatrix} 1 & 2 & -2 \\ 1 & 2 & -2 \end{bmatrix} \begin{bmatrix} x \\ y \\ z \end{bmatrix} = \begin{bmatrix} 5 \\ 2 \end{bmatrix}.$$
The LU decomposition of the coefficient matrix is 
$$ \begin{bmatrix} 1 & 2 & -2 \\ 1 & 2 & -2 \end{bmatrix} = \begin{bmatrix} 1 & 0 \\ 1 & 1\end{bmatrix}\begin{bmatrix} 1 & 2 & -2 \\ 0 & 0 & 0 \end{bmatrix},$$
and the system reduces to solving 
$$ \begin{bmatrix} 1 & 2 & -2 \\ 0 & 0 & 0 \end{bmatrix}\begin{bmatrix} x\\y\\z\end{bmatrix} = \begin{bmatrix} 5 \\ -3 \end{bmatrix}.$$
This is precisely the problematic case discussed above where we have a zero row corresponding to the equation $ 0 =-3$, forcing there the system to have no solutions.

Now let us consider examples of overdetermined systems. Consider for example
$$\begin{bmatrix} 2 & 1 \\ 0 & 3 \\ 4 & -1\end{bmatrix}\begin{bmatrix} x \\ y \end{bmatrix} = \begin{bmatrix} 1 \\ 3 \\5\end{bmatrix}.$$
Finding the solution to this system is equivalent to finding the intersection of the following three lines in $\mathbb{R}^2$:
$$ 2x + y= 1,\ 3y=3,\ 4x-y = 5.$$
Note that three lines in $\mathbb{R}^2$ almost always never intersect at one point. This is the case for this system as well:
<!-- Insert picture of three lines on eucl plane -->
By the LU decomposition
$$ \begin{bmatrix} 2 & 1 \\ 0 & 3 \\ 4 & -1\end{bmatrix} = \begin{bmatrix} 1 & 0 & 0 \\ 0 & 1 & 0 \\ 2 & -1 & 0\end{bmatrix} \begin{bmatrix} 2 & 1 \\ 0 & 3 \\ 0 & 0\end{bmatrix},$$
this amounts solving the system
$$ \begin{bmatrix} 2 & 1 \\ 0 & 3 \\ 0 & 0\end{bmatrix}\begin{bmatrix} x\\y\\z \end{bmatrix} = \begin{bmatrix} 1 \\ 3 \\ 6\end{bmatrix}.$$
This is the system we saw earlier -- there are no solutions.

But there are rare cases where three lines meet at one point; for instance consider the same system with the new constant vector 
$$\begin{bmatrix} 2 & 1 \\ 0 & 3 \\ 4 & -1\end{bmatrix}\begin{bmatrix} x \\ y \end{bmatrix} = \begin{bmatrix} 1 \\ 3 \\-1\end{bmatrix}.$$
LU decomposition leads to the system
$$ \begin{bmatrix} 2 & 1 \\ 0 & 3 \\ 0 & 0\end{bmatrix}\begin{bmatrix} x\\y\\z \end{bmatrix} = \begin{bmatrix} 1 \\ 3 \\ 0\end{bmatrix},$$
which has one solution.

It is also possible for overdetermined systems to have infinitely many solutions. Geometrically, this corresponds to precisely when three of the lines are exactly the same as each other.
