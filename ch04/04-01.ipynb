# 4.1: Writing systems using matrices

In Chapter 2, we solved systems by looking for intersections of lines and planes. In Chapter 3, we learned how matrices describe linear transformations. Now we'll put these ideas together: **solving a linear system means finding a vector that a matrix sends to a specified output**.

We'll start with a small electrical network, then develop a method that also works for much larger systems.


---

## From equations to a matrix

Consider the system

$$\begin{aligned}3x-y&=15,\\-x+2y&=0.\end{aligned}$$

Put the coefficients of $x$ in the first column and the coefficients of $y$ in the second. Then matrix-vector multiplication gives

$$\begin{bmatrix}3&-1\\-1&2\end{bmatrix}\begin{bmatrix}x\\y\end{bmatrix}
=\begin{bmatrix}3x-y\\-x+2y\end{bmatrix}.$$

So our two equations can be written as one matrix equation:

$$\underbrace{\begin{bmatrix}3&-1\\-1&2\end{bmatrix}}_A
\underbrace{\begin{bmatrix}x\\y\end{bmatrix}}_{\vec x}
=\underbrace{\begin{bmatrix}15\\0\end{bmatrix}}_{\vec b}.$$

:::{note} Definition: Matrix form of a linear system
A system of $n$ linear equations in $d$ unknowns can be written as

$$\boxed{A\vec x=\vec b,}$$

where the **coefficient matrix** $A$ has shape $n\times d$, the vector of unknowns $\vec x$ belongs to $\mathbb R^d$, and the right-hand side $\vec b$ belongs to $\mathbb R^n$.

Each row of $A$ corresponds to an equation. Each column corresponds to a variable.
:::

Keep the variables in the **same order in every equation**, and use a zero when a variable is missing. For example,

$$\begin{aligned}2x-z&=4,\\x+3y&=7\end{aligned}
\quad\Longleftrightarrow\quad
\begin{bmatrix}2&0&-1\\1&3&0\end{bmatrix}\begin{bmatrix}x\\y\\z\end{bmatrix}
=\begin{bmatrix}4\\7\end{bmatrix}.$$

The coefficient matrix need not be square! Here there are two equations and three unknowns.

$$2x-y=6,\qquad3y+4z=-2.$$

## A circuit with two unknown voltages

An engineer connects a $10$-volt supply to a resistor network. There are two junctions with unknown voltages $V_1$ and $V_2$. The ground connection has voltage $0$.

The connections are:

| From | To | Resistance |
| --- | --- | --- |
| $10$-volt supply | $V_1$ | $2\,\Omega$ |
| $V_1$ | Ground | $6\,\Omega$ |
| $V_1$ | $V_2$ | $3\,\Omega$ |
| $V_2$ | Ground | $3\,\Omega$ |

We need two physical rules. **Ohm's law** says that current from voltage $V$ toward voltage $W$ through resistance $R$ is $(V-W)/R$. A negative value means the current flows in the opposite direction. **Current balance** says that total current entering a junction equals total current leaving it.

At $V_1$, current arrives from the supply and leaves toward ground and $V_2$:

$$\frac{10-V_1}{2}=\frac{V_1}{6}+\frac{V_1-V_2}{3}.$$

Multiplying by 6 gives $30-3V_1=V_1+2(V_1-V_2)$. Collecting terms and dividing by 2 gives

$$3V_1-V_2=15.$$

At $V_2$, current arriving from $V_1$ leaves toward ground:

$$\frac{V_1-V_2}{3}=\frac{V_2}{3}
\quad\Longrightarrow\quad -V_1+2V_2=0.$$

Together, these equations give

$$\boxed{\begin{bmatrix}3&-1\\-1&2\end{bmatrix}\begin{bmatrix}V_1\\V_2\end{bmatrix}
=\begin{bmatrix}15\\0\end{bmatrix}.}$$

The resistances determine the coefficient matrix. Changing the supply voltage changes the right-hand side. We'll return to this system in [Chapter 4.3](04-03.ipynb).

## What does a solution mean?

A **solution** is one vector whose entries satisfy every equation at the same time. For our circuit,

$$\begin{bmatrix}3&-1\\-1&2\end{bmatrix}\begin{bmatrix}6\\3\end{bmatrix}
=\begin{bmatrix}18-3\\-6+6\end{bmatrix}
=\begin{bmatrix}15\\0\end{bmatrix},$$

so $V_1=6$ and $V_2=3$ solve the system.

We can interpret this in three familiar ways:

- **Equations:** the point $(6,3)$ lies on both lines $3x-y=15$ and $-x+2y=0$.
- **Columns:** $6\begin{bmatrix}3\\-1\end{bmatrix}+3\begin{bmatrix}-1\\2\end{bmatrix}=\begin{bmatrix}15\\0\end{bmatrix}$.
- **Transformations:** the map $T(\vec x)=A\vec x$ sends $\begin{bmatrix}6\\3\end{bmatrix}$ to $\begin{bmatrix}15\\0\end{bmatrix}$.

For a matrix of shape $n\times d$, this transformation has domain $\mathbb R^d$ and codomain $\mathbb R^n$.

If $A$ is invertible, [Chapter 3.6](../ch03/03-06.ipynb) tells us that the unique solution is $\vec x=A^{-1}\vec b$. But we don't need to compute the entire inverse to solve a system. Our goal is to find the solution by working with simpler matrices.

## Another model: Temperatures in a metal bar

A metal bar is divided into three equal segments. Its endpoints are kept at $0^\circ\mathrm C$ and $60^\circ\mathrm C$, and the interior temperatures are $T_1$ and $T_2$, in that order:

$$0^\circ\mathrm C\quad\longleftrightarrow\quad T_1\quad\longleftrightarrow\quad T_2\quad\longleftrightarrow\quad60^\circ\mathrm C.$$

No heat escapes through the sides. Assume each segment has thermal conductance $1$ watt per degree Celsius, so the heat flow toward the left is the temperature at its right endpoint minus the temperature at its left endpoint, in watts.

The temperatures have stopped changing: at each interior point, heat arriving from the right equals heat leaving to the left. **How can we use this balance to write a system for the two unknown temperatures?**

### Setting up the equations

At the point with temperature $T_1$, heat arrives from the right at a rate of $T_2-T_1$ watts and leaves toward the left at a rate of $T_1-0$ watts. Setting these equal gives

$$T_2-T_1=T_1-0\quad\Longrightarrow\quad 2T_1-T_2=0.$$

At the point with temperature $T_2$, heat arrives from the right endpoint at a rate of $60-T_2$ watts and leaves toward $T_1$ at a rate of $T_2-T_1$ watts. Thus,

$$60-T_2=T_2-T_1\quad\Longrightarrow\quad -T_1+2T_2=60.$$

Our system is therefore

$$\boxed{\begin{aligned}2T_1-T_2&=0,\\-T_1+2T_2&=60.\end{aligned}}$$

### Writing the system using a matrix

Order the unknowns as $T_1,T_2$. The coefficients of $T_1$ form the first column, and the coefficients of $T_2$ form the second:

$$\boxed{\underbrace{\begin{bmatrix}2&-1\\-1&2\end{bmatrix}}_A
\underbrace{\begin{bmatrix}T_1\\T_2\end{bmatrix}}_{\vec T}
=\underbrace{\begin{bmatrix}0\\60\end{bmatrix}}_{\vec b}.}$$

If the right endpoint changes to $90^\circ\mathrm C$, the coefficient matrix stays the same, while the right-hand side becomes $\begin{bmatrix}0\\90\end{bmatrix}$. Just as in the circuit example, changing a boundary value changes $\vec b$ without changing $A$.

We'll revisit this system in [Chapter 4.3](04-03.ipynb), where we'll use an LU factorization to find the two temperatures.
