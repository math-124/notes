# 4.7: Row operations and LU factorization

We have found factors by multiplying and matching entries. Row operations give another way to organize those same calculations.

---

## Operations that preserve solutions

Until now, we have found the factors by multiplying and matching entries. The same calculations can also be understood as simplifying the equations themselves.

:::{note} Definition: Row operations
Each row represents an equation. We may:

-   interchange two equations;

-   multiply an equation by a nonzero number;

-   add a multiple of one equation to another equation.
:::

These changes preserve the solution set because each can be reversed.

:::{attention} Operate on the entire equation
Apply the change to the **entire equation, including its right-hand side**. For the LU procedure, we need only row interchanges and adding multiples of rows; we do not need to scale pivots to one.
:::



---

## An example we have already factored

In Chapter 4.5 we considered 

$$\begin{aligned}
x_1+2x_2-x_3&=3,\\
2x_1+4x_2&=10,\\
-x_1-2x_2+7x_3&=9.
\end{aligned}$$

 Subtract twice the first equation from the second, and add the first equation to the third. The equivalent system is 

$$\begin{aligned}
x_1+2x_2-x_3&=3,\\
2x_3&=4,\\
6x_3&=12.
\end{aligned}$$

 Now subtract three times the *new second equation* from the third: 

$$\begin{aligned}
x_1+2x_2-x_3&=3,\\
2x_3&=4,\\
0&=0.
\end{aligned}$$

 The coefficient matrix is now in row echelon form: 

$$U=\begin{bmatrix}1&2&-1\\0&0&2\\0&0&0\end{bmatrix},
\qquad
\vec y=\begin{bmatrix}3\\4\\0\end{bmatrix}.$$

 These are exactly the matrix $U$ and vector $\vec y$ found using LU.

Notice that column $2$ needed no further work after the first step: its entries below the first row were already zero. We moved to column $3$ for the next pivot.

---

## How L records the operations

Let $R_i$ mean the current row $i$, including its right-hand side. The operations in the example were 

$$R_2\rightsquigarrow R_2-2R_1,\qquad
R_3\rightsquigarrow R_3-(-1)R_1,\qquad
R_3\rightsquigarrow R_3-3R_2.$$

 The notation $R_2\rightsquigarrow R_2-2R_1$ means that we replace $R_2$ by $R_2-2R_1$. The other arrows have the same meaning. In the last operation, $R_2$ is the updated second row.

The numbers we *subtracted as multiples* were $2$, $-1$, and $3$. They are exactly the entries below the diagonal in 

$$\boxed{L=\begin{bmatrix}1&0&0\\2&1&0\\-1&3&1\end{bmatrix}.}$$

 In general, when we subtract a multiple of pivot row $k$ from a lower row $i$, we record that multiple in entry $(i,k)$ of $L$. The column of $L$ records the *pivot row*, even when the pivot itself lies farther to the right in $U$.

---

## Recovering the original system

In this example, 

$$A=LU=
\begin{bmatrix}1&0&0\\2&1&0\\-1&3&1\end{bmatrix}
\begin{bmatrix}1&2&-1\\0&0&2\\0&0&0\end{bmatrix}.$$

 Read the product row by row:

-   The original first row equals the first row of $U$.

-   The original second row equals twice the first row of $U$, plus its second row.

-   The original third row equals minus the first row of $U$, plus three times its second row, plus its third row.

Thus $L$ stores the information needed to reconstruct the original rows from the echelon rows. Its entries record the multiples used in the row operations.

**What happens to the right-hand side?** The same operations change $\vec b$ into $\vec y$. Equivalently, $L\vec y=\vec b$. Here, 

$$y_1=3,\qquad y_2=10-2y_1=4,\qquad
y_3=9+y_1-3y_2=0.$$

 Forward substitution performs exactly the same subtractions on the right-hand side.

**What if rows are interchanged?** Record the row interchanges in $P$. If a swap happens after some multiples have been recorded in $L$, swap those already computed entries as well. The result is 

$$PA=LU,\qquad L\vec y=P\vec b.$$

 Finding LU and using row operations to bring a system to echelon form are two ways of organizing the same calculation. The factorization keeps a record of that calculation, so we can reuse it for a different right-hand side.
