# 4.2: PLU decomposition for invertible matrices

Recall that given the LU decomposition of the coefficient matrix:
$$A  = LU,$$
we can easily solve the equation $A\vec x = \vec b$ by setting $U\vec x = \vec y$ and solving the two triangular systems:
$$L\vec y = \vec b \quad \text{and} \quad U\vec x = \vec y$$
by forward and backward substitution, respectively.

## From elimination to factorization

One question we must ask here is: How does one find the LU decomposition of a square matrix $A$? The answer is rather surprising. We will first look at how to put matrices in upper-triangular form using _row reduction_, and then see how this gives us the LU factorization.

Let's look at a simple example. Suppose we are given the following $2\times 2$ matrix:
$$ \begin{bmatrix} 1 & 1 \\ 2 & 0 \end{bmatrix}.$$
Note that subtracting 2 times the first row from the second row gives us the following upper triangular matrix:
$$ \begin{bmatrix} 1 & 1 \\ 0 & -2 \end{bmatrix}.$$
This act of adding scalar multiples of one row to another is called an **elementary row operation** (there will be another type of elementary row operation that we will see later). In many cases, one can repeat elementary row operations to put a matrix in upper-triangular form. Consider this time the following $3\times 3$ matrix:
$$ \begin{bmatrix} 1 & 1 & 1 \\ 2 & 0 & 1 \\ 3 & 2 & 0 \end{bmatrix}.$$
We can perform the following elementary row operations to put this matrix in upper-triangular form:
1. Subtract 2 times the first row from the second row, and subtract 3 times the first row from the third row:
$$ \begin{bmatrix} 1 & 1 & 1 \\ 0 & -2 & -1 \\ 0 & -1 & -3 \end{bmatrix}.$$
2. Subtract 1/2 times the second row from the third row:
$$ \begin{bmatrix} 1 & 1 & 1 \\ 0 & -2 & -1 \\ 0 & 0 & -5/2 \end{bmatrix}.$$

**Row reduction** is the processing of repeating elementary row operations to put a matrix in upper-triangular form. Here is the procedure described more precisely: we always start by subtracting multiples of the first row from the rows below it to make the first column have zeros below the first entry. Then we move on to the second row, and subtract multiples of the second row from the rows below it to make the second column have zeros below the second entry. We continue this process until we reach the last row. The final result is an upper-triangular matrix!

How does row reduction relate to LU factorization? Here is an extremely important calculation:
$$ \begin{bmatrix} 1 & 0 \\ -2 & 1 \end{bmatrix} \begin{bmatrix} 1 & 1 \\ 2 & 0 \end{bmatrix} = \begin{bmatrix} 1 & 1 \\ -2\cdot 1 + 1\cdot 0 & -2\cdot 1 + 0\cdot 0 \end{bmatrix} = \begin{bmatrix} 1 & 1 \\ 0 & -2 \end{bmatrix}.$$
Note that the effect of multiplying the matrix $\begin{bmatrix} 1 & 0 \\ -2 & 1 \end{bmatrix}$ on the left is equivalent to performing the elementary row operation of subtracting 2 times the first row from the second row. This is true in more general settings, which we can see from the calculation:
$$ \begin{bmatrix} 1 & 0 \\ -2 & 1 \end{bmatrix}\begin{bmatrix} a & b \\ c & d \end{bmatrix} = \begin{bmatrix} a & b \\ c-2a & d-2b \end{bmatrix}.$$
How do we `reverse' this elementary row operation? We should add 2 times the first row to the second row. This is equivalent to multiplying by the matrix $\begin{bmatrix} 1 & 0 \\ 2 & 1 \end{bmatrix}$ on the left:
$$ \begin{bmatrix} 1 & 0 \\ 2 & 1 \end{bmatrix}\begin{bmatrix} a & b \\ c-2a & d-2b \end{bmatrix} = \begin{bmatrix} a & b \\ 2a + (c-2a) & 2b + (d-2b) \end{bmatrix} = \begin{bmatrix} a & b \\ c & d \end{bmatrix}.$$
Applying this observation to our previous example, we see that the matrix $\begin{bmatrix} 1 & 0 \\ -2 & 1 \end{bmatrix}$ is invertible with its inverse is given by $\begin{bmatrix} 1 & 0 \\ 2 & 1 \end{bmatrix}$, and that
$$ \begin{bmatrix} 1 & 1 \\ 2 & 0 \end{bmatrix} = \begin{bmatrix} 1 & 0 \\ 2 & 1 \end{bmatrix}\begin{bmatrix} 1 & 1 \\ 0 & -2 \end{bmatrix}.$$
At the end, we obtain the LU decomposition of the matrix $\begin{bmatrix} 1 & 1 \\ 2 & 0 \end{bmatrix}$.

Let's look at the $3\times 3$ example. The first elementary row operation of subtracting 2 times the first row from the second row and subtracting 3 times the first row from the third row is equivalent to multiplying by the matrix
$$ \begin{bmatrix} 1 & 0 & 0 \\ -2 & 1 & 0 \\ -3 & 0 & 1 \end{bmatrix}$$
on the left. The second elementary row operation of subtracting 1/2 times the second row from the third row is equivalent to multiplying by the matrix
$$ \begin{bmatrix} 1 & 0 & 0 \\ 0 & 1 & 0 \\ 0 & -1/2 & 1 \end{bmatrix}$$
on the left. Therefore, we have
$$ \begin{bmatrix} 1 & 0 & 0 \\ 0 & 1 & 0 \\ 0 & -1/2 & 1 \end{bmatrix} \begin{bmatrix} 1 & 0 & 0 \\ -2 & 1 & 0 \\ -3 & 0 & 1 \end{bmatrix} \begin{bmatrix} 1 & 1 & 1 \\ 2 & 0 & 1 \\ 3 & 2 & 0 \end{bmatrix} = \begin{bmatrix} 1 & 1 & 1 \\ 0 & -2 & -1 \\ 0 & 0 & -5/2 \end{bmatrix}.$$
Now multiplying the inverses of the two lower-triangular matrices on the left, we obtain $\begin{bmatrix} 1 & 1 & 1 \\ 2 & 0 & 1 \\ 3 & 2 & 0 \end{bmatrix}$:
$$ \begin{bmatrix} 1 & 1 & 1 \\ 2 & 0 & 1 \\ 3 & 2 & 0 \end{bmatrix} =
\begin{bmatrix} 1 & 0 & 0 \\ 2 & 1 & 0 \\ 3 & 0 & 1 \end{bmatrix}
\begin{bmatrix} 1 & 0 & 0 \\ 0 & 1 & 0 \\ 0 & 1/2 & 1 \end{bmatrix}
\begin{bmatrix} 1 & 1 & 1 \\ 0 & -2 & -1 \\ 0 & 0 & -5/2 \end{bmatrix}.$$
Note that now the $3\times 3$ example is a bit more complicated than the $2\times 2$ example, since we have two lower-triangular matrices on the left. However, multiplying these two lower-triangular matrices results in another lower-triangular matrix:
$$
\begin{bmatrix} 1 & 1 & 1 \\ 2 & 0 & 1 \\ 3 & 2 & 0 \end{bmatrix} =
\begin{bmatrix} 1 & 0 & 0 \\ 2 & 1 & 0 \\ 3 & 1/2 & 1 \end{bmatrix} \begin{bmatrix} 1 & 1 & 1 \\ 0 & -2 & -1 \\ 0 & 0 & -5/2 \end{bmatrix}$$
which again gives us the LU decomposition! This worked because **products of lower-triangular matrices are again lower-triangular matrices**.

Summarizing these observations, we see that
1. Row reduction where we subtract a multiple of one row from another row on the bottom is equivalent to multiplying a matrix on the left by a lower-triangular matrix.
2. These lower-triangular matrices are invertible, since we can `reverse' the row operation by adding a multiple of one row to another row on the bottom, which is equivalent to multiplying by another lower-triangular matrix on the left.
3. Tracking the lower-triangular matrices that we multiply on the left during row reduction gives us the LU factorization of a matrix.

## Permutation matrices and PLU decomposition

One problem with the LU decomposition is that it does not always exist. For example, consider the matrix
$$ \begin{bmatrix} 0 & 1 & 0\\ 1 & 1 & 1 \\ 2 & 2 & 1 \end{bmatrix}.$$
If we try to do row reduction as we did before, we immediately run into the problem -- we cannot put it in upper-triangular form by subtracting multiples of the first row from the second row, since the first entry is zero.

A solution would be to swap the first two rows, which gives us
$$ \begin{bmatrix} 1 & 1 & 1 \\ 0 & 1 & 0 \\ 2 & 2 & 1 \end{bmatrix},$$
and now we can subtract twice the first row from the third row to put our matrix in upper-triangular form. It is in fact known that **after suitably swapping around the rows, a square matrix always admits an LU decomposition**.

For this fact to be useful, we should re-cast it as a decomposition of the matrix $A$, just like how row operations on $A$ gave us the LU decomposition. In order to do this, we need to introduce the notion of a **permutation matrix**. A permutation matrix is a square matrix that has exactly one entry of 1 in each row and each column and 0s elsewhere. The following are examples of permutation matrices:
$$ \begin{bmatrix} 0 & 1 \\ 1 & 0 \end{bmatrix},\begin{bmatrix} 0 & 1 & 0 \\ 0 & 0 & 1 \\ 1 & 0 & 0 \end{bmatrix}.$$

Just like multiplying by a lower-triangular matrix with ones on the diagonal on the left corresponds to performing an elementary row operation, multiplying by a permutation matrix on the left corresponds to swapping rows. For example, note that
$$ \begin{bmatrix} 0 & 1 \\ 1 & 0 \end{bmatrix} \begin{bmatrix} a & b \\ c & d \end{bmatrix} = \begin{bmatrix} c & d \\ a & b \end{bmatrix},$$
i.e. multiplying the permutation matrix swaps the first and second rows of the matrix.
Similarly, note that
$$ \begin{bmatrix} 0 & 1 & 0 \\ 0 & 0 & 1 \\ 1 & 0 & 0 \end{bmatrix} \begin{bmatrix} a & b & c \\ d & e & f \\ g & h & i \end{bmatrix} = \begin{bmatrix} d & e & f \\ g & h & i \\ a & b & c \end{bmatrix}.$$
Carrying out the matrix multiplication manually, we see that the second entry of the first row being 1 means that the second row of the original matrix is sent to the first row. Similarly note that
* the third entry of the second row being 1 means that the third row of the original matrix is sent to the second row;
* the first entry of the third row being 1 means that the first row of the original matrix is sent to the third row.

Therefore the previous statement about the existence of LU decompositions after swapping rows can be re-written as follows:
$$ PA = LU$$
for a suitable permutation matrix $P$. Note also that the permutation matrix $P$ is invertible, since we can undo the row swaps which corresponds to multiplying by another permutation matrix on the left. Hence we can also write
$$ A = P^{-1}LU$$
where $P^{-1}$ is also a permutation matrix. This is called the **PLU decomposition** of the matrix $A$.

## Solving equations with the PLU decomposition

How does the PLU decomposition help us solve equations? Suppose we have a square matrix $A$ and we want to solve the equation
$$ A\vec x = \vec b.$$
Let us pick a suitable permutation matrix $P$ such that $PA = LU$. Multiplying both sides by $P$, we have
$$ PA\vec x = LU\vec x = P\vec b.$$
Note that multiplying $P$ on the left of $\vec b$ also permutes the entries of $\vec b$ in the same way we described above. Now we are back to the situation of the LU decomposition, ad we can solve the equation $LU\vec x = P\vec b$ by forward and backward substitution!

For instance, suppose we are given the equation
$$ \begin{bmatrix} 0 & 1 & 0\\ 1 & 1 & 1 \\ 2 & 2 & 1 \end{bmatrix}\begin{bmatrix} x \\ y \\ z\end{bmatrix} = \begin{bmatrix} 1 \\ 2 \\ 3 \end{bmatrix}.$$
Recall that we only perform row reductions on the coefficient matrix after swapping the first and second rows. This swapping corresponds to multiplying the permutation matrix
$$ P = \begin{bmatrix} 0 & 1 & 0\\ 1 & 0 & 0 \\ 0 & 0 & 1 \end{bmatrix}$$
on the left. Multiplying both sides of the equation by $P$, we have
$$
\begin{bmatrix} 0 & 1 & 0\\ 1 & 0 & 0 \\ 0 & 0 & 1 \end{bmatrix}\begin{bmatrix} 0 & 1 & 0\\ 1 & 1 & 1 \\ 2 & 2 & 1 \end{bmatrix}\begin{bmatrix} x \\ y \\ z\end{bmatrix} = \begin{bmatrix} 1 & 1 & 1 \\ 0 & 1 & 0 \\ 2 & 2 & 1 \end{bmatrix}\begin{bmatrix} x \\ y \\ z\end{bmatrix},\quad \begin{bmatrix} 0 & 1 & 0\\ 1 & 0 & 0 \\ 0 & 0 & 1 \end{bmatrix}\begin{bmatrix} 1 \\ 2 \\ 3 \end{bmatrix} = \begin{bmatrix} 2 \\ 1 \\ 3 \end{bmatrix}.$$
Therefore the equation is now
$$ \begin{bmatrix} 1 & 1 & 1 \\ 0 & 1 & 0 \\ 2 & 2 & 1 \end{bmatrix}\begin{bmatrix} x \\ y \\ z\end{bmatrix} = \begin{bmatrix} 2 \\ 1 \\ 3 \end{bmatrix}.$$
We can put this coefficient matrix in upper-triangular form by subtracting twice the first row from the third row, thus obtaining the LU decomposition
$$ \begin{bmatrix} 1 & 0 & 0 \\ 0 & 1 & 0 \\ 2 & 0 & 1 \end{bmatrix} \begin{bmatrix} 1 & 1 & 1 \\ 0 & 1 & 0 \\ 0 & 0 & -1\end{bmatrix}.$$
We first solve the system
$$ \begin{bmatrix} 1 & 0 & 0 \\ 0 & 1 & 0 \\ 2 & 0 & 1 \end{bmatrix}\begin{bmatrix} x'\\ y'\\ z'\end{bmatrix} = \begin{bmatrix} 2 \\ 1 \\ 3 \end{bmatrix}$$
with forward-substitution to obtain $x'=2$, $y'=1$ and $z'=-1$. Then solving the system
$$ \begin{bmatrix} 1 & 1 & 1 \\ 0 & 1 & 0 \\ 0 & 0 & -1\end{bmatrix}\begin{bmatrix} x\\ y\\ z\end{bmatrix} = \begin{bmatrix} 2 \\ 1 \\ -1 \end{bmatrix}$$
with backward-substitution gives us $x=0$, $y=1$ and $z=-1$.
