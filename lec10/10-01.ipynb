# 10.1: Solving equations with LU decompositions

Recall that systems of linear equations, with $n$ variables and $m$ equations, can be represented as the following matrix-vector equation:
$$ A\vec x = \vec b,$$
where $A$ is an $m \times n$ matrix, $\vec x$ is variable vector with $n$ entries, and $\vec b$ is a vector of constants with $m$ entries.

Let us focus on the case of $A$ being a square matrix. How do we solve this system of equations? It turns out that if $A$ is given in specific shapes, it becomes much easier to solve.

## Upper/lower triangular matrices

Given a square matrix $A$, we say that $A$ is an **upper triangular** matrix if all entries below the main diagonal are zero. Similarly, we say that $A$ is a **lower triangular** matrix if all entries above the main diagonal are zero.

Here are some examples of upper triangular matrices:
$$ \begin{bmatrix} 1 & 2 & 3 \\ 0 & 4 & 5 \\ 0 & 0 & 6 \end{bmatrix},\quad \begin{bmatrix} 7 & 8 \\ 0 & 9 \end{bmatrix},\quad \begin{bmatrix} 1 & 2 \\ 0 & 0 \end{bmatrix}.$$
Similarly, here are some examples of lower triangular matrices:
$$ \begin{bmatrix} 1 & 0 & 0 \\ 4 & 5 & 0 \\ 7 & 8 & 9 \end{bmatrix},\quad \begin{bmatrix}-1 & 0 \\ 2 & 3 \end{bmatrix},\quad \begin{bmatrix} 1 & 0 \\ 0 & 0 \end{bmatrix}.$$

## Back/forward substitution

Suppose we have a system of equations represented in matrix form as $A\vec x = \vec b$, where $A$ is an upper triangular matrix. We can easily solve this system using **back substitution**. This is a method where we solve for the last variable first, then substitute that value into the second to last equation, and so on until we have solved for all variables.

### Examples
Let us look at a lot of examples.

1. Consider the following system where the coefficient matrix $A$ is upper triangular:
    $$ \begin{bmatrix} 2 & 1 \\ 0 & 3 \end{bmatrix} \begin{bmatrix} x \\ y \end{bmatrix} = \begin{bmatrix} 5 \\ 9 \end{bmatrix}$$
    which is the following system of equations re-written in matrix form:
    $$ \begin{cases}
    2x + y = 5 \\
    3y = 9
    \end{cases}$$
    We first solve the bottom equation for $y$ by dividing out each side by 3 and get $y=3$. Then we substitute this value of $y$ into the top equation to solve for $x$:
    $$ 2x + 3 = 5,\quad x = 1.$$
    Hence we obtain the solution as $x = 1$ and $y = 3$.

1. Consider the following system with three variables:
    $$ \begin{bmatrix} 1 & -2 & 1 \\ 0 & 3 & -1 \\ 0 & 0 & 2 \end{bmatrix} \begin{bmatrix} x \\ y \\ z \end{bmatrix} = \begin{bmatrix} 4 \\ 0 \\ 6 \end{bmatrix}$$

    We solve the last equation first which is $2z = 6$, giving us $z=3$. Then substituting this into the second equation, we get $3y - 3 = 0$, which simplifies to $3y = 3$, giving us $y=1$. Finally, substituting both $y$ and $z$ into the first equation, we have $x-2\cdot 1 + 3 = 4$, which gives us $x=3$. Therefore, the solution to the system is $x=3$, $y=1$, and $z=3$.

1. Consider the following system:
    $$ \begin{bmatrix} 1 & 2 \\ 0 & 0 \end{bmatrix} \begin{bmatrix} x \\ y \end{bmatrix} = \begin{bmatrix} 4 \\ 0 \end{bmatrix}$$
    Note that the last equation is $0=0$, which is true for any value of $y$. This means that the system has infinitely many solutions, and we can express the solution in terms of a _free variable_ which we choose as $y$. Substituting into the first equation, we have $x + 2y = 4$, which gives us $x = 4-2y$.

1. Let us look at a system with the same coefficient matrix as the previous example, but with a different constant vector:
    $$ \begin{bmatrix} 1 & 2 \\ 0 & 0 \end{bmatrix} \begin{bmatrix} x \\ y \end{bmatrix} = \begin{bmatrix} 4 \\ 5 \end{bmatrix}$$
    In this case, the last equation is $0=5$, which is false. This means that the system has no solution.


Similarly, if $A$ is a lower triangular matrix, we can solve the system using **forward substitution**, where we solve for the first variable first, then substitute that value into the second equation, and so on. This is very similar to back substitution, but we start from the top equation instead of the bottom.

Let us look at just one example of forward substitution. Consider the following system with a lower triangular coefficient matrix:
$$ \begin{bmatrix} 4 & 0 \\ 3 & 1 \end{bmatrix} \begin{bmatrix} x \\ y \end{bmatrix} = \begin{bmatrix} 4 \\ 9 \end{bmatrix}$$
Solving the first equation $4x = 4$ gives us $x = 1$. Substituting this into the second equation, we obtain
$$ 3\cdot 1 + y = 9$$
which gives us $y=6$. Thus the solution of this system is $x=1$, $y=6$.

## LU decomposition

We've see that if the coefficient matrix $A$ is given by an upper/lower triangular matrix, we can easily solve the equation $A\vec x = \vec b$ using backward/forward substitution. How does this help us when we are given a general square matrix $A$?

It turns out that given any square matrix $A$, we can write $A$ as a product of a lower triangular matrix $L$ and an upper triangular matrix $U$ of the same size:
$$ A = LU.$$
This is called the **LU decomposition** of square matrices. Once we have an LU decomposition of the matrix $A$, we can solve the system $A\vec x = \vec b$ by first running forward-substitution and then back-substitution. More specifically, observe that the given system of equations can be written as
$$ A\vec x = LU\vec x = L\vec y = \vec b$$
where $\vec y = U\vec x$. We first solve the system
$$ L\vec y = \vec b$$
using forward-substitution. Then we solve the system
$$ U\vec x = \vec y$$
using back-substitution to get the final solution vector $\vec x$.

### Examples

Let's look at two examples.

1. Suppose we are given the following system:
    $$ \begin{bmatrix} 1 & -2 \\ 3 & 1 \end{bmatrix} \begin{bmatrix} x \\ y \end{bmatrix} = \begin{bmatrix} 3 \\ -2 \end{bmatrix}.$$
    The LU decomposition of the coefficient matrix $A$ is given as follows:
    $$ \begin{bmatrix} 1 & -2 \\ 3 & 1 \end{bmatrix} = \begin{bmatrix} 1 & 0 \\ 3 & 1 \end{bmatrix} \begin{bmatrix} 1 & -2 \\ 0 & 7 \end{bmatrix}.$$
    Then the system above can be rewritten as
    $$ \begin{bmatrix} 1 & 0 \\ 3 & 1 \end{bmatrix} \begin{bmatrix} 1 & -2 \\ 0 & 7 \end{bmatrix}\begin{bmatrix} x \\ y \end{bmatrix}
    = \begin{bmatrix} 1 & 0 \\ 3 & 1 \end{bmatrix} \begin{bmatrix} z \\ w \end{bmatrix} = \begin{bmatrix} 3 \\ -2 \end{bmatrix} \tag{$\ast$}$$
    where
    $$\begin{bmatrix} z \\ w \end{bmatrix} = \begin{bmatrix} 1 & -2 \\ 0 & 7 \end{bmatrix}\begin{bmatrix} x \\ y \end{bmatrix}. \tag{$\dagger$} $$
    Solving the equation ($*$) with forward-substitution, we obtain
    $$ z = 3,\quad w = -11.$$
    Then we solve the equation ($\dagger$) with back-substitution, which gives us
    $$ 7y = -11,\ y = -\frac{11}{7};\quad x - 2\left(-\frac{11}{7}\right) = 3,\ x = \frac{1}{7}.$$
    Thus the solution to the original system is $x = \frac{1}{7}$ and $y = -\frac{11}{7}$.

1. Suppose we are given the system
    $$ \begin{bmatrix} 1 & 2 \\ 2 & 4\end{bmatrix} \begin{bmatrix} x \\ y \end{bmatrix} = \begin{bmatrix} 0 \\ 1 \end{bmatrix}.$$
    The LU decomposition of the coefficient matrix $A$ is given as follows:
    $$ \begin{bmatrix} 1 & 2 \\ 2 & 4\end{bmatrix} = \begin{bmatrix} 1 & 0 \\ 2 & 1 \end{bmatrix} \begin{bmatrix} 1 & 2 \\ 0 & 0 \end{bmatrix}.$$
    Then the system above can be rewritten as
    $$ \begin{bmatrix} 1 & 0 \\ 2 & 1 \end{bmatrix} \begin{bmatrix} 1 & 2 \\ 0 & 0 \end{bmatrix}\begin{bmatrix} x \\ y \end{bmatrix}
    = \begin{bmatrix} 1 & 0 \\ 2 & 1 \end{bmatrix} \begin{bmatrix} z \\ w \end{bmatrix} = \begin{bmatrix} 0 \\ 1 \end{bmatrix} \tag{$\ast$}$$
    where
    $$\begin{bmatrix} z \\ w \end{bmatrix} = \begin{bmatrix} 1 & 2 \\ 0 & 0 \end{bmatrix}\begin{bmatrix} x \\ y \end{bmatrix}. \tag{$\dagger$} $$
    Solving the equation ($*$) with forward-substitution, we obtain
    $$ z = 0,\quad w = 1.$$
    Then solving the equation ($\dagger$) with back-substitution, we see that the last equation is $0=1$, which is always false. This means that the system has no solution.
