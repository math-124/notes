# 9.1: Solution spaces of systems of linear equations

Recall that systems of linear equations can be represented in matrix form as
$$ A \vec x = \vec b.$$
For instance, the systems
$$ \begin{cases} 2x + y = 1 \\ x + y = 2 \end{cases};\quad \begin{cases}x+y-z &= 3 \\ 2x +z &= -3 \\ -2y + 3z &= -9 \end{cases} $$
can each be written in matrix form as
$$ \begin{bmatrix} 2 & 1 \\ 1 & 1 \end{bmatrix} \begin{bmatrix} x \\ y \end{bmatrix} = \begin{bmatrix} 1 \\ 2 \end{bmatrix};\quad
\begin{bmatrix} 1 & 1 & -1 \\ 2 & 0 & 1 \\ 0 & -2 & 3 \end{bmatrix} \begin{bmatrix} x \\ y \\ z \end{bmatrix} = \begin{bmatrix} 3 \\ -3 \\ -9 \end{bmatrix} $$
Note that the coefficients of the variables in the system are gathered together to the matrix. For this reason, we call matrix $A$ the **coefficient matrix** of the system.

The objective of this section is to understand the solution space of systems of linear equations in terms of the coefficient matrix $A$ and the vector $\vec b$. It will turn out that there is always either no solution, a unique solution, or infinitely many solutions to the system, and this can be determined by the properties of the matrix $A$ and the vector $\vec b$.

## Homogeneous systems

We will first try to understand the case when vector $\vec b$ is zero. This is called a **homogeneous system** of linear equations, and can evidently be written as $$ A \vec x = 0.$$

The first observation is that for any homogeneous system, the zero vector $0$ is always a solution since $A\cdot 0 =  0$. We wish to understand whether there are any other solutions to the system, i.e. whether there are any _non-zero_ vectors $\vec x$ such that $A \vec x = 0$.

Let us assume that $A$ is an $m\times n$ matrix and that $\vec x$ is a vector with $n$ entries. Recall the following interpretation of the matrix-vector product $A \vec x$:
$$ A \vec x = x_1 \vec r_1 + x_2 \vec r_2 + \cdots + x_n \vec r_n $$
where $\vec r_1, \vec r_2, \ldots, \vec r_n$ are the column vectors of the matrix $A$, i.e.
$$ A = \begin{bmatrix} \vec r_1 & \vec r_2 & \cdots & \vec r_n \end{bmatrix}.$$

In other words, the product $A\vec x$ is a linear combination of the columns of $A$, with the entries of $\vec x$ as coefficients. Hence the equation $A\vec x = 0$ can be interpreted as asking for a linear combination of the columns of $A$ that equals the zero vector.

Suppose that there is a nonzero solution $\vec x$. For convenience, we assume the first entry $x_1$ is nonzero. Then we can write the equation $A\vec x = 0$ as
$$ x_1 \vec r_1 + x_2 \vec r_2 + \cdots + x_n \vec r_n = 0,$$
which is rearranged as
$$ \vec r_1 = -\frac{x_2}{x_1} \vec r_2 - \frac{x_3}{x_1} \vec r_3 - \cdots - \frac{x_n}{x_1} \vec r_n.$$
This shows that the first column vector $\vec r_1$ is a linear combination of the other column vectors $\vec r_2, \ldots, \vec r_n$. In other words, the column vectors of $A$ are linearly dependent.

Conversely, if the column vectors of $A$ are linearly dependent, we can run the process above backwards to find a nonzero solution $\vec x$ to the equation $A\vec x = 0$. Hence we have shown that there is a nonzero solution to the homogeneous system $A\vec x = 0$ if and only if the column vectors of $A$ are linearly dependent.

Let us look at the homogeneous systems given by the coefficient matrices of two examples given above.
$$ A_1 = \begin{bmatrix} 2 & 1 \\ 1 & 1 \end{bmatrix} \quad \text{and} \quad A_2 = \begin{bmatrix} 1 & 1 & -1 \\ 2 & 0 & 1 \\ 0 & -2 & 3 \end{bmatrix}.$$
Since the column vectors
$$ \begin{bmatrix} 2 \\ 1 \end{bmatrix} \quad \text{and} \quad \begin{bmatrix} 1 \\ 1 \end{bmatrix} $$
are linearly independent, the homogeneous system $A_1 \vec x = 0$ has only the zero solution. On the other hand, the column vectors
$$ \vec r_1 = \begin{bmatrix} 1 \\ 2 \\ 0 \end{bmatrix}, \quad \vec r_2 = \begin{bmatrix} 1 \\ 0 \\ -2 \end{bmatrix}, \quad \text{and} \quad \vec r_3 = \begin{bmatrix} -1 \\ 1 \\ 3 \end{bmatrix} $$
are linearly dependent, since the first column vector can be expressed as a linear combination of the other two:
$$ \vec r_1 = 3\vec r_2 + 2\vec r_3.$$
Rearranging, we have
$$ \vec r_1 - 3\vec r_2 - 2\vec r_3 = 0,$$
which shows that the vector $\vec x = \begin{bmatrix} 1 \\ -3 \\ -2 \end{bmatrix}$ is a nonzero solution to the homogeneous system $A_2 \vec x = 0$.

### Properties of solutions

What can we say about the solutions of the homogeneous system $A\vec x = 0$? Given that there is a nonzero solution, can we find infinitely many solutions, or maybe just two or three?

We can show that if there is a nonzero solution $\vec x$, then there are infinitely many solutions by analyzing the properties of the solutions. Suppose $\vec x$ and $\vec y$ are two solutions to the homogeneous system above (that are not necessarily distinct). Then we have
$$ A \vec x = 0 \quad \text{and} \quad A \vec y = 0.$$
We can then consider the sum of these two solutions:
$$ A(\vec x + \vec y) = A\vec x + A\vec y = 0 + 0 = 0.$$
This shows that the sum $\vec x + \vec y$ is also a solution to the homogeneous system. Similarly, for any scalar $c$, we have
$$ A(c\vec x) = c(A\vec x) = c \cdot 0 = 0,$$
which shows that the scalar multiple $c\vec x$ is also a solution to the homogeneous system. Hence we have shown that if there is a nonzero solution $\vec x$, then there are infinitely many solutions, since we can take any scalar multiple of $\vec x$ or take linear combinations with any other solutions $\vec y$ to get another solution.

For instance, recall that the homogeneous system $A_2 \vec x = 0$ has a solution $\vec x = \begin{bmatrix} 1 \\ -3 \\ -2 \end{bmatrix}$. By our observation, we know that
$$ c\vec x = c\begin{bmatrix} 1 \\ -3 \\ -2 \end{bmatrix} $$
for any scalar $c$ is a solution too.

Formally, we have shown that the set of solutions form a _vector subspace_ of $\mathbb{R}^n$. This concept will be explored in more detailed in future sections. For now, we summarize the results  as follows:
$$ A\vec x = 0 \text{ has } \begin{cases} \text{one solution $\vec x = 0$ if the column vectors of $A$ are linearly independent} \\ \text{$\infty$ many solutions if the column vectors of $A$  are linearly dependent}\end{cases}$$
