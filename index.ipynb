# Vectors, Matrices, and Applications: Course Notes for Math 124 at the University of Michigan

These are the course notes for Math 124 at the University of Michigan.

:::{note} Google Chrome Recommended
For the best experience, please view these notes using **Google Chrome** (or another Chromium-based browser, like Arc). You may experience slower scrolling and performance issues if using Safari.
:::

:::{warning} Content is under development!
The course is **brand-new** in Fall 2026, and as such, the course notes don't all exist yet! There will eventually be 9 chapters:
1. Vectors
2. Lines, Planes, and Orthogonality
3. Matrices and Linear Transformations
4. Systems of linear equations
5. Subspaces
6. Orthonormal Bases
7. Least Squares
8. Determinants
9. Eigenvalues and Eigenvectors

As sections are ready (which will usually be on the day of the corresponding lecture), we will publish them here and link them from the [course website](https://math124.org).
:::
