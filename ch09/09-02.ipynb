# 9.2: Diagonal Matrices and Diagonalization

The reason why eigenvectors and eigenvalues are important is that they provide a coordinate system in which complicated matrices can be describe in a much simpler way. This lecture will be dedicated to unpacking what this sentence means.

## Diagonal matrices

Let us first understand what we mean by a 'simpler' way. Under many circumstances, eigenvectors will allow us to 'describe' the given matrix in a diagonal matrix. Why are diagonal matrices nice?

Suppose we are given the diagonal matrix
$$ D = \begin{bmatrix} 1 & 0 \\ 0 & 3 \end{bmatrix}.$$
Let us interpret this as a linear transformation. It maps $\vec e_1$ to $\vec e_1$, and $\vec e_2$ to $3\vec e_2$. In other words, the transformation simply scales the standard basis vectors by the corresponding diagonal entry. More generally, given a diagonal matrix
$$ D = \begin {bmatrix} \lambda_1 & & & \\  &\lambda_2 & & \\  & & \ddots & \\ & & & \lambda_n\end{bmatrix},$$
multiplying $D$ with $\vec e_i$ gives us $\lambda_i\vec e_i$.

As we can see, diagonal matrix have a straightforward interpretation as linear transformations: they simply scale the standard basis vectors. How about arbitrary vectors and not the standard basis vectors? Taking the first $2\times 2$ example, we have for $ \vec v = \begin{bmatrix} 3 \\ 2\end{bmatrix}$ that
$$ D\vec v = \begin{bmatrix} 1 & 0 \\ 0 & 3 \end{bmatrix} \begin{bmatrix} 3 \\ 2\end{bmatrix} = \begin{bmatrix} 3 \\ 6 \end{bmatrix}.$$
Note that when we multiply the vector $\vec v$ to matrix $D$, we are simply each diagonal entry to the each entry of the vector. More generally, we have
$$ \begin {bmatrix} \lambda_1 & & & \\  &\lambda_2 & & \\  & & \ddots & \\ & & & \lambda_n\end{bmatrix}\begin{bmatrix} v_1 \\ v_2 \\ \vdots \\ v_n \end{bmatrix} = \begin{bmatrix} \lambda_1 v_1 \\ \lambda_2 v_2 \\ \vdots \\ \lambda_n v_n \end{bmatrix}.$$

The moral is that diagonal matrices are very easy to understand as linear transformations: they scale each entry of the vector by the diagonal entries.

Another reason why diagonal matrices are easy is that they behave well under matrix-matrix multiplication. In general, products of diagonal matrices are calculated by simply multiplying each diagonal entry by diagonal entry:
$$ \begin {bmatrix} \lambda_1 & & & \\  &\lambda_2 & & \\  & & \ddots & \\ & & & \lambda_n\end{bmatrix}\begin {bmatrix} \mu_1 & & & \\  &\mu_2 & & \\  & & \ddots & \\ & & & \mu_n\end{bmatrix} = \begin {bmatrix} \lambda_1 \mu_1 & & & \\  &\lambda_2 \mu_2 & & \\  & & \ddots & \\ & & & \lambda_n \mu_n\end{bmatrix}.  \tag{$*$}$$
This property is especially useful when we consider multiplying a diagonal matrix to itself many times over, i.e. taking _powers_ of matrices. The $k$-th power of a matrix $A$, denoted $A^k$, is defined in the same way it is for numbers: it is the product of $A$ by itself $k$ times. For diagonal matrices, the $k$-th power is obtained by simply taking the $k$-th power of each of the diagonal entries:
$$ D^k = \begin {bmatrix} \lambda_1^k & & & \\  &\lambda_2^k & & \\  & & \ddots & \\ & & & \lambda_n^k\end{bmatrix}. $$
This can be seen by repeatedly applying ($*$).

One can interpret the $k$-th power of a matrix in terms of linear transformations. Recall that given matrices $A$ and $B$, the product $AB$ can be interpreted as 'applying the transformation $B$ and then $A$'. Hence the power $A^k$ can be understood as 'applying the transformation $A$ by $k$ times'. In other words, the matrix $A^k$ corresponds to the $k$ _times iteration_ of the linear transformation $A$. This is important to note since iterations of linear transformations appear all over mathematics, physics, and engineering. We will see one particularly enlightening example -- the _Markov chain_ -- in the last lecture.

## Diagonalization

Now let us explore what we mean by 'describing' $A$ in a simpler form. We will focus on the following toy example: 
$$ A = \begin{bmatrix} 2 & 1 \\ 1 & 2 \end{bmatrix}. $$
Recall from last class that this matrix has eigenvalues 1 and 3, each with corresponding eigenvectors
$$ \vec v_1 = \begin{bmatrix} 1 \\ -1 \end{bmatrix},\quad \vec v_2 = \begin{bmatrix} 1 \\ 1 \end{bmatrix}.$$
This means that $A\vec v_1 = \vec v_1$ and $A\vec v_2 = 3\vec v_2$.

Now suppose that we are given the vector 
$$ \vec w = \begin{bmatrix} 5 \\ -1\end{bmatrix}$$
and want to evaluate $A\vec w$. Instead of directly doing matrix multiplication, what we could do is first decompose $\vec w$ in terms of $\vec v_1$ and $\vec v_2$, which gives us
$$ \vec w = 3 \vec v_1  +2\vec v_2.$$
This makes it much more convenient to find the product $A\vec w$, since we simply have
$$ A \vec w = A(3 \vec v_1  +2\vec v_2) = 3A\vec v_1 + 2 A\vec v_2 = 3\vec v_1 +2(3\vec v_2)= 3\vec v_1 +6\vec v_2 = \begin{bmatrix} 9 \\ 3 \end{bmatrix}.$$
In other words, once we decompose the vector $\vec w$ as a combination of $\vec v_1$ and $\vec v_2$, multiplying $A$ to each of them just amounts to multiplying each of their eigenvectors.

The heart of diagonalization is that the process above can be formalized so that it works for arbitary vectors $\vec w$, and that formalization results in a matrix decomposition of $A$. Let us define $S$ to have column vectors be the eigenvectors of $A$:
$$ S = \begin{bmatrix} \vec v_1 & \vec v_2 \end{bmatrix} = \begin{bmatrix} 1 & 1 \\ -1 & 1 \end{bmatrix}.$$

We can interpret this as a linear transformation sending the standard basis vectors $\vec e_1$, $\vec e_2$ each to $\vec v_1$, $\vec v_2$ respectively. Note that the inverse matrix
$$ S^{-1} = \frac{1}{\det S} \begin{bmatrix} 1 & -1 \\ 1 & 1 \end{bmatrix} = \frac{1}{2}\begin{bmatrix} 1 & -1 \\ 1 & 1 \end{bmatrix}$$
can then can be interpreted as a linear transformation mapping $\vec v_1$, $\vec v_2$ to $\vec e_1$, $\vec e_2$ respectively.

The process we went through in the example above can be more generally summarized as
* Take a vector $\vec w$ and express it as the linear combination $\vec w = a \vec v_1 + b\vec v_2$;
* applying $A$ gives us $A\vec w = a\vec v_1 + 3b\vec v_2$.

Now suppose we are given the breakdown $\vec w = a \vec v_1 + b\vec v_2$ for us, where $a$ and $b$ are yet unknown. Observe that applying $S^{-1}$, we obtain
$$ S^{-1}\vec w = a (S^{-1}\vec v_1) + b (S^{-1}\vec v_2) = a \vec e_1 + b\vec e_2 = \begin{bmatrix} a \\ b\end{bmatrix}.$$
In other words, multiplying the matrix $S^{-1}$ to $\vec w$ reveals what the scalars $a$ and $b$ are.

Now let us take the diagonal matrix
$$ D = \begin{bmatrix} 1 & 0 \\ 0 & 3 \end{bmatrix}.$$
Recall that multiplying a vector to this matrix is simply multiplying each of the diagonal entries:
$$ D (S^{-1}\vec w) = \begin{bmatrix} 1 & 0 \\ 0 & 3 \end{bmatrix}\begin{bmatrix} a \\ b\end{bmatrix} = \begin{bmatrix} a \\ 3b\end{bmatrix} = a\vec e_1 + 3b\vec e_2.$$
Finally multiplying with the matrix $S$, we obtain
$$ S(DS^{-1}\vec w) = a(S\vec e_1) +3b(S\vec e_2) = a\vec v_1 + b \vec v_2,$$
which is exactly what $A\vec w$ should be, i.e.
$$ A\vec w = SDS^{-1}\vec w.$$
Since this is true for any vector $\vec w$, we see that 
$$ A = SDS^{-1}. \tag{$\dagger$}$$
(In particular, one can see that the two matrices have the same first column vector by putting $\vec w = \vec e_1$, and same for the second column.)

The expression in ($\dagger$) is called the **diagonalization** of matrix $A$. Writing out in entries, the diagonalization is
$$ \begin{bmatrix} 2 & 1 \\ 1 & 2 \end{bmatrix} = \begin{bmatrix} 1 & 1 \\ -1 & 1 \end{bmatrix}\begin{bmatrix} 1 & 0 \\ 0 & 3 \end{bmatrix}\begin{bmatrix} 1 & 1 \\ -1 & 1 \end{bmatrix}^{-1} = \begin{bmatrix} 1 & 1 \\ -1 & 1 \end{bmatrix}\begin{bmatrix} 1 & 0 \\ 0 & 3 \end{bmatrix}\begin{bmatrix} 1/2 & -1/2 \\ 1/2 & 1/2 \end{bmatrix}.$$

## Meaning of diagonalization: Change of coordinates

There is much more meaning to diagonalization than what seems to be another one of those matrix decompositions that makes calculation easier. Let's break down the steps we've took. We started out by multiplying the matrix $S^{-1}$ to the given vector $\vec w$, which gave us
$$ S^{-1}\vec w = \begin{bmatrix} a \\ b\end{bmatrix},$$
where $a$ and $b$ are the coefficients of the decomposition $\vec w = a\vec v_1 + b\vec v_2$. Here we can understand the role of matrix $S^{-1}$ as _changing coordinates_ from $\vec e_1, \vec e_2$ to $\vec v_1$, $\vec v_2$. Let us try to see this from our previous example of
$$\vec w = \begin{bmatrix} 5 \\ -1\end{bmatrix} = 5\vec e_1 - \vec e_2.$$
By calculating, we see that
$$ S^{-1}\vec w = \begin{bmatrix} 1/2 & -1/2 \\ 1/2 & 1/2 \end{bmatrix}\begin{bmatrix} 5 \\ -1\end{bmatrix} = \begin{bmatrix} 3 \\ 2\end{bmatrix},$$
which matches our previous observation of $\vec w = 3\vec v_1 + 2\vec v_2$.

The next part is the most important: the diagonal matrix
$$ D = \begin{bmatrix} 1 & 0 \\ 0 & 3 \end{bmatrix}$$
can be understood as the expression of matrix $A$ in the coordinate system defined by $\vec v_1, \vec v_2$. Indeed, $A$ acts by scaling its eigenvectors $\vec v_1$ and $\vec v_2$ by its corresponding eigenvalues 1 and 3, respectively. Recall that that is precisely how the diagonal matrix works. Hence we get
$$ D\begin{bmatrix} 3 \\ 2\end{bmatrix} = \begin{bmatrix} 3 \\ 6\end{bmatrix},$$
which matches with
$$ A\vec w = 3\vec v_1 + 6\vec v_2.$$
This is what we mean by the matrix $A$ being 'described' in a simple way: in the new coordinate system defined by its eigenvectors, $A$ is written as the diagonal matrix $D$.

The last part is to convert back from the coordinate system $\vec v_1, \vec v_2$ to $\vec e_1, \vec e_2$ by multiplying $S$. Indeed, note that given
$$\vec w = 3\vec v_1 + 2\vec v_2,\quad A\vec w = 3\vec v_1 + 6\vec v_2,$$
multiplying $S$ gives us
$$ S\begin{bmatrix} 3 \\ 6 \end{bmatrix} = \begin{bmatrix} 1 & 1 \\ -1 & 1 \end{bmatrix}\begin{bmatrix} 3 \\ 6 \end{bmatrix} = \begin{bmatrix} 9 \\ 3 \end{bmatrix},$$
which precisely matches our final result obtained earlier.


Let us summarize our findings with the following chain of equations:

$$
\begin{aligned}
\vec w = \begin{bmatrix} 5 \\ -1 \end{bmatrix}
&\qquad && \text{coordinates in }(\vec e_1,\vec e_2), \\[4pt]
S^{-1}\vec w = \begin{bmatrix} 3 \\ 2 \end{bmatrix}
&\qquad && \text{change coordinates: }(\vec e_1,\vec e_2)\to(\vec v_1,\vec v_2), \\[4pt]
DS^{-1}\vec w = \begin{bmatrix} 3 \\ 6 \end{bmatrix}
&\qquad && \text{apply }A\text{ in eigenvector coordinates}, \\[4pt]
SDS^{-1}\vec w = \begin{bmatrix} 9 \\ 3 \end{bmatrix} = A\vec w
&\qquad && \text{change coordinates: }(\vec v_1,\vec v_2)\to(\vec e_1,\vec e_2).
\end{aligned}
$$

This perspective of viewing diagonalization as a coordinate change is quite useful for conceptual purposes. For example, the $k$-th power of $A$ can simply be understood as
* changing coordinates to $\vec v_1,\vec v_2$ by multiplying $S^{-1}$;
* taking the $k$-th power of $D$; and
* changing back coordinates by multiplying $S$.

This is mathematically expressed as
$$ A^k = SD^{k}S^{-1}.$$
One can check the validity of this expression by simply multiplying $A$ by $k$ times:
$$ A^k = (SDS^{-1})(SDS^{-1})\cdots (SDS^{-1}) = SD^{k}S^{-1},$$
where we note that the adjacent $S^{-1}$ and $S$ cancel out when multiplied, leaving $k$ copies of $D$ multiplied together to form $D^{k}$. This provides another reason why diagonalization is so powerful: it makes it much easier to calculate powers of matrices, since $D^k$ above can be found simply by taking the $k$-th powers of each of the diagonal entries.

Going back to our toy example, without diagonalization finding an expression for the $k$-th power of
$$ A = \begin{bmatrix} 2  & 1 \\ 1 & 2 \end{bmatrix}$$
seems very hard. But once we diagonalize, we can easily calculate it as follows:
$$\begin{align*} A^k = SD^{k}S^{-1} &= \begin{bmatrix} 1  & 1 \\ -1 & 1 \end{bmatrix}\begin{bmatrix} 1  & 0 \\ 0 & 3 \end{bmatrix}^k\begin{bmatrix} 1/2  & -1/2 \\ 1/2 & 1/2 \end{bmatrix} = \begin{bmatrix} 1  & 1 \\ -1 & 1 \end{bmatrix}\begin{bmatrix} 1  & 0 \\ 0 & 3^k \end{bmatrix}\begin{bmatrix} 1/2  & -1/2 \\ 1/2 & 1/2 \end{bmatrix} \\ &= \begin{bmatrix} 1  & 3^k \\ -1 & 3^k \end{bmatrix}\begin{bmatrix} 1/2  & -1/2 \\ 1/2 & 1/2 \end{bmatrix} \\ &= \begin{bmatrix} (1+3^k)/2  & (-1+3^k)/2 \\ (-1+3^k)/2 & (1+3^k)/2 \end{bmatrix}.\end{align*}$$


## Diagonalization in general

Now let us look back at our toy example and attempt to generalize the concept of diagonalization to any $n\times n$ matrix $A$. The very first step in doing this will hint to us that diagonalization may not be possible for all matrices. Indeed, we will later see examples of matrices that cannot be diagonalized.

The first crucial step in our toy example was expressing an arbitrary vector $\vec w$ as a linear combination of the eigenvectors $\vec v_1$, $\vec v_2$ of $A$. This is equivalent to the eigenvectors of $A$ forming a basis of $\mathbb{R}^2$. In general, **for $A$ to be diagonalizable, it must have eigenvectors $\vec v_1, \vec v_2,\ldots, \vec v_n$ that form a basis of $\mathbb{R}^n$**; this may not always be the possible.

Suppose we do have a basis of eigenvectors given as above. Then diagonalization works in the same way it did for the toy example. We form the **change of basis matrix** $S$ by taking the eigenvectors to be its column vectors:
$$ S = \begin{bmatrix} \vec v_1 & \vec v_2 & \cdots & \vec v_n \end{bmatrix}.$$
Then we take the diagonal matrix $D$ with the eigenvalues of $A$:
$$ D = \begin {bmatrix} \lambda_1 & & & \\  &\lambda_2 & & \\  & & \ddots & \\ & & & \lambda_n\end{bmatrix},$$
where $\lambda_i$ is the eigenvalue of $\vec v_i$. We obtain the diagonalization of $A$ as
$$ A = SDS^{-1}.$$
Note that $\vec v_1, \vec v_2,\ldots, \vec v_n$ must necessarily form a basis for the matrix $S$ to be invertible.


### Non-diagonalizable matrices

There are indeed matrices that are not diagonalizable. The simplest example is the shear matrix:
$$ A = \begin{bmatrix} 1 & a \\ 0 & 1 \end{bmatrix}\qquad (a\neq 0).$$
Recall that its only eigenvector is $\begin{bmatrix} 1 \\ 0 \end{bmatrix}$ with eigenvalue 1. This is not enough to form a basis of $\mathbb{R}^2$.
