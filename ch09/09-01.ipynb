# 9.1: Eigenvalues and Eigenvectors

For the final chapter of this course, we turn to the subject of eigenvectors and eigenvalues, which are considered one of the most important concepts in linear algebra. They allow us to understand complicated looking matrices (and the associated linear transformations) as scaling maps in a different coordinate system.

## Definitions

Suppose we are given an $n\times n$ square matrix $A$. We say that a nonzero vector $\vec v$ with $n$ entries is an **eigenvector** of $A$ if
$$ A\vec v = \lambda \vec v $$
for some scalar $\lambda$. We say that $\lambda$ is the **eigenvalue** associated to the eigenvector $\vec v$. The eigenvalues of matrix $A$ are simply all the possible eigenvalues for different eigenvectors.

Before looking at a series of numerical examples, it helps to meditate on what eigenvectors/eigenvalues mean geometrically. Viewing the matrix $A$ as a linear transformation, $A\vec v$ is the output we get from inputting the vector $\vec v$. If we have that $\vec v$ is an eigenvector with eigenvalue $\lambda$, i.e.
$$ A\vec v = \lambda \vec v, $$
then $A\vec v$ does not change the direction of $\vec v$ and simply scales it by a factor of $\lambda$, its eigenvalue. In other words, eigenvectors represent all directions which are _preserved_ by $A$, and the associated eigenvalue is the factor in which that direction is scaled.

Note that it is valid to say that eigenvectors represent 'direction' and not just 'vector', since any scalar multiple of an eigenvector is also an eigenvector. Indeed, note for any scalar $c$ that
$$ A(c\vec v) = cA \vec v = c(\lambda \vec v) = \lambda (c\vec v),$$
so $c\vec v$ is also an eigenvector of $A$ with eigenvalue $\lambda$.

Now let us look at some simple numerical examples to get ourselves warmed up.

#### Example 1: A $2\times 2$ diagonal matrix

Let
$$A=\begin{bmatrix}2&0\\0&-1\end{bmatrix}. $$
The standard basis vectors are eigenvectors of $A$. More precisely,
$$\vec e_1=\begin{bmatrix}1\\0\end{bmatrix} \quad\text{has eigenvalue}\quad \lambda_1=2.$$
and
$$\vec e_2=\begin{bmatrix}0\\1\end{bmatrix} \quad\text{has eigenvalue}\quad \lambda_2=-1.$$
We can verify these eigenpairs directly:
$$A\vec e_1=\begin{bmatrix}2&0\\0&-1\end{bmatrix}\begin{bmatrix}1\\0\end{bmatrix}=\begin{bmatrix}2\\0\end{bmatrix}=2\begin{bmatrix}1\\0\end{bmatrix}=2\vec e_1.$$
and
$$A\vec e_2=\begin{bmatrix}2&0\\0&-1\end{bmatrix}\begin{bmatrix}0\\1\end{bmatrix}=\begin{bmatrix}0\\-1\end{bmatrix}=(-1)\begin{bmatrix}0\\1\end{bmatrix}=-\vec e_2.$$
Thus, multiplying by $A$ scales $\vec e_1$ by $2$ and reverses $\vec e_2$ without changing its length. The eigenvalues of $A$ are 2 and $-1$.

#### Example 2: A $3\times 3$ upper-triangular matrix

Let
$$B=\begin{bmatrix}1&1&0\\0&2&0\\0&0&3\end{bmatrix}. $$
Three eigenvectors and their associated eigenvalues are
$$\vec w_1=\begin{bmatrix}1\\0\\0\end{bmatrix}, \quad \lambda_1=1; \qquad \vec w_2=\begin{bmatrix}1\\1\\0\end{bmatrix}, \quad \lambda_2=2; \qquad \vec w_3=\begin{bmatrix}0\\0\\1\end{bmatrix}, \quad \lambda_3=3.$$
Again, we verify each pair by multiplying:
$$B\vec w_1=\begin{bmatrix}1&1&0\\0&2&0\\0&0&3\end{bmatrix}\begin{bmatrix}1\\0\\0\end{bmatrix}=\begin{bmatrix}1\\0\\0\end{bmatrix}=1\vec w_1.$$
$$B\vec w_2=\begin{bmatrix}1&1&0\\0&2&0\\0&0&3\end{bmatrix}\begin{bmatrix}1\\1\\0\end{bmatrix}=\begin{bmatrix}2\\2\\0\end{bmatrix}=2\begin{bmatrix}1\\1\\0\end{bmatrix}=2\vec w_2.$$
and
$$B\vec w_3=\begin{bmatrix}1&1&0\\0&2&0\\0&0&3\end{bmatrix}\begin{bmatrix}0\\0\\1\end{bmatrix}=\begin{bmatrix}0\\0\\3\end{bmatrix}=3\begin{bmatrix}0\\0\\1\end{bmatrix}=3\vec w_3.$$
Notice that $\vec w_2$ is not a standard basis vector, but multiplication by $B$ still only scales it and does not change its direction. The eigenvalues of $B$ are 1, 2, and 3.

### Eigenvectors and eigenvalues of special $2\times 2$ matrices

Let's analyze what eigenvalues and eigenvectors we get for more general classes of matrices in the $2\times 2$ case. Recall that we've studied the following maps: scalings, reflections, projections, and shears.

### Scalings

A matrix representing a scaling transformation is given by a diagonal matrix:
$$ A = \begin{bmatrix} a & 0 \\ 0  & b \end{bmatrix}.$$
As we've saw in the first example above, the eigenvectors of this matrix are the standard basis vectors $\vec e_1$ and $\vec e_2$, each with eigenvalues $a$ and $b$.

### Reflections

Rather than looking at general reflections over lines, let's focus on the simple example of reflecting across the line $y=x$. In this case, the vector $ \begin{bmatrix} x \\ y \end{bmatrix}$ is mapped to $ \begin{bmatrix} y \\ x \end{bmatrix}$. Hence the matrix can be written as
$$ R= \begin{bmatrix} 0 & 1 \\ 1 & 0 \end{bmatrix}.$$
Vectors along the line $y=x$ are unchanged by the reflection. For example,
$$R\begin{bmatrix}1\\1\end{bmatrix}=\begin{bmatrix}0&1\\1&0\end{bmatrix}\begin{bmatrix}1\\1\end{bmatrix}=\begin{bmatrix}1\\1\end{bmatrix}.$$
Thus, $\begin{bmatrix}1\\1\end{bmatrix}$ is an eigenvector with eigenvalue $1$. On the other hand, vectors perpendicular to the line of reflection, which lie along $y=-x$, have their directions reversed. For example,
$$R\begin{bmatrix}1\\-1\end{bmatrix}=\begin{bmatrix}0&1\\1&0\end{bmatrix}\begin{bmatrix}1\\-1\end{bmatrix}=\begin{bmatrix}-1\\1\end{bmatrix}=-\begin{bmatrix}1\\-1\end{bmatrix}.$$
Therefore, $\begin{bmatrix}1\\-1\end{bmatrix}$ is an eigenvector with eigenvalue $-1$, as is every nonzero vector along $y=-x$. This matches our geometric intuition: the reflection acts as a scaling map on these two special directions. It scales the direction parallel to the mirror by $1$ and the perpendicular direction by $-1$.

### Orthogonal projections

Consider the orthogonal projection onto the $x$-axis. It sends $\begin{bmatrix}x\\ y\end{bmatrix}$ to $\begin{bmatrix}x\\0\end{bmatrix}$, so its matrix is
$$P=\begin{bmatrix}1&0\\0&0\end{bmatrix}. $$
Every nonzero vector along the $x$-axis is left unchanged. In particular,
$$P\begin{bmatrix}1\\0\end{bmatrix}=\begin{bmatrix}1&0\\0&0\end{bmatrix}\begin{bmatrix}1\\0\end{bmatrix}=\begin{bmatrix}1\\0\end{bmatrix}.$$
so $\begin{bmatrix}1\\0\end{bmatrix}$ is an eigenvector with eigenvalue $1$.

Every nonzero vector along the $y$-axis is collapsed to the zero vector. For example,
$$P\begin{bmatrix}0\\1\end{bmatrix}=\begin{bmatrix}1&0\\0&0\end{bmatrix}\begin{bmatrix}0\\1\end{bmatrix}=\begin{bmatrix}0\\0\end{bmatrix}=0\begin{bmatrix}0\\1\end{bmatrix}.$$
so $\begin{bmatrix}0\\1\end{bmatrix}$ is an eigenvector with eigenvalue $0$. Hence the eigenvalues of this projection are $1$ and $0$: the direction onto which we project is scaled by $1$, while the perpendicular direction is scaled by $0$. Once again, the eigenvectors identify precisely the directions on which the transformation acts as a simple scaling map.

### Shears

Consider the horizontal shear
$$S=\begin{bmatrix}1&a\\0&1\end{bmatrix}, \qquad a\neq0.$$
It sends $\begin{bmatrix}x\\y\end{bmatrix}$ to $\begin{bmatrix}x+ay\\y\end{bmatrix}$, shifting a vector horizontally by an amount proportional to its vertical component. Vectors along the $x$-axis have no vertical component, so they are unchanged. In particular,
$$S\begin{bmatrix}1\\0\end{bmatrix}=\begin{bmatrix}1&a\\0&1\end{bmatrix}\begin{bmatrix}1\\0\end{bmatrix}=\begin{bmatrix}1\\0\end{bmatrix}.$$
Thus, $\begin{bmatrix}1\\0\end{bmatrix}$ is an eigenvector with eigenvalue $1$, and the same is true of every nonzero vector along the $x$-axis.

A vector not on the $x$-axis has $y\neq0$, so the shear adds the nonzero horizontal displacement $ay$ while leaving its vertical component unchanged. The output therefore cannot a scalar multiple of the input. Hence, when $a\neq0$, the $x$-axis is the only direction preserved by $A$ and hence the only eigenvalue is 1.



## Finding eigenvalues

In the examples above, we simply listed the eigenvectors and eigenvalues of a matrix. How does one actually find them, and how can we be sure that the ones we've found are everything?

This is when the concept of determinants comes in incredibly handy. Suppose that we have an eigenvector $\vec v$ with eigenvalue $\lambda$:
$$ A\vec v = \lambda \vec v.$$
Note that $\lambda \vec v$ can be rewritten as $\lambda I_n\vec v$, where we recall that $I_n$ is the $n\times n$ identity matrix. Subtracting this from each side, we obtain:
$$ (A - \lambda I_n)\vec v = 0. \tag{$*$} $$

Recall that we stipulate eigenvectors $\vec v$ to be nonzero (this is just for sanity reasons, since the zero vector is always an eigenvector with eigenvalue zero). Hence expanding the product in the equation ($*$), we see that there is some linear combination of the columns of $A-\lambda I_n$ that equals zero. This means that the matrix $A - \lambda I_n$ is not invertible, which in turn is the same as
$$ \det (A - \lambda I_n) = 0.$$

The solutions to this equation about $\lambda$ are precisely the eigenvalues of $A$.

Let us demonstrate this method with a simple $2\times 2$ matrix. Consider
$$A=\begin{bmatrix}2&1\\1&2\end{bmatrix}. $$
To find its eigenvalues, we solve $\det(A-\lambda I_2)=0$. First,
$$A-\lambda I_2=\begin{bmatrix}2&1\\1&2\end{bmatrix}-\begin{bmatrix}\lambda&0\\0&\lambda\end{bmatrix}=\begin{bmatrix}2-\lambda&1\\1&2-\lambda\end{bmatrix}.$$
Therefore,
$$\begin{align*}0 &=\det(A-\lambda I_2)\\ &=\det\begin{bmatrix}2-\lambda&1\\1&2-\lambda\end{bmatrix}\\ &=(2-\lambda)^2-1\\ &=(1-\lambda)(3-\lambda).\end{align*}$$
Hence the solutions to this equation are $\lambda=1$ and $\lambda=3$, and the eigenvalues of $A$ are $1$ and $3$.

Now that we've found the eigenvalues of $A$, how do we find the corresponding eigenvectors? The answer can be found again in equation ($*$): for each eigenvalue $\lambda$, the corresponding eigenvectors are precisely the nonzero solutions to the homogeneous equation
$$(A-\lambda I_n)\vec v = 0.$$

Let's put this observation into use on the previous example.

- For $\lambda=1$, we solve
    $$(A-I_n)\vec v=\begin{bmatrix}2-1&1\\1&2-1\end{bmatrix}\begin{bmatrix}x\\y\end{bmatrix}=\begin{bmatrix}1&1\\1&1\end{bmatrix}\begin{bmatrix}x\\y\end{bmatrix}=\begin{bmatrix}0\\0\end{bmatrix}.$$
    This gives the equation $x+y=0$, so $y=-x$. Thus, the nonzero solutions are
    $$\vec v=t\begin{bmatrix}1\\-1\end{bmatrix},\qquad t\neq 0.$$
    Hence the vector $\begin{bmatrix}1\\-1\end{bmatrix}$ (and every nonzero multiple of it) is an eigenvector with eigenvalue $1$.

- For $\lambda=3$, we solve
    $$(A-3I_n)\vec v=\begin{bmatrix}2-3&1\\1&2-3\end{bmatrix}\begin{bmatrix}x\\y\end{bmatrix}=\begin{bmatrix}-1&1\\1&-1\end{bmatrix}\begin{bmatrix}x\\y\end{bmatrix}=\begin{bmatrix}0\\0\end{bmatrix}.$$
    This gives $-x+y=0$, so $y=x$. Hence the nonzero solutions are
    $$\vec v=t\begin{bmatrix}1\\1\end{bmatrix},\qquad t\neq0.$$
    Hence $\begin{bmatrix}1\\1\end{bmatrix}$ (and every nonzero multiple of it) is an eigenvector with eigenvalue $3$.
In summary, once an eigenvalue is known, we find its eigenvectors by finding solutions $\vec v$ to the equation $(A-\lambda I_n)\vec v = 0$ for each $\lambda$.

## Eigenvalues of triangular matrices

Just like how determinants were easy to calculate for triangular matrices, so are its eigenvalues. Given an $n\times n$ upper triangular matrix $A$, we find the eigenvalues $\lambda$ by solving
$$ \det (A - \lambda I_n) = 0.$$

Let us label the diagonal entries of $A$ as $a_1, a_2, \ldots, a_n$. Note that $A - \lambda I_n$ remains triangular with diagonal entries
$$a_1 - \lambda, a_2 - \lambda, \ldots, a_n - \lambda.$$
Hence the determinant of $A - \lambda I_n$ is simply the product of all of its diagonal entries:
$$ 0 =  \det (A - \lambda I_n) = (a_1 - \lambda)(a_2 - \lambda)\cdots(a_n - \lambda).$$
We can immediately solve the equation above and obtain the solutions as
$$ \lambda = a_1, a_2, \ldots, a_n. $$

In other words, the eigenvalues of triangular matrix $A$ are precisely its diagonal entries. We have already seen this phenomenon from various examples, including the $3\times 3$ upper triangular matrix, the $2\times 2$ scaling and shearing map.
