# 9.3: Symmetric Matrices and the Spectral Theorem

Last time, we learned about the concept of diagonalization and how useful it can be both conceptually and computationally. Once we have a diagonalization of a matrix, the corresponding linear map can be thought of as a scaling map along its eigenvectors where the scale factor are the eigenvalues. We also saw that it is immensely useful for calculating inverses and $k$-th powers as well.

One caveat is that not all matrices are diagonalizable; there needs to be enough eigenvectors to form a basis. Two example of non-diagonalizable matrices we saw was the shear and 90-degree rotation map:
$$ \begin{bmatrix} 1 & a \\ 0 & 1 \end{bmatrix},\quad \begin{bmatrix} 0 & -1 \\ 1 & 0 \end{bmatrix}.$$

Nevertheless, these exceptions are rather rare and one can say that 'most' matrices are diagonalizable. Instead of making this statement precise, we will instead focus on a certain category of matrices -- the symmetric matrices -- which are not only always diagonalizable, but in a very nice manner.

## The spectral theorem

Suppose $A$ is an $n\times n$ square matrix. Recall that $A$ is **symmetric** if it is equal to its own transpose:
$$ A^T = A.$$
This means that the entries of $A$ are symmetric with respect to the diagonal. Some examples of symmetric matrices are:
$$ \begin{bmatrix} 1 & 2 \\ 2 & 3 \end{bmatrix}, \qquad \begin{bmatrix} 4 & -1 \\ -1 & 0 \end{bmatrix}, \qquad \begin{bmatrix} 2 & 1 & 3 \\ 1 & 5 & 0 \\ 3 & 0 & -2 \end{bmatrix}. $$

The **spectral theorem** says that a symmetric matrix $A$ is always diagonalizable and has an _orthonormal_ basis of eigenvectors. This is often summarized as saying that symmetric matrices are **orthogonally diagonalizable**.

What does this mean precisely? It means that there are eigenvectors $\vec v_1,\vec v_2,\ldots, \vec v_n$ of $A$ that form an orthonormal basis of $\mathbb{R}^n$, i.e. they are of unit length and orthogonal to each other. Let $\lambda_1, \lambda_2, \ldots, \lambda_n$ be the corresponding eigenvalues, then diagonalization gives us
$$ A = QDQ^{-1},$$
where $D$ is a diagonal matrix with entries $\lambda_1, \lambda_2, \ldots, \lambda_n$ down the diagonal and $Q$ is the matrix given by
$$ Q = \begin{bmatrix} \vec v_1 & \vec v_2 & \cdots & \vec v_n \end{bmatrix}.$$

Recall here that when the column vectors of a matrix form an orthonormal basis, we call that matrix an orthogonal matrix. These matrices have the property that their inverse is equal to their transpose:
$$ Q^{T} = Q^{-1}.$$
Hence the diagonalization above can be rephrased as:
$$ A = QDQ^T.$$


## Geometric meaning of the spectral theorem

What does the orthogonal diagonalization $A = QDQ^T$ provided by the symmetric theorem mean geometrically? According to our understanding of diagonlization, this means that the map $A$ is a linear transformation that acts by scaling along the directions $\vec v_1, \vec v_2, \ldots, \vec v_n$, and the scaling factors are the corresponding eigenvalues. An equivalent way of describing this was to view the diagonalization as a coordinate change: we change coordinates from the standard basis $\vec e_1, \vec e_2, \ldots, \vec e_n$ to $\vec v_1, \vec v_2, \ldots, \vec v_n$ and then perform a scaling along each of the coordinates.

Again, a particularly nice feature here is that the eigenvectors $\vec v_1, \vec v_2, \ldots, \vec v_n$ -- which form the basis of the new coordinate system -- form an orthonormal basis. Sometimes this collection of eigenvectors may be called the **orthonormal eigenbasis**.

One may also interpret the diagonalization $A = QDQ^T$ as a composition of linear transformations. Recall that the orthogonal matrix $Q$ corresponds to a _rigid transformation_, i.e. one that preserves lengths and angles, like rotations and reflections, and that $Q^T$ is its inverse transformation. Hence the diagonalization tells us that $A$ is a composition of:
* applying the inverse transformation $Q^T = Q^{-1}$ of a rigid transformation $Q$;
* scaling $D$ along the standard basis vectors; and
* applying the rigid transformation $Q$.

This should really come as a surprise since it's not clear at all that $A$ can be understood as a composition of such maps just from the fact that its entries are symmetric across the diagonal.

## Symmetric matrices in science/engineering

Symmetric matrices may seem aesthetically pleasing from a mathematical viewpoint, but it may seem unclear if they are natural objects that arise in science and engineering. Here we discuss some examples of symmetric matrices that appear in science/engineering, which should give extra motivation and appreciation for the spectral theorem. We won't define any of these objects in detail.

### Least squares problem and normal equation

This is the one example which we can fully explain. Recall that the exact solution of the least squares problem $A\vec x = b$ could be found by solving the **normal equation**:
$$ A^T A \vec x = A^T \vec b.$$
Here the matrix $A^T A$ is an example of a symmetric matrix arising naturally in mathematics. This follows from a simple calculation:
$$ (A^T A)^T = A^T (A^T)^T = A^T A,$$
where we recall that transposing a product of matrices swaps their order of multiplication. Hence once we obtain the diagonalization:
$$ A^T A = QDQ^T$$
from the spectral theorem, the normal equation can be rephrased as
$$ QDQ^T \vec x = A^T \vec b,$$
and the solution can be obtained by
$$ \vec x = (QDQ^T)^{-1}A^T \vec b = QD^{-1}Q^T A^T \vec b.$$
This may seem like just moving around symbols, but it's profound from a computational viewpoint since it circumvents the need to calculate the inverse of a (usually massive) matrix $A$. We still need to find the inverse of $D$, but this is much easier to find since $D^{-1}$ is simply the diagonal matrix obtained by taking the reciprocal of each of the diagonal entries of $D$.

### Covariance matrix

The main object studied in statistics are **random variables**. These are variables that may take specific sets of values with some prescribed probability, e.g. a random variable $X$ can equal zero with probability 1/2 or equal 1 with probability 1/2. Some pairs of random variables $X,Y$ may have some sort of relation between each other, and one way to capture this is by the **covariance**, denoted $\textrm{cov}(X,Y)$. Given $n$ many random variables $X_1, X_2, \ldots, X_n$, one may wish to study the **covariance matrix**
$$ C = \begin{bmatrix} \textrm{cov}(X_1, X_1) & \textrm{cov}(X_1, X_2) & \cdots & \textrm{cov}(X_1, X_n) \\ \textrm{cov}(X_2, X_1) & \textrm{cov}(X_2, X_2) & \cdots & \textrm{cov}(X_1, X_n) \\ \vdots & & \ddots & \\ \textrm{cov}(X_n, X_1) & \textrm{cov}(X_n, X_2) & \cdots & \textrm{cov}(X_n, X_n) \\
 \end{bmatrix} $$
which encodes at once how the $n$ random variables are related to each other. A basic property of covariance is that the order does not matter, i.e. $\textrm{cov}(X,Y) = \textrm{cov}(Y,X)$. It follows that the covariance matrix above is a symmetric matrix.

### Stress tensor

In mechanics, the **stress tensor** describes the internal forces acting inside a material. Its entries are usually denoted by the Greek letter $\sigma$ (sigma), and in three-dimensional space it has the form
$$ \sigma = \begin{bmatrix} \sigma_{xx} & \sigma_{xy} & \sigma_{xz} \\ \sigma_{xy} & \sigma_{yy} & \sigma_{yz} \\ \sigma_{xz} & \sigma_{yz} & \sigma_{zz} \end{bmatrix}. $$
The entries along the diagonal are called **normal stresses**, while the off-diagonal entries are called **shear stresses**. Roughly speaking, conservation of angular momentum prevents a tiny piece of material from spontaneously spinning, which forces the matching shear stresses to be equal. Thus $\sigma_{xy}=\sigma_{yx}$, $\sigma_{xz}=\sigma_{zx}$, and $\sigma_{yz}=\sigma_{zy}$, so the stress tensor is symmetric.
